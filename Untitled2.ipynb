```python
data_file = get_example_data("/gpfs/home/sa24m/scratch/tqprof/run2/osse_out/dart_cycle/output/2015071412/obs_seq.final")

obs_seq = obsq.ObsSequence(data_file)

df = obs_seq.df.copy()
# Filter your data
# df = df[df["prior_ensemble_mean"] > -1e5]
df = df[(df['type']==ObsTypeString)]

lon = ((df["longitude"] + 180.0) % 360.0) - 180.0
lat = df["latitude"]
pressure = df["vertical"]
obs = df["observation"]
# obs = df[(df['type']=='RADIOSONDE_TEMPERATURE')]["observation"]

# Axes bounds
lon_min, lon_max = -150, -50
lat_min, lat_max = 10, 70

fig = plt.figure(figsize=(12,8))
ax = fig.add_subplot(111, projection='3d')

# -------- Plot observations ----------
sc = ax.scatter(lon, lat, pressure,
                c=obs,
                s=50,
                cmap="viridis")

# -------- Ground plane (coastlines, borders, states) ----------
z0 = pressure.max()

def plot_feature_3d(ax, feature, z, lon_bounds, lat_bounds, **kwargs):
    """Convert a Cartopy feature to 3D lines at z=z0"""
    for geom in feature.geometries():
        if isinstance(geom, LineString):
            xs, ys = geom.xy
        elif isinstance(geom, MultiLineString):
            xs, ys = [], []
            for line in geom.geoms:
                x, y = line.xy
                xs.extend(x)
                xs.append(np.nan)  # break line
                ys.extend(y)
                ys.append(np.nan)
        else:
            continue

        # Clip to axis bounds
        xs = np.array(xs)
        ys = np.array(ys)
        mask = (xs >= lon_bounds[0]) & (xs <= lon_bounds[1]) & (ys >= lat_bounds[0]) & (ys <= lat_bounds[1])
        ax.plot(xs[mask], ys[mask], zs=z, **kwargs)

# Convert features to geometries
coast = cfeature.COASTLINE.with_scale("110m")
borders = cfeature.BORDERS.with_scale("110m")
states = cfeature.STATES.with_scale("50m")

plot_feature_3d(ax, coast, z0, (lon_min, lon_max), (lat_min, lat_max), color="k", lw=0.8)
plot_feature_3d(ax, borders, z0, (lon_min, lon_max), (lat_min, lat_max), color="0.2", lw=0.6)
plot_feature_3d(ax, states, z0, (lon_min, lon_max), (lat_min, lat_max), color="0.2", lw=0.6)

# -------- Axes and labels ----------
ax.set_xlim(lon_min, lon_max)
ax.set_ylim(lat_min, lat_max)
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
ax.set_zlabel("Pressure (Pa)")
ax.set_title("3D Radiosonde Profile")

# Invert pressure axis (meteorological convention)
ax.invert_zaxis()

plt.colorbar(sc, label=f"{ObsTypeString}", shrink=0.5)
# plt.savefig('figs/radiosonde_profile.png')
```

```python
def thm_at_zagl(ds, zagl_m=1000.0, time_index=0):
    # geopotential height on staggered levels -> meters
    Zstag = (ds["PH"].isel(Time=time_index).values + ds["PHB"].values) / 9.81   # (bt_stag, y, x)
    Zmass = 0.5 * (Zstag[:-1, :, :] + Zstag[1:, :, :])                         # (bt, y, x)
    HGT   = ds["HGT"].values                                                    # (y, x)
    ZAGL  = Zmass - HGT[None, :, :]                                             # (bt, y, x)

    thm3d = ds["THM"].isel(Time=time_index).values                              # (bt, y, x)

    k = np.argmin(np.abs(ZAGL - zagl_m), axis=0)                                # (y, x)
    j, i = np.indices(k.shape)
    thm2d = thm3d[k, j, i]                                                      # (y, x)
    return thm2d
axes = []

fig = plt.figure(figsize=(6*ncols, 6*nrows), constrained_layout=True)
copy = 'totalspread'
markersize = 6
dom = 'd01'
var = "T2"

# ----------------------------
# Pass 1: global range for consistent background colorbar
# ----------------------------
bg_min = np.inf
bg_max = -np.inf
for cdir in cycle_dirs:
    post_file = Path(cdir) / f"postassim_mean_{dom}.nc"
    with xr.open_dataset(post_file) as ds_post_tmp:
        arr = ds_post_tmp[var].isel(Time=0).values
        bg_min = min(bg_min, np.nanmin(arr))
        bg_max = max(bg_max, np.nanmax(arr))

levels_bg = np.linspace(bg_min, bg_max, 21)

# --- marker cycle (extend if needed) ---
markers = ['o', 's', '^', 'v', 'D', 'P', 'X', '*', '<', '>', 'h', '8']
all_types_seen = set()
obs_type_name_map = {}   # type_id -> label

# ----------------------------
# Pass 2: plotting
# ----------------------------
for i, cdir in enumerate(cycle_dirs):
    inc_file = Path(cdir) / f"analysis_increment_{dom}.nc"
    post_file = Path(cdir) / f"postassim_mean_{dom}.nc"
    obs_file = Path(cdir) / f"obs_epoch_001.nc"    
    x, y, m = setup_basemap_from_file(post_file)
    ax = fig.add_subplot(nrows, ncols, i + 1)
    axes.append(ax)

    m.drawcoastlines(ax=ax, linewidth=0.5)
    m.drawstates(ax=ax, linewidth=0.3)
    m.drawcountries(ax=ax, linewidth=0.3)

    # Background field (consistent levels across all panels)
    with xr.open_dataset(post_file) as ds_post:
        ds2d = ds_post[var].isel(Time=0) #* 1000
        cs = m.contourf(
            x, y, ds2d.values,
            levels=levels_bg,
            extend="both",
            ax=ax,
            cmap="Greens"
        )

    # Increments + winds
    with xr.open_dataset(inc_file) as ds:
        pos_inc = ds['THM'].isel(Time=0)[0].where(lambda a: a > 0) 
        neg_inc = ds['THM'].isel(Time=0)[0].where(lambda a: a < -0.00028)

        h2 = m.contour(
            x, y, pos_inc.values,
            extend="both",
            ax=ax,
            cmap="Wistia"
        )
        h3 = m.contour(
            x, y, neg_inc.values,
            extend="both",
            ax=ax,
            cmap="cool"
        )

        u = ds["U10"].isel(Time=0)
        v = ds["V10"].isel(Time=0)
        lat = ds["XLAT"]
        lon = ds["XLONG"]

        skip = 5
        q = m.quiver(
            lon.values[::skip, ::skip],
            lat.values[::skip, ::skip],
            u.values[::skip, ::skip],
            v.values[::skip, ::skip],
            scale=40,
            width=0.002,
            ax=ax,
            latlon=True,
        )


        # --- open file ---
        ds = xr.open_dataset(obs_file)
        
        # --- choose which observation copy / qc copy to plot ---
        obs_copy_idx = 0
        qc_copy_idx = 1
        
        # --- extract core fields ---
        lon = ds["location"].isel(locdim=0).values
        lat = ds["location"].isel(locdim=1).values
        z   = ds["location"].isel(locdim=2).values
        wvt = ds["which_vert"].values
        obs = ds["observations"].isel(copy=obs_copy_idx).values
        qc  = ds["qc"].isel(qc_copy=qc_copy_idx).values if "qc" in ds else None
        otype = ds["obs_type"].values  # <-- add this
        
        # Detect radians (heuristic): radians typically within [-2π, 2π]
        if np.nanmax(np.abs(lon)) <= 2*np.pi + 0.1 and np.nanmax(np.abs(lat)) <= np.pi/2 + 0.1:
            lon = np.degrees(lon)
            lat = np.degrees(lat)
        
        # Normalize lon to [-180, 180) to match typical WRF XLONG
        lon = ((lon + 180.0) % 360.0) - 180.0

        # --- near-surface mask ---
        # near_sfc = (wvt == -1) | ((wvt == 3) & np.isfinite(z) & (z <= 20.0))
        near_sfc = (
            (wvt == -1) |
            ((wvt == 3) & np.isfinite(z) & (z <= 20.0)) |
            ((wvt == 2) & np.isfinite(z) & (z >= 98000.0))   
        )
        
        maxQC = 2
        if qc is not None:
            near_sfc &= np.isfinite(qc) & (qc <= maxQC)
        
        near_sfc &= np.isfinite(lon) & np.isfinite(lat) & np.isfinite(obs)
        # --- 3 km cap in a unified height coordinate ---
        z_m = z.astype(float).copy()
        
        # convert pressure (Pa) -> approximate height (m) for which_vert==2
        p0 = 101325.0
        H  = 8400.0
        mp = (wvt == 2) & np.isfinite(z_m)
        p = np.clip(z_m[mp], 1.0, p0)
        z_m[mp] = H * np.log(p0 / p)
        
        # which_vert==-1 and which_vert==3 are already meters-ish
        zmax_m = 18000.0
        near_sfc &= np.isfinite(z_m) & (z_m >= 0.0) & (z_m <= zmax_m)
        
        # --- subset ---
        lon_s   = lon[near_sfc]
        lat_s   = lat[near_sfc]
        obs_s   = obs[near_sfc]
        z_s     = z[near_sfc]
        wvt_s   = wvt[near_sfc]
        otype_s = otype[near_sfc]  # <-- add this
        
        # --- optional: obs type id -> name mapping ---
        # DART obs_type values are typically 1-based ids matching ObsTypesMetaData
        obs_type_names = ds["ObsTypesMetaData"].values if "ObsTypesMetaData" in ds else None
        
        def obs_type_label(type_id):
            if obs_type_names is None:
                return f"type {int(type_id)}"
            i = int(type_id) - 1  # DART ids are usually 1-based
            if 0 <= i < len(obs_type_names):
                raw = obs_type_names[i]
                return raw.decode().strip() if isinstance(raw, (bytes, np.bytes_)) else str(raw).strip()
            return f"type {int(type_id)}"

        
        # --- plot one scatter per obs type ---
        sc_last = None
        unique_types = np.unique(otype_s)
        all_types_seen.update(unique_types.tolist())
        for t_id in unique_types:
            obs_type_name_map[int(t_id)] = obs_type_label(t_id)
        
        for k, t_id in enumerate(unique_types):
            msk = (otype_s == t_id)
            mk = markers[k % len(markers)]
        
            sc_last = m.scatter(
                lon_s[msk], lat_s[msk],
                c=obs_s[msk],
                s=30,
                cmap="bwr",
                alpha=0.85,
                marker=mk,
                label=obs_type_label(t_id),
                latlon=True,
                ax=ax,
                zorder=20, 
            )

# ----------------------------
# One figure-level legend (custom handles)
# ----------------------------
# legend_handles = [
#     Line2D([0], [0], color="k", marker=r"$\rightarrow$", markersize=12, lw=0,
#            linestyle="None", label="10 m wind increment"),
#     Line2D([0], [0], color="goldenrod", lw=2, label="Positive increments"),
#     Line2D([0], [0], color="deepskyblue", lw=2, label="Negative increments"),
# ]

# fig.legend(
#     handles=legend_handles,
#     bbox_to_anchor=(1, 0.03),
#     ncol=3,
#     frameon=True
# )
legend_handles = [
    Line2D([0], [0], color="k", marker=r"$\rightarrow$", markersize=12, lw=0,
           linestyle="None", label="10 m wind increment"),
    Line2D([0], [0], color="goldenrod", lw=2, label="Positive increments"),
    Line2D([0], [0], color="deepskyblue", lw=2, label="Negative increments"),
]

# Add marker-only handles for each observation type seen across all panels
obs_marker_handles = []
for k, t_id in enumerate(sorted(all_types_seen)):
    mk = markers[k % len(markers)]
    obs_marker_handles.append(
        Line2D([0], [0],
               marker=mk,
               linestyle="None",
               color="k",
               markerfacecolor="none",
               markeredgecolor="k",
               markersize=7,
               label=obs_type_name_map.get(int(t_id), f"type {int(t_id)}"))
    )

legend_handles_all = legend_handles + obs_marker_handles

fig.legend(
    handles=legend_handles_all,
    bbox_to_anchor=(0.87, 0.03),
    ncol=3,
    frameon=True
)

cax = fig.add_axes([0.25, 0.663, 0.5, 0.018])  # make it a bit wider
cbar = fig.colorbar(cs, cax=cax, orientation="horizontal")

cax2 = fig.add_axes([0.25, 0.33, 0.5, 0.018])  # make it a bit wider
cbar2 = fig.colorbar(sc_last, cax=cax2, orientation="horizontal", extend='both',label='observation Values')

from matplotlib.ticker import MaxNLocator, ScalarFormatter
cbar.ax.xaxis.set_major_locator(MaxNLocator(3))

fmt = ScalarFormatter(useOffset=False)#==fmt.set_scientific(False)   
cbar.ax.xaxis.set_major_formatter(fmt)

cbar.ax.tick_params(axis="x", labelsize=8, pad=1)
cbar.update_ticks()         
cbar.set_label(f"Background {var}")
# plt.savefig(f"fig/increments_{str(cycle_dirs[0])[49:-11]}.jpeg")
```

```python
# ── Cell 1 : imports ──────────────────────────────────────────────────────────
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from matplotlib.ticker import AutoMinorLocator

# ── Cell 2 : file paths ───────────────────────────────────────────────────────
pre_file  = "/gpfs/home/sa24m/stephen_asare/runs/unfinished_kde_qv_only_inf_0/2015071418/preassim_mean_d01.nc"
post_file = "/gpfs/home/sa24m/stephen_asare/runs/unfinished_kde_qv_only_inf_0/2015071418/postassim_mean_d01.nc"
obs_file  = "/gpfs/home/sa24m/stephen_asare/runs/unfinished_kde_qv_only_inf_0/2015071418/obs_epoch_001.nc"

ds_pre  = xr.open_dataset(pre_file)
ds_post = xr.open_dataset(post_file)
ds_obs  = xr.open_dataset(obs_file)

# ── Cell 3 : decode DART string metadata ─────────────────────────────────────
def decode_str_array(da):
    """
    Safely decode a DART char/byte metadata array to a list of plain strings.

    DART writes fixed-length strings into netCDF as dtype |S32 (or similar).
    xarray loads each element as a scalar numpy bytes_ object — NOT as an
    iterable of individual bytes — so b"".join(row) fails.  This function
    handles that case plus the older character-array layout where each row IS
    an iterable of single-byte numpy bytes_ values.
    """
    result = []
    for item in da.values:
        if isinstance(item, (bytes, np.bytes_)):
            # scalar byte string  e.g. b'observation               '
            result.append(item.decode("ascii", errors="replace").strip())
        elif isinstance(item, (np.str_, str)):
            result.append(str(item).strip())
        else:
            # fallback: character-array layout — join individual byte elements
            try:
                result.append(b"".join(
                    c if isinstance(c, bytes) else bytes([c]) for c in item
                ).decode("ascii", errors="replace").strip())
            except Exception:
                result.append("".join(str(c) for c in item).strip())
    return result

copy_meta     = decode_str_array(ds_obs["CopyMetaData"])
obs_type_meta = decode_str_array(ds_obs["ObsTypesMetaData"])

print("CopyMetaData entries:")
for i, m in enumerate(copy_meta):
    print(f"  [{i+1:3d}]  {m}")

# ── Cell 4 : locate key copy indices (0-based) ────────────────────────────────
def find_copy(meta_list, keyword, exclude=()):
    for i, m in enumerate(meta_list):
        ml = m.lower()
        if keyword in ml and not any(ex in ml for ex in exclude):
            return i
    raise ValueError(f"Copy containing '{keyword}' not found in CopyMetaData.")

obs_cidx        = find_copy(copy_meta, "observation",
                             exclude=["prior", "posterior", "ensemble"])
prior_mean_cidx = find_copy(copy_meta, "prior ensemble mean")
post_mean_cidx  = find_copy(copy_meta, "posterior ensemble mean")

print(f"\nobs copy index        : {obs_cidx + 1}  → '{copy_meta[obs_cidx]}'")
print(f"prior mean copy index : {prior_mean_cidx + 1}  → '{copy_meta[prior_mean_cidx]}'")
print(f"post  mean copy index : {post_mean_cidx + 1}  → '{copy_meta[post_mean_cidx]}'")

# ── Cell 5 : filter for SPECIFIC_HUMIDITY observations ───────────────────────
qv_type_nums = [
    i + 1   # DART type numbers are 1-based
    for i, m in enumerate(obs_type_meta)
    if "specific_humidity" in m.lower()
]
print(f"\nSpecific-humidity obs type numbers: {qv_type_nums}")
if not qv_type_nums:
    raise RuntimeError("No SPECIFIC_HUMIDITY obs type found in ObsTypesMetaData.")

obs_type_vals = ds_obs["obs_type"].values          # (ObsIndex,)
mask          = np.isin(obs_type_vals, qv_type_nums)
print(f"Number of QVAPOR observations passing filter : {mask.sum()}")
if mask.sum() == 0:
    raise RuntimeError("No specific-humidity obs found — check obs_type values.")

# ── Cell 6 : extract values, locations, vertical coordinate ──────────────────
observations = ds_obs["observations"].values           # (ObsIndex, copy)
obs_vals     = observations[mask, obs_cidx]            # (n_qv,)  kg/kg
prior_mod    = observations[mask, prior_mean_cidx]     # (n_qv,)  kg/kg
post_mod     = observations[mask, post_mean_cidx]      # (n_qv,)  kg/kg

location   = ds_obs["location"].values[mask]           # (n_qv, 3)  radians
which_vert = ds_obs["which_vert"].values[mask]         # (n_qv,)

obs_lon_deg  = np.degrees(location[:, 0])
obs_lat_deg  = np.degrees(location[:, 1])
obs_vert_raw = location[:, 2]

# which_vert == 2  →  pressure in Pa
pres_Pa  = np.where(which_vert == 2, obs_vert_raw, np.nan)
pres_hPa = pres_Pa / 100.0

print(f"\nPressure range : {np.nanmin(pres_hPa):.1f} – {np.nanmax(pres_hPa):.1f} hPa")
print(f"Lat  range     : {obs_lat_deg.min():.3f} – {obs_lat_deg.max():.3f} °N")
print(f"Lon  range     : {obs_lon_deg.min():.3f} – {obs_lon_deg.max():.3f} °E")

# ── Cell 7 : compute altitude from WRF geopotential ──────────────────────────
g        = 9.80665
PH       = ds_pre["PH"].values[0]    # (nz+1, ny, nx)  perturbation geopotential
PHB      = ds_pre["PHB"].values      # (nz+1, ny, nx)  base geopotential
geo_stag = PH + PHB                  # total geopotential on staggered (w) levels
z_3d     = 0.5 * (geo_stag[:-1] + geo_stag[1:]) / g   # mass-level altitude m MSL

try:
    MU  = ds_pre["MU"].values[0]    # (ny, nx)
    MUB = ds_pre["MUB"].values      # (ny, nx)
    C1H = ds_pre["C1H"].values      # (nz,)
    C2H = ds_pre["C2H"].values      # (nz,)

    mu_tot_mean = (MU + MUB).mean()
    p_col_hPa   = (C1H * mu_tot_mean + C2H) / 100.0   # (nz,) domain-mean pressure hPa
    z_col       = z_3d.mean(axis=(1, 2))               # (nz,) domain-mean altitude m MSL

    # np.interp needs ascending x; pressure decreases with altitude, so flip
    sort_asc = np.argsort(p_col_hPa)          # ascending pressure → ascending altitude  
    obs_alt_m    = np.interp(pres_hPa,
                              p_col_hPa[sort_asc],
                              z_col[sort_asc])
    use_altitude = True
    print("Altitude computed from WRF geopotential.")
except Exception as e:
    print(f"Altitude computation failed ({e}); using pressure (hPa) on y-axis.")
    use_altitude = False

# ── Cell 8 : choose x-axis ───────────────────────────────────────────────────
lon_range = obs_lon_deg.max() - obs_lon_deg.min()
lat_range = obs_lat_deg.max() - obs_lat_deg.min()

is_single_station = (lon_range < 0.05 and lat_range < 0.05)

if not is_single_station:
    if lat_range >= lon_range:
        x_vals  = obs_lat_deg
        x_label = "Latitude (°N)"
    else:
        x_vals  = obs_lon_deg
        x_label = "Longitude (°E)"

y_vals  = obs_alt_m  if use_altitude else pres_hPa
y_label = "Altitude (m MSL)" if use_altitude else "Pressure (hPa)"

# ── Cell 9 : compute departures (g/kg) ───────────────────────────────────────
dep_prior = (prior_mod - obs_vals) * 1e3
dep_post  = (post_mod  - obs_vals) * 1e3

print(f"\nPrior departure — mean: {dep_prior.mean():.4f}  std: {dep_prior.std():.4f} g/kg")
print(f"Post  departure — mean: {dep_post.mean():.4f}  std: {dep_post.std():.4f} g/kg")

# ── Cell 10 : shared diverging colour scale ───────────────────────────────────
vmax = max(np.abs(dep_prior).max(), np.abs(dep_post).max(), 0.01)
norm = mcolors.TwoSlopeNorm(vmin=-vmax, vcenter=0, vmax=vmax)
cmap = "RdBu_r"

# ── Cell 11 : plot ────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(13, 7), sharey=True, constrained_layout=True)

fig.suptitle(
    "QVAPOR Departure  (Model − Obs)  |  2015-07-14 18Z\n"
    f"Assimilated RADIOSONDE_SPECIFIC_HUMIDITY  (n = {mask.sum()})",
    fontsize=13, fontweight="bold",
)

for ax, dep, panel_title in [
    (axes[0], dep_prior, "Prior  (pre-assimilation)"),
    (axes[1], dep_post,  "Posterior  (post-assimilation)"),
]:
    if is_single_station:
        sc = ax.scatter(dep, y_vals,
                        c=dep, cmap=cmap, norm=norm,
                        s=70, edgecolors="k", linewidths=0.4, zorder=4)
        ax.axvline(0, color="k", lw=1, ls="--", alpha=0.7, zorder=3)
        sort_y = np.argsort(y_vals)
        ax.plot(dep[sort_y], y_vals[sort_y],
                color="gray", lw=0.8, ls="-", zorder=2, alpha=0.6)
        ax.set_xlabel("Model − Obs  (g kg⁻¹)", fontsize=11)
        ax.xaxis.set_minor_locator(AutoMinorLocator())
    else:
        sc = ax.scatter(x_vals, y_vals,
                        c=dep, cmap=cmap, norm=norm,
                        s=80, edgecolors="k", linewidths=0.4, zorder=4)
        ax.set_xlabel(x_label, fontsize=11)
        cbar = plt.colorbar(sc, ax=ax, pad=0.02, aspect=30)
        cbar.set_label("Model − Obs  (g kg⁻¹)", fontsize=10)
        cbar.ax.axhline(0, color="k", lw=1)

    ax.set_ylabel(y_label, fontsize=11)
    if not use_altitude:
        ax.invert_yaxis()
    ax.set_title(panel_title, fontsize=11, pad=6)
    ax.grid(True, alpha=0.3, which="both")
    ax.yaxis.set_minor_locator(AutoMinorLocator())

if is_single_station:
    sm   = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
    cbar = fig.colorbar(sm, ax=axes.ravel().tolist(), pad=0.02, aspect=35)
    cbar.set_label("Model − Obs  (g kg⁻¹)", fontsize=10)
    cbar.ax.axhline(0, color="k", lw=1)
    fig.text(0.5, -0.01,
             f"Station  lat = {obs_lat_deg.mean():.3f}°N,  "
             f"lon = {obs_lon_deg.mean():.3f}°E",
             ha="center", fontsize=10, style="italic")

out_path = "/gpfs/home/sa24m/stephen_asare/runs/unfinished_kde_qv_only_inf_0/2015071418/qv_departure_profile.png"
# plt.savefig(out_path, dpi=150, bbox_inches="tight")
# print(f"\nFigure saved → {out_path}")
plt.show()
```

```python
#!/usr/bin/env python3
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize

# ============================================================
# USER SETTINGS
# ============================================================
# path = Path("/gpfs/home/sa24m/scratch/tqprof/run2/osse_out/dart_cycle/output/2015071412/")

pre_file  = "/gpfs/home/sa24m/scratch/tqprof/run2/osse_out/dart_cycle/output/2015071412/preassim_mean_d01.nc"
post_file = "/gpfs/home/sa24m/scratch/tqprof/run2/osse_out/dart_cycle/output/2015071412/postassim_mean_d01.nc"
obs_file  = "/gpfs/home/sa24m/scratch/tqprof/run2/osse_out/dart_cycle/output/2015071412/obs_epoch_001.nc"         # produced by obs_diag/obs_epoch step

TIME_INDEX = 0

# Choose what to analyze from the posterior/preassim files:
# Examples: "THM", "QVAPOR", "QCLOUD", "QRAIN", "QICE", "QSNOW", "QGRAUP", "U", "V", "T2", "U10", "V10"
VAR = "THM"

# Choose what you want to visualize: "increment" (post-pre) or "post" or "pre"
MODE = "increment"

# Vertical coordinate (your files support AGL via PH/PHB/HGT)
VERT_MODE = "AGL"      # only "AGL" implemented here

# If VAR is 3D, you can restrict to a height range for plotting
ZMIN_M = 0.0
ZMAX_M = 18000.0

# Obs: only radiosonde temperature (since that's what you assimilated)
OBS_TYPE_NAME = "RADIOSONDE_TEMPERATURE"
OBS_COPY_IDX = 0
QC_COPY_IDX  = 1
MAX_QC = 2

# Convert obs pressure -> height approximation (only for plotting)
P0 = 101325.0
H  = 8400.0

# ============================================================
# HELPERS (minimal, only what this script needs)
# ============================================================
def get_obs_type_id(ds_obs, name):
    """Return obs_type integer id for a given ObsTypesMetaData name."""
    meta = ds_obs["ObsTypesMetaData"].values
    names = []
    for v in meta:
        if isinstance(v, (bytes, np.bytes_)):
            names.append(v.decode().strip())
        else:
            names.append(str(v).strip())
    # DART obs_type ids are usually 1-based
    for k, nm in enumerate(names, start=1):
        if nm == name:
            return k
    raise ValueError(f"Obs type '{name}' not found in ObsTypesMetaData. Available (sample): {names[:10]} ...")

def nearest_ij(lat2d, lon2d, lat0, lon0):
    dist2 = (lat2d - lat0)**2 + (lon2d - lon0)**2
    j, i = np.unravel_index(np.argmin(dist2), dist2.shape)
    return int(j), int(i)

def destagger_to_mass(ds, var, time_index=0):
    """
    Returns a mass-grid version of U/V/W if staggered.
    - U: (bt,y,x_stag) -> average to (bt,y,x)
    - V: (bt,y_stag,x) -> average to (bt,y,x)
    - W: (bt_stag,y,x) -> average to (bt,y,x)
    Other vars returned as-is.
    """
    da = ds[var].isel(Time=time_index)

    if var == "U":  # west_east_stag
        return 0.5 * (da.isel(west_east_stag=slice(0, -1)) + da.isel(west_east_stag=slice(1, None))).rename({"west_east_stag":"west_east"})
    if var == "V":  # south_north_stag
        return 0.5 * (da.isel(south_north_stag=slice(0, -1)) + da.isel(south_north_stag=slice(1, None))).rename({"south_north_stag":"south_north"})
    if var == "W":  # bottom_top_stag
        return 0.5 * (da.isel(bottom_top_stag=slice(0, -1)) + da.isel(bottom_top_stag=slice(1, None))).rename({"bottom_top_stag":"bottom_top"})
    return da

def z_agl_mass_levels(ds, time_index=0):
    """AGL height (m) on mass levels: shape (bottom_top, y, x)."""
    Zstag = (ds["PH"].isel(Time=time_index).values + ds["PHB"].values) / 9.81   # (bt_stag,y,x)
    Zmass = 0.5 * (Zstag[:-1, :, :] + Zstag[1:, :, :])                         # (bt,y,x)
    HGT   = ds["HGT"].values                                                    # (y,x)
    return Zmass - HGT[None, :, :]

# ============================================================
# LOAD DATA
# ============================================================
ds_pre  = xr.open_dataset(pre_file)
ds_post = xr.open_dataset(post_file)
ds_obs  = xr.open_dataset(obs_file)

# ============================================================
# 1) OBS: Extract ONLY radiosonde temperature, compute a single station lat/lon
# ============================================================
sonde_id = get_obs_type_id(ds_obs, OBS_TYPE_NAME)

lon = ds_obs["location"].isel(locdim=0).values.astype(float)
lat = ds_obs["location"].isel(locdim=1).values.astype(float)
z   = ds_obs["location"].isel(locdim=2).values.astype(float)
wvt = ds_obs["which_vert"].values
otype = ds_obs["obs_type"].values

obs = ds_obs["observations"].isel(copy=OBS_COPY_IDX).values.astype(float)
qc  = ds_obs["qc"].isel(qc_copy=QC_COPY_IDX).values.astype(float) if "qc" in ds_obs else None

# radians -> degrees if needed
if np.nanmax(np.abs(lon)) <= 2*np.pi + 0.1 and np.nanmax(np.abs(lat)) <= np.pi/2 + 0.1:
    lon = np.degrees(lon); lat = np.degrees(lat)

# filter to radiosonde temperature + QC
mask = (otype == sonde_id) & np.isfinite(lon) & np.isfinite(lat) & np.isfinite(obs) & np.isfinite(z)
if qc is not None:
    mask &= np.isfinite(qc) & (qc <= MAX_QC)

lon_s = lon[mask]
lat_s = lat[mask]
z_s   = z[mask]
wvt_s = wvt[mask]
obs_s = obs[mask]

# station location = median of obs locations (robust)
lat0 = float(np.nanmedian(lat_s))
lon0 = float(np.nanmedian(lon_s))

# convert obs vertical to "height-like" meters for plotting
# - if pressure Pa (which_vert==2): convert to height using scale height
z_obs_m = z_s.copy()
mp = (wvt_s == 2) & np.isfinite(z_obs_m)
p = np.clip(z_obs_m[mp], 1.0, P0)
z_obs_m[mp] = H * np.log(P0 / p)

# keep within plot range
okz = np.isfinite(z_obs_m) & (z_obs_m >= ZMIN_M) & (z_obs_m <= ZMAX_M)
z_obs_m = z_obs_m[okz]
obs_s   = obs_s[okz]

# ============================================================
# 2) MODEL: extract vertical profile at nearest gridpoint to the station
# ============================================================
lat2d = ds_post["XLAT"].values.astype(float)
lon2d = ds_post["XLONG"].values.astype(float)

# normalize lon0 to match WRF lon convention
if np.nanmax(lon2d) > 180.0 and np.nanmin(lon2d) >= 0.0:
    lon0_use = lon0 % 360.0
else:
    lon0_use = ((lon0 + 180.0) % 360.0) - 180.0

j0, i0 = nearest_ij(lat2d, lon2d, lat0, lon0_use)

# vertical coordinate for model (AGL, meters)
ZAGL = z_agl_mass_levels(ds_post, time_index=TIME_INDEX)  # (bt,y,x)
z_mod_m = ZAGL[:, j0, i0]

# model variable profile at mass levels (or 2D vars)
def extract_profile(ds):
    if VAR in ["T2", "U10", "V10", "Q2", "PSFC"]:
        # 2D: treat as a single point at z=0
        val = float(ds[VAR].isel(Time=TIME_INDEX, south_north=j0, west_east=i0).values)
        return np.array([0.0]), np.array([val])
    else:
        da = destagger_to_mass(ds, VAR, time_index=TIME_INDEX)
        val = da[:, j0, i0].values.astype(float)   # (bottom_top,)
        return z_mod_m.astype(float), val

z_pre,  v_pre  = extract_profile(ds_pre)
z_post, v_post = extract_profile(ds_post)

if MODE == "increment":
    z_plot = z_post
    v_plot = v_post - v_pre
    model_label = f"Δ{VAR} (post-pre)"
elif MODE == "post":
    z_plot = z_post
    v_plot = v_post
    model_label = f"{VAR} (post)"
elif MODE == "pre":
    z_plot = z_pre
    v_plot = v_pre
    model_label = f"{VAR} (pre)"
else:
    raise ValueError("MODE must be one of: increment, post, pre")

# trim model height range (for 3D vars)
mz = np.isfinite(z_plot) & np.isfinite(v_plot) & (z_plot >= ZMIN_M) & (z_plot <= ZMAX_M)
z_plot = z_plot[mz]
v_plot = v_plot[mz]

# ============================================================
# 3) CONSISTENT COLOR LEVELS for model & obs
# ============================================================
# Note: obs are temperature (K?) while model variable can be anything.
# Consistent color scales only makes physical sense if VAR is also temperature-like.
# You asked for consistent levels, so we enforce it numerically regardless.
vals_all = np.concatenate([v_plot[np.isfinite(v_plot)], obs_s[np.isfinite(obs_s)]])
vmin = float(np.nanmin(vals_all))
vmax = float(np.nanmax(vals_all))
norm = Normalize(vmin=vmin, vmax=vmax)

# ============================================================
# 4) PLOT: single axes, two colorbars (left=model, right=obs)
# ============================================================
fig, ax = plt.subplots(figsize=(6.5, 8.0), constrained_layout=True)

# Model profile points (colored)
sc_model = ax.scatter(
    v_plot, z_plot,
    c=v_plot, norm=norm, cmap="viridis",
    s=35, edgecolor="none", label="Model (at station gridpoint)"
)

# Obs points (colored with same norm)
sc_obs = ax.scatter(
    obs_s, z_obs_m,
    c=obs_s, norm=norm, cmap="viridis",
    s=55, marker="x", linewidths=1.5, label=f"Obs: {OBS_TYPE_NAME}"
)

ax.axvline(0.0, color="k", linewidth=1) if MODE == "increment" else None

ax.set_xlabel(model_label if MODE != "increment" else f"{model_label}  (station: j={j0}, i={i0})")
ax.set_ylabel("Height AGL (m)")
ax.set_ylim(ZMIN_M, ZMAX_M)
ax.grid(True, alpha=0.3)
ax.set_title(f"Single-sonde impact @ ({lat0:.3f}, {lon0:.3f})  |  VAR={VAR}  |  MODE={MODE}")

# Two colorbars: left for model, right for obs (same norm, same cmap)
cbar_left = fig.colorbar(sc_model, ax=ax, location="left", pad=0.02)
cbar_left.set_label("Model values (same scale)")

cbar_right = fig.colorbar(sc_obs, ax=ax, location="right", pad=0.02)
cbar_right.set_label("Obs values (same scale)")

ax.legend(loc="lower right", frameon=True)

plt.show()

ds_pre.close(); ds_post.close(); ds_obs.close()
```

```python
#!/usr/bin/env python3
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
from pathlib import Path
from matplotlib.ticker import MaxNLocator, ScalarFormatter

# =============================================================================
# USER OPTIONS
# =============================================================================
# CYCLE_DIR   = "/gpfs/home/sa24m/stephen_asare/runs/unfinished_kde_qv_only_inf_0/2015071418/"
CYCLE_DIR   = "/gpfs/home/sa24m/scratch/tqprof/run2/osse_out/dart_cycle/output/2015071418/"
TIME_INDEX  = 0
ZAGL_M      = 3000.0

DOMS        = ["d01", "d02"]             # row 0 = d01, row 1 = d02
VARS_2COL   = ["THM", "QVAPOR"]          # col 0..1 in each row

INC_SOURCE  = "file"                    # "file" (analysis_increment) or "diff" (post-pre)

# winds (background wind at same ZAGL in every panel)
PLOT_WIND     = True
WIND_U        = "U"
WIND_V        = "V"
WIND_SKIP     = 7
QUIVER_SCALE  = 350
QUIVER_WIDTH  = 0.002

# plotting style
BG_LEVELS    = 21
INC_LEVELS   = 17
INC_ALPHA    = 0.55
INC_THRESH   = None   # e.g. 1e-4, or None
CMAP_BG      = "Greens"
CMAP_INC     = "bwr"

# OBS OVERLAY (from obs_epoch_001.nc produced by obs_diag)
PLOT_OBS          = True
OBS_FILENAME      = "obs_epoch_001.nc"          # in each cycle dir, or give full path
OBS_TYPE_NAME     = "RADIOSONDE_TEMPERATURE"    # you assimilated only this
OBS_COPY_IDX      = 0
QC_COPY_IDX       = 1
MAX_QC            = 2
OBS_DZ_M          = 500.0   # plot obs within +/- this of ZAGL_M
OBS_MARKER        = "o"
OBS_SIZE          = 35
OBS_ALPHA         = 0.9
OBS_CMAP          = "bwr"

# =============================================================================
# ASSUMES you already have: setup_basemap_from_file(nc_path) -> (x, y, m)
# =============================================================================

# =============================================================================
# Helpers
# =============================================================================
def safe_levels(vmin, vmax, n, pad_frac=1e-6, pad_abs=1e-12):
    vmin = float(vmin); vmax = float(vmax)
    if not np.isfinite(vmin) or not np.isfinite(vmax):
        return np.linspace(0.0, 1.0, n)
    if vmax <= vmin:
        pad = max(pad_abs, pad_frac * (abs(vmin) + 1.0))
        vmin -= pad
        vmax += pad
    return np.linspace(vmin, vmax, n)

def _destagger_to_mass(ds, var, time_index=0):
    da = ds[var].isel(Time=time_index) if "Time" in ds[var].dims else ds[var]
    if var == "U":
        return 0.5 * (da.isel(west_east_stag=slice(0, -1)) + da.isel(west_east_stag=slice(1, None))).rename({"west_east_stag":"west_east"})
    if var == "V":
        return 0.5 * (da.isel(south_north_stag=slice(0, -1)) + da.isel(south_north_stag=slice(1, None))).rename({"south_north_stag":"south_north"})
    if var == "W":
        return 0.5 * (da.isel(bottom_top_stag=slice(0, -1)) + da.isel(bottom_top_stag=slice(1, None))).rename({"bottom_top_stag":"bottom_top"})
    return da

def _k_at_zagl(ds_ref, zagl_m, time_index=0):
    Zstag = (ds_ref["PH"].isel(Time=time_index).values + ds_ref["PHB"].values) / 9.81
    Zmass = 0.5 * (Zstag[:-1, :, :] + Zstag[1:, :, :])
    HGT   = ds_ref["HGT"].values
    ZAGL  = Zmass - HGT[None, :, :]
    return np.argmin(np.abs(ZAGL - zagl_m), axis=0)

def field_at_zagl(ds_field, ds_ref, var, zagl_m, time_index=0):
    # 2D (Time,y,x)
    if ds_field[var].ndim == 3 and ("bottom_top" not in ds_field[var].dims) and ("bottom_top_stag" not in ds_field[var].dims):
        return ds_field[var].isel(Time=time_index).values

    # 3D mass direct or destagger
    if "bottom_top" in ds_field[var].dims:
        v3d = ds_field[var].isel(Time=time_index).values
    else:
        da = _destagger_to_mass(ds_field, var, time_index=time_index)
        if "bottom_top" not in da.dims:
            return da.values
        v3d = da.values

    k = _k_at_zagl(ds_ref, zagl_m=zagl_m, time_index=time_index)
    j, i = np.indices(k.shape)
    return v3d[k, j, i]

def _obs_type_id_from_name(ds_obs, name):
    if "ObsTypesMetaData" not in ds_obs:
        return None
    meta = ds_obs["ObsTypesMetaData"].values
    names = []
    for v in meta:
        if isinstance(v, (bytes, np.bytes_)):
            names.append(v.decode().strip())
        else:
            names.append(str(v).strip())
    for k, nm in enumerate(names, start=1):   # DART ids are usually 1-based
        if nm == name:
            return k
    return None

def extract_obs_near_zagl(obs_path, zagl_m, dz_m, *, obs_type_name, obs_copy_idx, qc_copy_idx, max_qc):
    """
    Returns lon_deg, lat_deg, obs_val for obs within +/- dz_m of zagl_m (using pressure->height approx).
    """
    ds = xr.open_dataset(obs_path)

    lon = ds["location"].isel(locdim=0).values.astype(float)
    lat = ds["location"].isel(locdim=1).values.astype(float)
    z   = ds["location"].isel(locdim=2).values.astype(float)
    wvt = ds["which_vert"].values
    obs = ds["observations"].isel(copy=obs_copy_idx).values.astype(float)

    qc = None
    if "qc" in ds:
        qc = ds["qc"].isel(qc_copy=qc_copy_idx).values.astype(float)

    otype = ds["obs_type"].values if "obs_type" in ds else None

    # radians heuristic
    if np.nanmax(np.abs(lon)) <= 2*np.pi + 0.1 and np.nanmax(np.abs(lat)) <= np.pi/2 + 0.1:
        lon = np.degrees(lon)
        lat = np.degrees(lat)

    # convert obs vertical to "height-like" meters
    z_m = z.astype(float).copy()
    p0 = 101325.0
    H  = 8400.0
    mp = (wvt == 2) & np.isfinite(z_m)        # pressure (Pa) -> height
    p = np.clip(z_m[mp], 1.0, p0)
    z_m[mp] = H * np.log(p0 / p)

    msk = np.isfinite(lon) & np.isfinite(lat) & np.isfinite(obs) & np.isfinite(z_m)

    if qc is not None:
        msk &= np.isfinite(qc) & (qc <= max_qc)

    # obs type filter (only radiosonde temperature)
    if otype is not None:
        tid = _obs_type_id_from_name(ds, obs_type_name)
        if tid is not None:
            msk &= (otype == tid)

    # near requested level
    msk &= (np.abs(z_m - zagl_m) <= dz_m)

    lon_s = lon[msk]
    lat_s = lat[msk]
    obs_s = obs[msk]

    ds.close()
    return lon_s, lat_s, obs_s

# =============================================================================
# Load fields for both domains
# =============================================================================
cdir = Path(CYCLE_DIR)
DATA = {}   # (dom,var) -> dict(bg=..., inc=...)
GEO  = {}   # dom -> dict(x,y,m,lon,lat,u,v)

for dom in DOMS:
    post_file = cdir / f"postassim_mean_{dom}.nc"
    pre_file  = cdir / f"preassim_mean_{dom}.nc"
    inc_file  = cdir / f"analysis_increment_{dom}.nc"

    if not post_file.exists():
        raise FileNotFoundError(f"Missing: {post_file}")
    if INC_SOURCE == "diff" and (not pre_file.exists()):
        raise FileNotFoundError(f"Missing (INC_SOURCE='diff'): {pre_file}")
    if INC_SOURCE == "file" and (not inc_file.exists()):
        raise FileNotFoundError(f"Missing (INC_SOURCE='file'): {inc_file}")

    x, y, m = setup_basemap_from_file(str(post_file))

    ds_post = xr.open_dataset(post_file)
    ds_pre  = xr.open_dataset(pre_file) if (INC_SOURCE == "diff") else None
    ds_inc  = xr.open_dataset(inc_file) if (INC_SOURCE == "file") else None

    lon2d = ds_post["XLONG"].values
    lat2d = ds_post["XLAT"].values

    if PLOT_WIND:
        u2d = field_at_zagl(ds_post, ds_post, WIND_U, ZAGL_M, time_index=TIME_INDEX)
        v2d = field_at_zagl(ds_post, ds_post, WIND_V, ZAGL_M, time_index=TIME_INDEX)
    else:
        u2d = v2d = None

    GEO[dom] = dict(x=x, y=y, m=m, lon=lon2d, lat=lat2d, u=u2d, v=v2d, post_file=str(post_file))

    for var in VARS_2COL:
        bg2d = field_at_zagl(ds_post, ds_post, var, ZAGL_M, time_index=TIME_INDEX)

        if INC_SOURCE == "file":
            inc2d = field_at_zagl(ds_inc, ds_post, var, ZAGL_M, time_index=TIME_INDEX)
        else:
            pre2d = field_at_zagl(ds_pre, ds_post, var, ZAGL_M, time_index=TIME_INDEX)
            inc2d = bg2d - pre2d

        if INC_THRESH is not None:
            inc2d = np.where(np.abs(inc2d) >= INC_THRESH, inc2d, np.nan)

        DATA[(dom, var)] = dict(bg=bg2d, inc=inc2d)

    if ds_inc is not None:
        ds_inc.close()
    if ds_pre is not None:
        ds_pre.close()
    ds_post.close()

# =============================================================================
# Shared levels per COLUMN across d01/d02
# =============================================================================
LEVELS = {}
for var in VARS_2COL:
    bmins, bmaxs = [], []
    for dom in DOMS:
        b = DATA[(dom, var)]["bg"]
        bmins.append(np.nanmin(b))
        bmaxs.append(np.nanmax(b))
    levels_bg = safe_levels(np.nanmin(bmins), np.nanmax(bmaxs), BG_LEVELS)

    absmaxs = []
    for dom in DOMS:
        inc = DATA[(dom, var)]["inc"]
        absmaxs.append(np.nanmax(np.abs(inc)))
    inc_absmax = float(np.nanmax(absmaxs))
    if not np.isfinite(inc_absmax) or inc_absmax == 0.0:
        inc_absmax = 1e-12
    levels_inc = np.linspace(-inc_absmax, inc_absmax, INC_LEVELS)

    LEVELS[var] = dict(levels_bg=levels_bg, levels_inc=levels_inc)

# =============================================================================
# Plot 2 rows x 2 columns + OBS overlay
# =============================================================================
fig, axs = plt.subplots(2, 2, figsize=(13.5, 8.5), constrained_layout=False)
plt.subplots_adjust(left=0.06, right=0.97, bottom=0.06, top=0.92, wspace=0.26, hspace=0.14)

for r, dom in enumerate(DOMS):
    g = GEO[dom]
    m = g["m"]
    x = g["x"]; y = g["y"]

    # obs for this cycle (same for both vars in row)
    lon_obs = lat_obs = obs_val = None
    if PLOT_OBS:
        obs_path = (cdir / OBS_FILENAME) if not Path(OBS_FILENAME).is_absolute() else Path(OBS_FILENAME)
        if obs_path.exists():
            lon_obs, lat_obs, obs_val = extract_obs_near_zagl(
                str(obs_path), ZAGL_M, OBS_DZ_M,
                obs_type_name=OBS_TYPE_NAME,
                obs_copy_idx=OBS_COPY_IDX,
                qc_copy_idx=QC_COPY_IDX,
                max_qc=MAX_QC
            )
        else:
            print(f"WARNING: obs file not found: {obs_path} (skipping obs overlay)")

    for c, var in enumerate(VARS_2COL):
        ax = axs[r, c]

        m.drawcoastlines(ax=ax, linewidth=0.5)
        m.drawstates(ax=ax, linewidth=0.3)
        m.drawcountries(ax=ax, linewidth=0.3)

        bg2d  = DATA[(dom, var)]["bg"]
        inc2d = DATA[(dom, var)]["inc"]
        lev_bg  = LEVELS[var]["levels_bg"]
        lev_inc = LEVELS[var]["levels_inc"]

        cs_bg  = m.contourf(x, y, bg2d,  levels=lev_bg,  extend="both", ax=ax, cmap=CMAP_BG)
        pos_inc = np.where(inc2d > 0, inc2d, np.nan)
        neg_inc = np.where(inc2d < 0, inc2d, np.nan)
        # pos_levels = np.linspace(np.nanmax([np.nanmin(inc2d[inc2d > 0]), 1e-12]), np.nanmax(inc2d), INC_LEVELS)
        # neg_levels = np.linspace(np.nanmin(inc2d), np.nanmin([np.nanmax(inc2d[inc2d < 0]), -1e-12]), INC_LEVELS)

        pos_levels = np.linspace(0, np.nanmax(inc2d), INC_LEVELS)
        neg_levels = np.linspace(np.nanmin(inc2d), 0, INC_LEVELS)
        
        cs_inc_pos = m.contour(x, y, pos_inc, levels=pos_levels, ax=ax, cmap="Wistia", alpha=INC_ALPHA)
        cs_inc_neg = m.contour(x, y, neg_inc, levels=neg_levels, ax=ax, cmap="cool",   alpha=INC_ALPHA)
        # cs_inc = m.contour(x, y, inc2d, levels=lev_inc, extend="both", ax=ax, cmap=CMAP_INC, alpha=INC_ALPHA)
        # cs_inc = m.contour(x, y, inc2d.where(lambda a: a > 0).values, levels=lev_inc, extend="both", ax=ax, cmap=CMAP_INC, alpha=INC_ALPHA)
        # pos_inc = ds['THM'].isel(Time=0)[0].where(lambda a: a > 0)
        # neg_inc = ds['THM'].isel(Time=0)[0].where(lambda a: a < -0.00028)
        # cs_inc1 = m.contour(x, y, pos_inc.values, extend="both", ax=ax, cmap="Wistia")
        # cs_inc2 = m.contour(x, y, neg_inc.values, extend="both", ax=ax, cmap="cool")

        if PLOT_WIND:
            m.quiver(
                g["lon"][::WIND_SKIP, ::WIND_SKIP],
                g["lat"][::WIND_SKIP, ::WIND_SKIP],
                g["u"][::WIND_SKIP, ::WIND_SKIP],
                g["v"][::WIND_SKIP, ::WIND_SKIP],
                latlon=True,
                ax=ax,
                scale=QUIVER_SCALE,
                width=QUIVER_WIDTH,
                color="k",
                zorder=10,
            )

        # OBS overlay (only radiosonde temp near ZAGL_M)
        if PLOT_OBS and (lon_obs is not None) and (lon_obs.size > 0):
            m.scatter(
                lon_obs, lat_obs,
                c=obs_val,
                cmap=OBS_CMAP,
                s=OBS_SIZE,
                marker=OBS_MARKER,
                alpha=OBS_ALPHA,
                edgecolor="k",
                linewidth=0.4,
                latlon=True,
                ax=ax,
                zorder=30,
            )

        ax.set_title(f"{dom}: {var} @ {ZAGL_M:.0f} m AGL", fontsize=11)

        # default colorbars: left=increments, right=background
        # cb_inc = fig.colorbar(cs_inc, ax=ax, location="left", pad=0.02, fraction=0.046)
        # cb_inc.set_label(f"Δ{var}", fontsize=8)
        # cb_inc.ax.yaxis.set_major_locator(MaxNLocator(4))
        # cb_inc.ax.yaxis.set_major_formatter(ScalarFormatter(useOffset=False))

        cb_bg = fig.colorbar(cs_bg, ax=ax, location="right", pad=0.02, fraction=0.046)
        cb_bg.set_label(f"{var}", fontsize=8)
        cb_bg.ax.yaxis.set_major_locator(MaxNLocator(4))
        cb_bg.ax.yaxis.set_major_formatter(ScalarFormatter(useOffset=False))

fig.suptitle(
    f"Impact at {ZAGL_M:.0f} m AGL (obs: {OBS_TYPE_NAME} within ±{OBS_DZ_M:.0f} m) | cycle={Path(CYCLE_DIR).name}",
    fontsize=12
)
# plt.savefig("fig/increments_temp-eakf.jpeg")
```

```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from pathlib import Path

axes = []

fig = plt.figure(figsize=(4*ncols, 4*nrows), constrained_layout=True)
copy = 'totalspread'
markersize = 6
dom = 'd01'
var = "T2"

# ----------------------------
# Pass 1: global range for consistent background colorbar
# ----------------------------
bg_min = np.inf
bg_max = -np.inf
for cdir in cycle_dirs:
    post_file = Path(cdir) / f"postassim_mean_{dom}.nc"
    with xr.open_dataset(post_file) as ds_post_tmp:
        arr = ds_post_tmp[var].isel(Time=0).values
        bg_min = min(bg_min, np.nanmin(arr))
        bg_max = max(bg_max, np.nanmax(arr))

levels_bg = np.linspace(bg_min, bg_max, 21)

# ----------------------------
# Pass 2: plotting
# ----------------------------
        
# --- marker cycle (extend if needed) ---
markers = ['o', '*', '^', 'v', 'D', 'P', 'X', 's', '<', '>', 'h', '8']
all_types_seen = set()
obs_type_name_map = {}   # type_id -> label
for i, cdir in enumerate(cycle_dirs):
    inc_file = Path(cdir) / f"analysis_increFment_{dom}.nc"
    post_file = Path(cdir) / f"postassim_mean_{dom}.nc"
    obs_file = Path(cdir) / f"obs_epoch_001.nc"    
    x, y, m = setup_basemap_from_file(post_file)
    ax = fig.add_subplot(nrows, ncols, i + 1)
    axes.append(ax)

    m.drawcoastlines(ax=ax, linewidth=0.5)
    m.drawstates(ax=ax, linewidth=0.3)
    m.drawcountries(ax=ax, linewidth=0.3)

    # Background field (consistent levels across all panels)
    with xr.open_dataset(post_file) as ds_post:
        ds2d = ds_post[var].isel(Time=0) #* 1000
        cs = m.contourf(
            x, y, ds2d.values,
            levels=levels_bg,
            extend="both",
            ax=ax,
            cmap="Greens"
        )

    # Increments + winds
    with xr.open_dataset(inc_file) as ds:
        pos_inc = ds[var].isel(Time=0).where(lambda a: a > 0) * 1000
        neg_inc = ds[var].isel(Time=0).where(lambda a: a < -0.00028) * 1000

        h2 = m.contour(
            x, y, pos_inc.values,
            extend="both",
            ax=ax,
            cmap="Wistia"
        )
        h3 = m.contour(
            x, y, neg_inc.values,
            extend="both",
            ax=ax,
            cmap="cool"
        )

        u = ds["U10"].isel(Time=0)
        v = ds["V10"].isel(Time=0)
        lat = ds["XLAT"]
        lon = ds["XLONG"]

        skip = 5
        q = m.quiver(
            lon.values[::skip, ::skip],
            lat.values[::skip, ::skip],
            u.values[::skip, ::skip],
            v.values[::skip, ::skip],
            scale=40,
            width=0.002,
            ax=ax,
            latlon=True,
        )


        # --- open file ---
        ds = xr.open_dataset(obs_file)
        
        # --- choose which observation copy / qc copy to plot ---
        obs_copy_idx = 0
        qc_copy_idx = 1
        
        # --- extract core fields ---
        lon = ds["location"].isel(locdim=0).values
        lat = ds["location"].isel(locdim=1).values
        z   = ds["location"].isel(locdim=2).values
        wvt = ds["which_vert"].values
        obs = ds["observations"].isel(copy=obs_copy_idx).values
        qc  = ds["qc"].isel(qc_copy=qc_copy_idx).values if "qc" in ds else None
        otype = ds["obs_type"].values  # <-- add this
        
        # Detect radians (heuristic): radians typically within [-2π, 2π]
        if np.nanmax(np.abs(lon)) <= 2*np.pi + 0.1 and np.nanmax(np.abs(lat)) <= np.pi/2 + 0.1:
            lon = np.degrees(lon)
            lat = np.degrees(lat)
        
        # Normalize lon to [-180, 180) to match typical WRF XLONG
        lon = ((lon + 180.0) % 360.0) - 180.0

        # --- near-surface mask ---
        # near_sfc = (wvt == -1) | ((wvt == 3) & np.isfinite(z) & (z <= 20.0))
        near_sfc = (
            (wvt == -1) |
            ((wvt == 3) & np.isfinite(z) & (z <= 20.0)) |
            ((wvt == 2) & np.isfinite(z) & (z >= 900.0))   # Pa; tune (e.g., 95000)
        )
        
        maxQC = 2
        if qc is not None:
            near_sfc &= np.isfinite(qc) & (qc <= maxQC)
        
        near_sfc &= np.isfinite(lon) & np.isfinite(lat) & np.isfinite(obs)
    # in_box = (
    #     (lon >= lon_min) & (lon <= lon_max) &
    #     (lat >= lat_min) & (lat <= lat_max)
    # )
    # near_sfc &= in_box
        
        # --- subset ---
        lon_s   = lon[near_sfc]
        lat_s   = lat[near_sfc]
        obs_s   = obs[near_sfc]
        z_s     = z[near_sfc]
        wvt_s   = wvt[near_sfc]
        otype_s = otype[near_sfc]  # <-- add this
        
        # --- optional: obs type id -> name mapping ---
        # DART obs_type values are typically 1-based ids matching ObsTypesMetaData
        obs_type_names = ds["ObsTypesMetaData"].values if "ObsTypesMetaData" in ds else None
        
        def obs_type_label(type_id):
            if obs_type_names is None:
                return f"type {int(type_id)}"
            i = int(type_id) - 1  # DART ids are usually 1-based
            if 0 <= i < len(obs_type_names):
                raw = obs_type_names[i]
                return raw.decode().strip() if isinstance(raw, (bytes, np.bytes_)) else str(raw).strip()
            return f"type {int(type_id)}"


        
        # --- plot one scatter per obs type ---
        sc_last = None
        unique_types = np.unique(otype_s)
        all_types_seen.update(unique_types.tolist())
        for t_id in unique_types:
            obs_type_name_map[int(t_id)] = obs_type_label(t_id)
        
        for k, t_id in enumerate(unique_types):
            msk = (otype_s == t_id)
            mk = markers[k % len(markers)]
        
            sc_last = m.scatter(
                lon_s[msk], lat_s[msk],
                c=obs_s[msk],
                s=60,
                cmap="bwr",
                alpha=0.85,
                marker=mk,
                label=obs_type_label(t_id),
                latlon=True,
                ax=ax,
                zorder=20, 
            )

legend_handles = [
    Line2D([0], [0], color="k", marker=r"$\rightarrow$", markersize=12, lw=0,
           linestyle="None", label="10 m wind increment"),
    Line2D([0], [0], color="goldenrod", lw=2, label="Positive increments"),
    Line2D([0], [0], color="deepskyblue", lw=2, label="Negative increments"),
]

# Add marker-only handles for each observation type seen across all panels
obs_marker_handles = []
for k, t_id in enumerate(sorted(all_types_seen)):
    mk = markers[k % len(markers)]
    obs_marker_handles.append(
        Line2D([0], [0],
               marker=mk,
               linestyle="None",
               color="k",
               markerfacecolor="none",
               markeredgecolor="k",
               markersize=7,
               label=obs_type_name_map.get(int(t_id), f"type {int(t_id)}"))
    )

legend_handles_all = legend_handles + obs_marker_handles

fig.legend(
    handles=legend_handles_all,
    bbox_to_anchor=(0.72, 0.-0.2),
    ncol=3,
    frameon=True
)

cax = fig.add_axes([0.25, 0.08, 0.5, 0.028])  # make it a bit wider
cbar = fig.colorbar(cs, cax=cax, orientation="horizontal")

cax2 = fig.add_axes([0.25, -0.1, 0.5, 0.028])  # make it a bit wider
cbar2 = fig.colorbar(sc_last, cax=cax2, orientation="horizontal", extend='both',label='observation Values')

from matplotlib.ticker import MaxNLocator, ScalarFormatter
cbar.ax.xaxis.set_major_locator(MaxNLocator(3))

fmt = ScalarFormatter(useOffset=False)
fmt.set_scientific(False)   
cbar.ax.xaxis.set_major_formatter(fmt)

cbar.ax.tick_params(axis="x", labelsize=8, pad=1)
cbar.update_ticks()         
cbar.set_label(f"Background {var}")
```

```python
import pandas as pd
import plotly.express as px
import pydartdiags.obs_sequence.obs_sequence as obsq

# 1. Define your file paths
file_paths = [
    '/gpfs/research/scratch/sa24m/tqprof/run2/osse_out/sys_obs/window_2015071406_2015071524/ascii_to_obs/obs_seq2015071406',
    '/gpfs/research/scratch/sa24m/tqprof/run2/osse_out/sys_obs/window_2015071406_2015071524/ascii_to_obs/obs_seq2015071412',
    '/gpfs/research/scratch/sa24m/tqprof/run2/osse_out/sys_obs/window_2015071406_2015071524/ascii_to_obs/obs_seq2015071418',
    '/gpfs/research/scratch/sa24m/tqprof/run2/osse_out/dart_cycle/output/2015071412/obs_seq.final',
]

dfs = []

# 2. Loop through files with Error Handling
for file_path in file_paths:
    file_name = file_path.split('/')[-1]
    
    try:
        # Attempt to load the sequence
        obs_seq = obsq.ObsSequence(file_path)
        
        # We copy the ENTIRE dataframe here instead of filtering it
        d = obs_seq.df.copy()
        
        # Check if the dataframe is empty
        if d.empty:
            print(f"WARNING: {file_name} loaded, but is completely empty.")
            continue

        # Add the filename column for Plotly subplots
        d['source_file'] = file_name
        dfs.append(d)
        print(f"SUCCESS: {file_name} loaded.")
        
    except ValueError as e:
        print(f"FAILED: {file_name} could not be parsed. Error: {e}")
    except Exception as e:
        print(f"FAILED: Unexpected error with {file_name}. Error: {e}")

# 3. Combine and Plot (if we have valid data)
if dfs:
    combined_df = pd.concat(dfs, ignore_index=True)

    # Create the geographic plot with subplots
    fig = px.scatter_geo(
        combined_df,
        lat='latitude',   
        lon='longitude',  
        color='type',             # NEW: Colors the dots by observation type
        hover_data={
            'type': False,        # Turned off in hover since it's now in the legend
            'observation': True,
            'obs_num': True,
            'vertical': True      # Added vertical coordinate to hover info!
        },
        facet_col='source_file', 
        facet_col_wrap=2,        
        title='Geographic Plot of ALL Observations by Time Window'
    )

    # Update layout
    fig.update_layout(
        width=1100,  # Increased width slightly to make room for the new legend
        height=800  
    )

    # Clean up subplot titles to remove the "source_file=" prefix
    fig.for_each_annotation(lambda a: a.update(text=a.text.split("=")[-1]))

    # Display the plot
    fig.show()
else:
    print("CRITICAL: No files were successfully loaded. Cannot create the plot.")
```
