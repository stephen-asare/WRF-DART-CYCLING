```python
import pandas as pd
import plotly.express as px
import pydartdiags.obs_sequence.obs_sequence as obsq

# 1. Define your file paths
file_paths = [
    '/gpfs/home/sa24m/scratch/tqprof/run2/osse_out/dart_cycle/output/2015071418/obs_seq.final'
]

dfs = []

# 2. Loop through files with Error Handling
for file_path in file_paths:
    file_name = file_path.split('/')[-1]
    
    try:
        # Attempt to load the sequence
        obs_seq = obsq.ObsSequence(file_path)
        
        # We copy the ENTIRE dataframe here instead of filtering it
        d = obs_seq.df.copy()
        
        # Check if the dataframe is empty
        if d.empty:
            print(f"WARNING: {file_name} loaded, but is completely empty.")
            continue

        # Add the filename column for Plotly subplots
        d['source_file'] = file_name
        dfs.append(d)
        print(f"SUCCESS: {file_name} loaded.")
        
    except ValueError as e:
        print(f"FAILED: {file_name} could not be parsed. Error: {e}")
    except Exception as e:
        print(f"FAILED: Unexpected error with {file_name}. Error: {e}")

# 3. Combine and Plot (if we have valid data)
if dfs:
    combined_df = pd.concat(dfs, ignore_index=True)

    # Create the geographic plot with subplots
    fig = px.scatter_geo(
        combined_df,
        lat='latitude',   
        lon='longitude',  
        color='type',             # NEW: Colors the dots by observation type
        hover_data={
            'type': False,        # Turned off in hover since it's now in the legend
            'observation': True,
            'obs_num': True,
            'vertical': True,      # Added vertical coordinate to hover info!
            'DART_quality_control':True
        },
        facet_col='source_file', 
        facet_col_wrap=2,        
        title='Geographic Plot of ALL Observations by Time Window'
    )

    # Update layout
    fig.update_layout(
        width=1100,  # Increased width slightly to make room for the new legend
        height=800  
    )

    # Clean up subplot titles to remove the "source_file=" prefix
    fig.for_each_annotation(lambda a: a.update(text=a.text.split("=")[-1]))

    # Display the plot
    fig.show()
else:
    print("CRITICAL: No files were successfully loaded. Cannot create the plot.")
```

```python
#!/usr/bin/env python3
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
from pathlib import Path
from matplotlib.ticker import MaxNLocator, ScalarFormatter
from mpl_toolkits.basemap import Basemap

# Import the pydartdiags module for reading obs_seq.final files
import pydartdiags.obs_sequence.obs_sequence as obsq

# =============================================================================
# USER OPTIONS
# =============================================================================
# CYCLE_DIR   = "/gpfs/home/sa24m/scratch/tqprof/run2/osse_out/dart_cycle/output/2015071421/"
CYCLE_DIR   = "/gpfs/home/sa24m/stephen_asare/runs//output/2015071412/"

TIME_INDEX  = 0

# VERTICAL LEVEL SETTINGS
PLOT_LOWEST_LEVEL = True    # Set to True to plot lowest model layer (k=0) to surface
ZAGL_M            = 3000.0  # Ignored if PLOT_LOWEST_LEVEL is True

DOMS        = ["d01", "d02"]                      # row 0 = d01, row 1 = d02
VARS_COLS   = ["THM", "QVAPOR", "WSPD"]           # col 0..2 in each row

INC_SOURCE  = "file"                              # "file" (analysis_increment) or "diff" (post-pre)

# winds (background wind vectors overlaid on every panel)
PLOT_WIND     = True
WIND_U        = "U"
WIND_V        = "V"
WIND_SKIP     = 9
QUIVER_SCALE  = 150
QUIVER_WIDTH  = 0.002

# plotting style
BG_LEVELS    = 21
INC_LEVELS   = 17
INC_ALPHA    = 0.55
INC_THRESH   = None   # e.g. 1e-4, or None
CMAP_BG      = "Greens"
CMAP_INC     = "bwr"

# PYDARTDIAGS OBS OVERLAY OPTIONS
PLOT_OBS           = True
OBS_SEQ_FILE       = CYCLE_DIR+"obs_seq.final"
OBS_TIME_START     = "2015-07-14 11:30:00" 
OBS_TIME_END       = "2015-07-14 12:30:00" 
OBS_TYPES          = ['RADIOSONDE_TEMPERATURE']
OBS_VALUE_COL      = "observation"         
OBS_MARKER         = "o"
OBS_SIZE           = 35
OBS_ALPHA          = 0.9
OBS_CMAP           = "bwr"

# =============================================================================
# WRF BASEMAP HELPER
# =============================================================================
def setup_basemap_from_file(nc_path):
    ds = xr.open_dataset(nc_path)
    lat = ds['XLAT'].values[0] if 'Time' in ds['XLAT'].dims else ds['XLAT'].values
    lon = ds['XLONG'].values[0] if 'Time' in ds['XLONG'].dims else ds['XLONG'].values
    
    truelat1 = getattr(ds, 'TRUELAT1', lat.mean())
    truelat2 = getattr(ds, 'TRUELAT2', lat.mean())
    stand_lon = getattr(ds, 'STAND_LON', lon.mean())
    
    m = Basemap(
        projection='lcc', resolution='i',
        llcrnrlon=lon[0, 0], llcrnrlat=lat[0, 0],
        urcrnrlon=lon[-1, -1], urcrnrlat=lat[-1, -1],
        lat_1=truelat1, lat_2=truelat2, lon_0=stand_lon
    )
    x, y = m(lon, lat)
    ds.close()
    return x, y, m

# =============================================================================
# Helpers
# =============================================================================
def safe_levels(vmin, vmax, n, pad_frac=1e-6, pad_abs=1e-12):
    vmin = float(vmin); vmax = float(vmax)
    if not np.isfinite(vmin) or not np.isfinite(vmax):
        return np.linspace(0.0, 1.0, n)
    if vmax <= vmin:
        pad = max(pad_abs, pad_frac * (abs(vmin) + 1.0))
        vmin -= pad
        vmax += pad
    return np.linspace(vmin, vmax, n)
def _destagger_to_mass(ds, var, time_index=0):
    da = ds[var].isel(Time=time_index) if "Time" in ds[var].dims else ds[var]
    if var == "U" and "west_east_stag" in da.dims:
        return 0.5 * (da.isel(west_east_stag=slice(0, -1)) + da.isel(west_east_stag=slice(1, None))).rename({"west_east_stag":"west_east"})
    if var == "V" and "south_north_stag" in da.dims:
        return 0.5 * (da.isel(south_north_stag=slice(0, -1)) + da.isel(south_north_stag=slice(1, None))).rename({"south_north_stag":"south_north"})
    if var == "W" and "bottom_top_stag" in da.dims:
        return 0.5 * (da.isel(bottom_top_stag=slice(0, -1)) + da.isel(bottom_top_stag=slice(1, None))).rename({"bottom_top_stag":"bottom_top"})
    return da

def _k_at_zagl(ds_ref, zagl_m, time_index=0):
    Zstag = (ds_ref["PH"].isel(Time=time_index).values + ds_ref["PHB"].values) / 9.81
    Zmass = 0.5 * (Zstag[:-1, :, :] + Zstag[1:, :, :])
    HGT   = ds_ref["HGT"].values
    ZAGL  = Zmass - HGT[None, :, :]
    return np.argmin(np.abs(ZAGL - zagl_m), axis=0)

def field_at_zagl(ds_field, ds_ref, var, zagl_m, time_index=0, lowest_level=False):
    # 1. ALWAYS destagger the variable first
    da = _destagger_to_mass(ds_field, var, time_index=time_index)
    
    # 2. Check if it's a purely 2D surface field (e.g., T2, TSK)
    if "bottom_top" not in da.dims:
        return da.values

    # 3. Extract the 3D values
    v3d = da.values
    
    # 4. Grab the lowest layer directly if requested
    if lowest_level:
        return v3d[0, :, :]

    # 5. Interpolate to geopotential height
    k = _k_at_zagl(ds_ref, zagl_m=zagl_m, time_index=time_index)
    j, i = np.indices(k.shape)
    return v3d[k, j, i]
# =============================================================================
# Load fields for both domains
# =============================================================================
cdir = Path(CYCLE_DIR)
DATA = {}   
GEO  = {}   

for dom in DOMS:
    post_file = cdir / f"postassim_mean_{dom}.nc"
    pre_file  = cdir / f"preassim_mean_{dom}.nc"
    inc_file  = cdir / f"analysis_increment_{dom}.nc"

    if not post_file.exists(): raise FileNotFoundError(f"Missing: {post_file}")
    if INC_SOURCE == "diff" and (not pre_file.exists()): raise FileNotFoundError(f"Missing: {pre_file}")
    if INC_SOURCE == "file" and (not inc_file.exists()): raise FileNotFoundError(f"Missing: {inc_file}")

    x, y, m = setup_basemap_from_file(str(post_file))

    ds_post = xr.open_dataset(post_file)
    ds_pre  = xr.open_dataset(pre_file) if (INC_SOURCE == "diff") else None
    ds_inc  = xr.open_dataset(inc_file) if (INC_SOURCE == "file") else None

    lon2d = ds_post["XLONG"].values
    lat2d = ds_post["XLAT"].values

    if PLOT_WIND:
        u2d = field_at_zagl(ds_post, ds_post, WIND_U, ZAGL_M, time_index=TIME_INDEX, lowest_level=PLOT_LOWEST_LEVEL)
        v2d = field_at_zagl(ds_post, ds_post, WIND_V, ZAGL_M, time_index=TIME_INDEX, lowest_level=PLOT_LOWEST_LEVEL)
    else:
        u2d = v2d = None

    GEO[dom] = dict(x=x, y=y, m=m, lon=lon2d, lat=lat2d, u=u2d, v=v2d, post_file=str(post_file))

    for var in VARS_COLS:
        # Dynamic Wind Speed Calculation
        if var == "WSPD":
            u_post = field_at_zagl(ds_post, ds_post, "U", ZAGL_M, time_index=TIME_INDEX, lowest_level=PLOT_LOWEST_LEVEL)
            v_post = field_at_zagl(ds_post, ds_post, "V", ZAGL_M, time_index=TIME_INDEX, lowest_level=PLOT_LOWEST_LEVEL)
            bg2d = np.sqrt(u_post**2 + v_post**2)
            
            if INC_SOURCE == "file":
                u_inc = field_at_zagl(ds_inc, ds_post, "U", ZAGL_M, time_index=TIME_INDEX, lowest_level=PLOT_LOWEST_LEVEL)
                v_inc = field_at_zagl(ds_inc, ds_post, "V", ZAGL_M, time_index=TIME_INDEX, lowest_level=PLOT_LOWEST_LEVEL)
                u_pre = u_post - u_inc
                v_pre = v_post - v_inc
                pre2d = np.sqrt(u_pre**2 + v_pre**2)
                inc2d = bg2d - pre2d
            else:
                u_pre = field_at_zagl(ds_pre, ds_post, "U", ZAGL_M, time_index=TIME_INDEX, lowest_level=PLOT_LOWEST_LEVEL)
                v_pre = field_at_zagl(ds_pre, ds_post, "V", ZAGL_M, time_index=TIME_INDEX, lowest_level=PLOT_LOWEST_LEVEL)
                pre2d = np.sqrt(u_pre**2 + v_pre**2)
                inc2d = bg2d - pre2d
                
        else:
            bg2d = field_at_zagl(ds_post, ds_post, var, ZAGL_M, time_index=TIME_INDEX, lowest_level=PLOT_LOWEST_LEVEL)
            if INC_SOURCE == "file":
                inc2d = field_at_zagl(ds_inc, ds_post, var, ZAGL_M, time_index=TIME_INDEX, lowest_level=PLOT_LOWEST_LEVEL)
            else:
                pre2d = field_at_zagl(ds_pre, ds_post, var, ZAGL_M, time_index=TIME_INDEX, lowest_level=PLOT_LOWEST_LEVEL)
                inc2d = bg2d - pre2d

        if INC_THRESH is not None:
            inc2d = np.where(np.abs(inc2d) >= INC_THRESH, inc2d, np.nan)

        DATA[(dom, var)] = dict(bg=bg2d, inc=inc2d)

    if ds_inc is not None: ds_inc.close()
    if ds_pre is not None: ds_pre.close()
    ds_post.close()

# =============================================================================
# Shared levels per COLUMN across d01/d02
# =============================================================================
LEVELS = {}
for var in VARS_COLS:
    bmins, bmaxs = [], []
    for dom in DOMS:
        b = DATA[(dom, var)]["bg"]
        bmins.append(np.nanmin(b))
        bmaxs.append(np.nanmax(b))
    levels_bg = safe_levels(np.nanmin(bmins), np.nanmax(bmaxs), BG_LEVELS)

    absmaxs = []
    for dom in DOMS:
        inc = DATA[(dom, var)]["inc"]
        absmaxs.append(np.nanmax(np.abs(inc)))
    inc_absmax = float(np.nanmax(absmaxs))
    if not np.isfinite(inc_absmax) or inc_absmax == 0.0:
        inc_absmax = 1e-12
    levels_inc = np.linspace(-inc_absmax, inc_absmax, INC_LEVELS)

    LEVELS[var] = dict(levels_bg=levels_bg, levels_inc=levels_inc)

# =============================================================================
# Process pydartdiags Observations (Done ONCE before plotting)
# =============================================================================
lon_obs = lat_obs = obs_val = None

if PLOT_OBS:
    try:
        obs_seq = obsq.ObsSequence(OBS_SEQ_FILE)
        df = obs_seq.df.copy()
        
        if df.empty:
            print(f"WARNING: {Path(OBS_SEQ_FILE).name} loaded, but is completely empty.")
        else:
            df["time"] = pd.to_datetime(df["time"])
            
            t0 = pd.Timestamp(OBS_TIME_START)
            t1 = pd.Timestamp(OBS_TIME_END)
            
            df_filtered = df[
                (df["time"] >= t0) &
                (df["time"] <= t1) &
                (df["type"].isin(OBS_TYPES))
            ].copy()
            
            df_filtered["lon_plot"] = ((df_filtered["longitude"] + 180) % 360) - 180
            
            lon_obs = df_filtered["lon_plot"].values
            lat_obs = df_filtered["latitude"].values
            obs_val = df_filtered[OBS_VALUE_COL].values
            
            print(f"SUCCESS: Loaded {len(df_filtered)} observations from {Path(OBS_SEQ_FILE).name} for overlay.")
    except Exception as e:
        print(f"FAILED: Could not process obs_seq file. Error: {e}")

# =============================================================================
# Plot 2 rows x 3 columns + OBS overlay
# =============================================================================
# Widened figsize from 13.5 to 18.5 to accommodate the 3rd column
fig, axs = plt.subplots(2, 3, figsize=(18.5, 8.5), constrained_layout=False)
plt.subplots_adjust(left=0.04, right=0.98, bottom=0.06, top=0.92, wspace=0.20, hspace=0.14)

for r, dom in enumerate(DOMS):
    g = GEO[dom]
    m = g["m"]
    x = g["x"]; y = g["y"]

    for c, var in enumerate(VARS_COLS):
        ax = axs[r, c]

        m.drawcoastlines(ax=ax, linewidth=0.5)
        m.drawstates(ax=ax, linewidth=0.3)
        m.drawcountries(ax=ax, linewidth=0.3)

        bg2d  = DATA[(dom, var)]["bg"]
        inc2d = DATA[(dom, var)]["inc"]
        lev_bg  = LEVELS[var]["levels_bg"]

        cs_bg  = m.contourf(x, y, bg2d,  levels=lev_bg,  extend="both", ax=ax, cmap=CMAP_BG)
        
        pos_inc = np.where(inc2d > 0, inc2d, np.nan)
        neg_inc = np.where(inc2d < 0, inc2d, np.nan)
        inc_max = float(np.nanmax(inc2d))
        inc_min = float(np.nanmin(inc2d))

        # Safe contour plotting that prevents the zero-increment crash
        if inc_max > 1e-10:
            pos_levels = np.linspace(0, inc_max, INC_LEVELS)
            m.contour(x, y, pos_inc, levels=pos_levels, ax=ax, cmap="Wistia", alpha=INC_ALPHA)
        
        if inc_min < -1e-10:
            neg_levels = np.linspace(inc_min, 0, INC_LEVELS)
            m.contour(x, y, neg_inc, levels=neg_levels, ax=ax, cmap="cool", alpha=INC_ALPHA)

        if PLOT_WIND:
            m.quiver(
                g["lon"][::WIND_SKIP, ::WIND_SKIP],
                g["lat"][::WIND_SKIP, ::WIND_SKIP],
                g["u"][::WIND_SKIP, ::WIND_SKIP],
                g["v"][::WIND_SKIP, ::WIND_SKIP],
                latlon=True,
                ax=ax,
                scale=QUIVER_SCALE,
                width=QUIVER_WIDTH,
                color="k",
                zorder=10,
            )

        if PLOT_OBS and (lon_obs is not None) and (lon_obs.size > 0):
            m.scatter(
                lon_obs, lat_obs,
                c=obs_val,
                cmap=OBS_CMAP,
                s=OBS_SIZE,
                marker=OBS_MARKER,
                alpha=OBS_ALPHA,
                edgecolor="k",
                linewidth=0.4,
                latlon=True,
                ax=ax,
                zorder=30,
            )

        # Dynamic Title Logic based on PLOT_LOWEST_LEVEL
        if PLOT_LOWEST_LEVEL:
            lvl_str = "Lowest Level (k=0)"
        else:
            lvl_str = f"{ZAGL_M:.0f} m AGL"
            
        ax.set_title(f"{dom}: {var}", fontsize=11)

        cb_bg = fig.colorbar(cs_bg, ax=ax, location="right", pad=0.02, fraction=0.046)
        cb_bg.set_label(f"{var}", fontsize=8)
        cb_bg.ax.yaxis.set_major_locator(MaxNLocator(4))
        cb_bg.ax.yaxis.set_major_formatter(ScalarFormatter(useOffset=False))

fig.suptitle(
    f"{OBS_TYPES[0]} | cycle={Path(CYCLE_DIR).name}",
    fontsize=12
)
plt.savefig(f"fig/increments_{CYCLE_DIR[36:-27]}_{CYCLE_DIR[49:-19]}_{CYCLE_DIR[-3:-1]}.jpeg")
plt.show()
```

```python
#!/usr/bin/env python3
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
from pathlib import Path
from matplotlib.ticker import MaxNLocator, ScalarFormatter
from mpl_toolkits.basemap import Basemap

import pydartdiags.obs_sequence.obs_sequence as obsq

# =============================================================================
# USER OPTIONS
# =============================================================================
CYCLE_DIR   = "/gpfs/home/sa24m/stephen_asare/runs/temp_only_18_eakf/output/2015071412/"
# CYCLE_DIR = '/gpfs/home/sa24m/scratch/tqprof/run2/osse_out/dart_cycle/output/2015071412/'

NATURE_FILES = {
    "d01": f"/gpfs/research/scratch/sa24m/tqprof/run2/osse_out/ens_wrf/e002/wrfout_d01_2015-07-14_{CYCLE_DIR[-3:-1]}:00:00",
    "d02": f"/gpfs/research/scratch/sa24m/tqprof/run2/osse_out/ens_wrf/e002/wrfout_d02_2015-07-14_{CYCLE_DIR[-3:-1]}:00:00",
}

TIME_INDEX = 0

# VERTICAL LEVEL SETTINGS
PLOT_LOWEST_LEVEL = True
ZAGL_M = 3000.0

DOMS = ["d01", "d02"]
VARS_COLS = ["THM", "QVAPOR", "WSPD"]

# plotting style
DIFF_LEVELS = 21
CMAP_DIFF = "bwr"

# wind overlay from posterior
PLOT_WIND = False
WIND_U = "U"
WIND_V = "V"
WIND_SKIP = 9
QUIVER_SCALE = 150
QUIVER_WIDTH = 0.002

# observation overlay
PLOT_OBS = True
OBS_SEQ_FILE = CYCLE_DIR + "obs_seq.final"
# OBS_TIME_START = "2015-07-14 20:30:00"
# OBS_TIME_END   = "2015-07-14 21:30:00"
OBS_TIME_START = f"2015-07-14 {int(CYCLE_DIR[-3:-1]) - 1:02d}:30:00"
OBS_TIME_END   = f"2015-07-14 {int(CYCLE_DIR[-3:-1]):02d}:30:00"
OBS_TYPES = ['RADIOSONDE_TEMPERATURE'] # ['RADIOSONDE_SPECIFIC_HUMIDITY'] #['RADIOSONDE_TEMPERATURE'] #["RADIOSONDE_U_WIND_COMPONENT"]

# location-only obs markers
OBS_MARKER = "o"
OBS_SIZE = 35
OBS_ALPHA = 0.9
OBS_EDGE_COLOR = "k"
OBS_FACE_COLOR = "none"

# output
OUTDIR = Path("fig")
OUTDIR.mkdir(exist_ok=True)
OUTFIG = OUTDIR / f"posterior_minus_nature_{CYCLE_DIR[36:-27]}_{CYCLE_DIR[49:-19]}_{CYCLE_DIR[-3:-1]}.jpeg"

# =============================================================================
# WRF BASEMAP HELPER
# =============================================================================
def setup_basemap_from_file(nc_path):
    ds = xr.open_dataset(nc_path)

    lat = ds["XLAT"].values[0] if "Time" in ds["XLAT"].dims else ds["XLAT"].values
    lon = ds["XLONG"].values[0] if "Time" in ds["XLONG"].dims else ds["XLONG"].values

    truelat1 = getattr(ds, "TRUELAT1", lat.mean())
    truelat2 = getattr(ds, "TRUELAT2", lat.mean())
    stand_lon = getattr(ds, "STAND_LON", lon.mean())

    m = Basemap(
        projection="lcc",
        resolution="i",
        llcrnrlon=lon[0, 0],
        llcrnrlat=lat[0, 0],
        urcrnrlon=lon[-1, -1],
        urcrnrlat=lat[-1, -1],
        lat_1=truelat1,
        lat_2=truelat2,
        lon_0=stand_lon,
    )

    x, y = m(lon, lat)
    ds.close()

    return x, y, m


# =============================================================================
# Helpers
# =============================================================================
def _destagger_to_mass(ds, var, time_index=0):
    da = ds[var].isel(Time=time_index) if "Time" in ds[var].dims else ds[var]

    if var == "U" and "west_east_stag" in da.dims:
        return 0.5 * (
            da.isel(west_east_stag=slice(0, -1))
            + da.isel(west_east_stag=slice(1, None))
        ).rename({"west_east_stag": "west_east"})

    if var == "V" and "south_north_stag" in da.dims:
        return 0.5 * (
            da.isel(south_north_stag=slice(0, -1))
            + da.isel(south_north_stag=slice(1, None))
        ).rename({"south_north_stag": "south_north"})

    if var == "W" and "bottom_top_stag" in da.dims:
        return 0.5 * (
            da.isel(bottom_top_stag=slice(0, -1))
            + da.isel(bottom_top_stag=slice(1, None))
        ).rename({"bottom_top_stag": "bottom_top"})

    return da


def _k_at_zagl(ds_ref, zagl_m, time_index=0):
    PH = ds_ref["PH"].isel(Time=time_index).values if "Time" in ds_ref["PH"].dims else ds_ref["PH"].values
    PHB = ds_ref["PHB"].isel(Time=time_index).values if "Time" in ds_ref["PHB"].dims else ds_ref["PHB"].values

    Zstag = (PH + PHB) / 9.81
    Zmass = 0.5 * (Zstag[:-1, :, :] + Zstag[1:, :, :])

    HGT = ds_ref["HGT"].isel(Time=time_index).values if "Time" in ds_ref["HGT"].dims else ds_ref["HGT"].values

    ZAGL = Zmass - HGT[None, :, :]
    return np.argmin(np.abs(ZAGL - zagl_m), axis=0)


def field_at_zagl(ds_field, ds_ref, var, zagl_m, time_index=0, lowest_level=False):
    da = _destagger_to_mass(ds_field, var, time_index=time_index)

    if "bottom_top" not in da.dims:
        return da.values

    v3d = da.values

    if lowest_level:
        return v3d[0, :, :]

    k = _k_at_zagl(ds_ref, zagl_m=zagl_m, time_index=time_index)
    j, i = np.indices(k.shape)

    return v3d[k, j, i]


def get_field_for_comparison(ds, ds_ref, var, zagl_m, time_index=0, lowest_level=False):
    """
    Returns comparable 2D fields.

    For THM:
    - use THM directly if present
    - otherwise use T + 300 from standard wrfout
    """
    if var == "WSPD":
        u = field_at_zagl(ds, ds_ref, "U", zagl_m, time_index=time_index, lowest_level=lowest_level)
        v = field_at_zagl(ds, ds_ref, "V", zagl_m, time_index=time_index, lowest_level=lowest_level)
        return np.sqrt(u**2 + v**2)

    if var == "THM":
        if "THM" in ds.variables:
            return field_at_zagl(ds, ds_ref, "THM", zagl_m, time_index=time_index, lowest_level=lowest_level)

        elif "T" in ds.variables:
            return field_at_zagl(ds, ds_ref, "T", zagl_m, time_index=time_index, lowest_level=lowest_level) + 300.0

        else:
            raise KeyError("Neither THM nor T exists in this file. Cannot compute THM comparison.")

    return field_at_zagl(ds, ds_ref, var, zagl_m, time_index=time_index, lowest_level=lowest_level)


def symmetric_levels(arrays, nlevels):
    absmax = 0.0

    for arr in arrays:
        local = np.nanmax(np.abs(arr))
        if np.isfinite(local):
            absmax = max(absmax, float(local))

    if absmax == 0.0:
        absmax = 1e-12

    return np.linspace(-absmax, absmax, nlevels)


# =============================================================================
# Load posterior - nature fields
# =============================================================================
cdir = Path(CYCLE_DIR)

DATA = {}
GEO = {}

for dom in DOMS:
    post_file = cdir / f"postassim_mean_{dom}.nc"
    nature_file = Path(NATURE_FILES[dom])

    if not post_file.exists():
        raise FileNotFoundError(f"Missing posterior file: {post_file}")

    if not nature_file.exists():
        raise FileNotFoundError(f"Missing nature file: {nature_file}")

    x, y, m = setup_basemap_from_file(str(post_file))

    ds_post = xr.open_dataset(post_file)
    ds_nat = xr.open_dataset(nature_file)

    lon2d = ds_post["XLONG"].values
    lat2d = ds_post["XLAT"].values

    if PLOT_WIND:
        u2d = field_at_zagl(
            ds_post, ds_post, WIND_U, ZAGL_M,
            time_index=TIME_INDEX,
            lowest_level=PLOT_LOWEST_LEVEL,
        )
        v2d = field_at_zagl(
            ds_post, ds_post, WIND_V, ZAGL_M,
            time_index=TIME_INDEX,
            lowest_level=PLOT_LOWEST_LEVEL,
        )
    else:
        u2d = v2d = None

    GEO[dom] = dict(
        x=x,
        y=y,
        m=m,
        lon=lon2d,
        lat=lat2d,
        u=u2d,
        v=v2d,
        post_file=str(post_file),
        nature_file=str(nature_file),
    )

    for var in VARS_COLS:
        post2d = get_field_for_comparison(
            ds_post,
            ds_post,
            var,
            ZAGL_M,
            time_index=TIME_INDEX,
            lowest_level=PLOT_LOWEST_LEVEL,
        )

        nat2d = get_field_for_comparison(
            ds_nat,
            ds_nat,
            var,
            ZAGL_M,
            time_index=TIME_INDEX,
            lowest_level=PLOT_LOWEST_LEVEL,
        )

        diff2d = post2d - nat2d

        DATA[(dom, var)] = dict(
            post=post2d,
            nature=nat2d,
            diff=diff2d,
        )

    ds_post.close()
    ds_nat.close()


# =============================================================================
# Shared difference levels per variable across d01/d02
# =============================================================================
LEVELS = {}

for var in VARS_COLS:
    diff_arrays = [DATA[(dom, var)]["diff"] for dom in DOMS]
    LEVELS[var] = symmetric_levels(diff_arrays, DIFF_LEVELS)


# =============================================================================
# Process observations
# =============================================================================
lon_obs = lat_obs = None

if PLOT_OBS:
    try:
        obs_seq = obsq.ObsSequence(OBS_SEQ_FILE)
        df = obs_seq.df.copy()

        if df.empty:
            print(f"WARNING: {Path(OBS_SEQ_FILE).name} loaded, but is empty.")
        else:
            df["time"] = pd.to_datetime(df["time"])

            t0 = pd.Timestamp(OBS_TIME_START)
            t1 = pd.Timestamp(OBS_TIME_END)

            df_filtered = df[
                (df["time"] >= t0)
                & (df["time"] <= t1)
                & (df["type"].isin(OBS_TYPES))
            ].copy()

            df_filtered["lon_plot"] = ((df_filtered["longitude"] + 180) % 360) - 180

            lon_obs = df_filtered["lon_plot"].values
            lat_obs = df_filtered["latitude"].values

            print(
                f"SUCCESS: Loaded {len(df_filtered)} observations "
                f"from {Path(OBS_SEQ_FILE).name} for overlay."
            )

    except Exception as e:
        print(f"FAILED: Could not process obs_seq file. Error: {e}")


# =============================================================================
# Plot 2 rows x 3 columns
# =============================================================================
fig, axs = plt.subplots(2, 3, figsize=(18.5, 8.5), constrained_layout=False)

plt.subplots_adjust(
    left=0.04,
    right=0.98,
    bottom=0.06,
    top=0.92,
    wspace=0.20,
    hspace=0.14,
)

for r, dom in enumerate(DOMS):
    g = GEO[dom]
    m = g["m"]
    x = g["x"]
    y = g["y"]

    for c, var in enumerate(VARS_COLS):
        ax = axs[r, c]

        m.drawcoastlines(ax=ax, linewidth=0.5)
        m.drawstates(ax=ax, linewidth=0.3)
        m.drawcountries(ax=ax, linewidth=0.3)

        diff2d = DATA[(dom, var)]["diff"]
        lev_diff = LEVELS[var]

        cs = m.contourf(
            x,
            y,
            diff2d,
            levels=lev_diff,
            cmap=CMAP_DIFF,
            extend="both",
            ax=ax,
        )

        if PLOT_WIND:
            m.quiver(
                g["lon"][::WIND_SKIP, ::WIND_SKIP],
                g["lat"][::WIND_SKIP, ::WIND_SKIP],
                g["u"][::WIND_SKIP, ::WIND_SKIP],
                g["v"][::WIND_SKIP, ::WIND_SKIP],
                latlon=True,
                ax=ax,
                scale=QUIVER_SCALE,
                width=QUIVER_WIDTH,
                color="k",
                zorder=10,
            )

        if PLOT_OBS and (lon_obs is not None) and (lon_obs.size > 0):
            m.scatter(
                lon_obs,
                lat_obs,
                s=OBS_SIZE,
                marker=OBS_MARKER,
                facecolors=OBS_FACE_COLOR,
                edgecolors=OBS_EDGE_COLOR,
                linewidth=0.7,
                alpha=OBS_ALPHA,
                latlon=True,
                ax=ax,
                zorder=30,
            )

        if PLOT_LOWEST_LEVEL:
            lvl_str = "Lowest Level (k=0)"
        else:
            lvl_str = f"{ZAGL_M:.0f} m AGL"

        ax.set_title(f"{dom}: posterior - nature {var}", fontsize=11)

        cb = fig.colorbar(cs, ax=ax, location="right", pad=0.02, fraction=0.046)
        cb.set_label(f"{var} posterior - nature", fontsize=8)
        cb.ax.yaxis.set_major_locator(MaxNLocator(4))
        cb.ax.yaxis.set_major_formatter(ScalarFormatter(useOffset=False))

fig.suptitle(
    f"Posterior minus Nature Run | all | cycle={Path(CYCLE_DIR).name}",
    fontsize=12,
)

plt.savefig(OUTFIG, dpi=300, bbox_inches="tight")
plt.show()

# print(f"Saved figure: {OUTFIG}")
```

```python
#!/usr/bin/env python3
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
from pathlib import Path
from matplotlib.ticker import MaxNLocator, ScalarFormatter
from mpl_toolkits.basemap import Basemap

import pydartdiags.obs_sequence.obs_sequence as obsq

# =============================================================================
# USER SETTINGS
# =============================================================================
CYCLE_DIR   = "/gpfs/home/sa24m/stephen_asare/runs/all_kde/output/2015071412/"
# CYCLE_DIR   =  '/gpfs/home/sa24m/scratch/tqprof/run2/osse_out/dart_cycle/output/2015071421/'
    
NATURE_FILES = {
    "d01": "/gpfs/research/scratch/sa24m/tqprof/run2/osse_out/ens_wrf/e002/wrfout_d01_2015-07-14_12:00:00",
    "d02": "/gpfs/research/scratch/sa24m/tqprof/run2/osse_out/ens_wrf/e002/wrfout_d02_2015-07-14_12:00:00",
}

DOMS = ["d01", "d02"]
VARS_COLS = ["THM", "QVAPOR", "WSPD"]
TIME_INDEX = 0
OBS='RADIOSONDE_U_WIND_COMPONENT'
OBS_SEQ_FILE = CYCLE_DIR + "obs_seq.final"
OBS_TYPES_ALL = [
    "RADIOSONDE_SPECIFIC_HUMIDITY",
    "RADIOSONDE_TEMPERATURE",
    "RADIOSONDE_U_WIND_COMPONENT",
    "RADIOSONDE_V_WIND_COMPONENT",
]

# # obs time window
# OBS_TIME_START = "2015-07-14 11:30:00"
# OBS_TIME_END   = "2015-07-14 12:30:00"

OBS_TIME_START = f"2015-07-14 {int(CYCLE_DIR[-3:-1]) - 1:02d}:30:00"
OBS_TIME_END   = f"2015-07-14 {int(CYCLE_DIR[-3:-1]):02d}:30:00"

# plot style
CMAP_DIFF = "bwr"
OBS_CMAPS = {
    "THM": "coolwarm",
    "QVAPOR": "viridis",
    "WSPD": "plasma",
}
N_DIFF_LEVELS = 21
N_XSEC_PRESSURE_LEVELS = 121

# marker style
OBS_SIZE = 90
OBS_EDGE_COLOR = "k"
OBS_LINEWIDTH = 1.4
OBS_TEXT_FONTSIZE = 7

# output
FIGDIR = Path("fig")
FIGDIR.mkdir(exist_ok=True)

# =============================================================================
# BASIC HELPERS
# =============================================================================
def wrap_lon(lon):
    return ((lon + 180.0) % 360.0) - 180.0

def fmt_obs_value(var, val):
    if var == "THM":
        return f"{val:.1f}"
    elif var == "QVAPOR":
        return f"{val:.4f}"
    elif var == "WSPD":
        return f"{val:.1f}"
    return f"{val:.2f}"

def symmetric_levels(arrays, nlevels=21):
    absmax = 0.0
    for arr in arrays:
        local = np.nanmax(np.abs(arr))
        if np.isfinite(local):
            absmax = max(absmax, float(local))
    if absmax == 0.0:
        absmax = 1.0e-12
    return np.linspace(-absmax, absmax, nlevels)

def get_da(ds, name, time_index=0):
    da = ds[name]
    if "Time" in da.dims:
        da = da.isel(Time=time_index)
    return da

def get_2d_latlon(ds):
    lat = get_da(ds, "XLAT", TIME_INDEX).values
    lon = get_da(ds, "XLONG", TIME_INDEX).values
    return lat, lon

def setup_basemap_from_file(nc_path):
    ds = xr.open_dataset(nc_path)
    lat, lon = get_2d_latlon(ds)

    truelat1 = getattr(ds, "TRUELAT1", float(np.nanmean(lat)))
    truelat2 = getattr(ds, "TRUELAT2", float(np.nanmean(lat)))
    stand_lon = getattr(ds, "STAND_LON", float(np.nanmean(lon)))

    m = Basemap(
        projection="lcc",
        resolution="i",
        llcrnrlon=float(lon[0, 0]),
        llcrnrlat=float(lat[0, 0]),
        urcrnrlon=float(lon[-1, -1]),
        urcrnrlat=float(lat[-1, -1]),
        lat_1=truelat1,
        lat_2=truelat2,
        lon_0=stand_lon,
    )

    x, y = m(lon, lat)
    ds.close()
    return x, y, m

def destagger_to_mass(da, var_name):
    if var_name == "U" and "west_east_stag" in da.dims:
        da = 0.5 * (
            da.isel(west_east_stag=slice(0, -1)) +
            da.isel(west_east_stag=slice(1, None))
        ).rename({"west_east_stag": "west_east"})
    elif var_name == "V" and "south_north_stag" in da.dims:
        da = 0.5 * (
            da.isel(south_north_stag=slice(0, -1)) +
            da.isel(south_north_stag=slice(1, None))
        ).rename({"south_north_stag": "south_north"})
    elif var_name == "W" and "bottom_top_stag" in da.dims:
        da = 0.5 * (
            da.isel(bottom_top_stag=slice(0, -1)) +
            da.isel(bottom_top_stag=slice(1, None))
        ).rename({"bottom_top_stag": "bottom_top"})
    return da

def get_mass_field(ds, var, time_index=0):
    if var == "THM":
        if "THM" in ds.variables:
            return get_da(ds, "THM", time_index).values
        elif "T" in ds.variables:
            return get_da(ds, "T", time_index).values + 300.0
        else:
            raise KeyError("THM/T not found in file.")
    elif var == "QVAPOR":
        return get_da(ds, "QVAPOR", time_index).values
    elif var == "U":
        da = get_da(ds, "U", time_index)
        return destagger_to_mass(da, "U").values
    elif var == "V":
        da = get_da(ds, "V", time_index)
        return destagger_to_mass(da, "V").values
    elif var == "WSPD":
        u = get_mass_field(ds, "U", time_index)
        v = get_mass_field(ds, "V", time_index)
        return np.sqrt(u**2 + v**2)
    else:
        raise KeyError(f"Unsupported variable {var}")

def get_pressure_mass(ds, time_index=0):
    p = get_da(ds, "P", time_index).values
    pb = get_da(ds, "PB", time_index).values
    return p + pb

def interp_column_to_pressure(p_col, f_col, p_targets):
    mask = np.isfinite(p_col) & np.isfinite(f_col)
    if mask.sum() < 2:
        return np.full_like(np.atleast_1d(p_targets), np.nan, dtype=float)

    p_use = p_col[mask]
    f_use = f_col[mask]

    order = np.argsort(p_use)   # ascending pressure
    p_use = p_use[order]
    f_use = f_use[order]

    p_use, unique_idx = np.unique(p_use, return_index=True)
    f_use = f_use[unique_idx]

    p_targets = np.atleast_1d(p_targets).astype(float)
    out = np.interp(p_targets, p_use, f_use, left=np.nan, right=np.nan)

    out[(p_targets < p_use.min()) | (p_targets > p_use.max())] = np.nan
    return out

def interp_3d_to_pressure(field3d, p3d, p_target):
    nz, ny, nx = field3d.shape
    out = np.full((ny, nx), np.nan, dtype=float)

    field_flat = field3d.reshape(nz, -1)
    p_flat = p3d.reshape(nz, -1)

    for idx in range(field_flat.shape[1]):
        out.ravel()[idx] = interp_column_to_pressure(
            p_flat[:, idx],
            field_flat[:, idx],
            np.array([p_target])
        )[0]
    return out

def nearest_gridpoint(lon2d, lat2d, lon0, lat0):
    lon2d_w = wrap_lon(lon2d)
    d2 = (lon2d_w - lon0)**2 + (lat2d - lat0)**2
    j, i = np.unravel_index(np.nanargmin(d2), d2.shape)
    return j, i

# =============================================================================
# OBSERVATION HANDLING
# =============================================================================
def load_obs_profiles(obs_seq_file, t0, t1):
    obs_seq = obsq.ObsSequence(obs_seq_file)
    df = obs_seq.df.copy()

    if df.empty:
        raise ValueError(f"{obs_seq_file} loaded but contains no observations.")

    df["time"] = pd.to_datetime(df["time"])
    t0 = pd.Timestamp(t0)
    t1 = pd.Timestamp(t1)

    df = df[
        (df["time"] >= t0) &
        (df["time"] <= t1) &
        (df["type"].isin(OBS_TYPES_ALL))
    ].copy()

    if df.empty:
        raise ValueError("No observations remain after filtering by time/type.")

    df["lon_plot"] = wrap_lon(df["longitude"].values)

    # split by type
    q_df = df[df["type"] == "RADIOSONDE_SPECIFIC_HUMIDITY"].copy()
    t_df = df[df["type"] == "RADIOSONDE_TEMPERATURE"].copy()
    u_df = df[df["type"] == "RADIOSONDE_U_WIND_COMPONENT"].copy()
    v_df = df[df["type"] == "RADIOSONDE_V_WIND_COMPONENT"].copy()

    # build wind-speed obs from U/V at matching pressure levels
    u_work = u_df[["vertical", "longitude", "latitude", "lon_plot", "observation"]].copy()
    v_work = v_df[["vertical", "longitude", "latitude", "lon_plot", "observation"]].copy()

    u_work = u_work.rename(columns={"observation": "u_obs"})
    v_work = v_work.rename(columns={"observation": "v_obs"})

    u_work["vertical_key"] = u_work["vertical"].round(3)
    v_work["vertical_key"] = v_work["vertical"].round(3)

    wspd_df = pd.merge(
        u_work,
        v_work[["vertical_key", "v_obs"]],
        on="vertical_key",
        how="inner"
    )

    wspd_df["observation"] = np.sqrt(wspd_df["u_obs"]**2 + wspd_df["v_obs"]**2)
    wspd_df["vertical"] = wspd_df["vertical"]
    wspd_df["type"] = "RADIOSONDE_WIND_SPEED"

    # common pressure levels across all three plotted variable families
    p_temp = set(np.round(t_df["vertical"].unique(), 3))
    p_qv   = set(np.round(q_df["vertical"].unique(), 3))
    p_u    = set(np.round(u_df["vertical"].unique(), 3))
    p_v    = set(np.round(v_df["vertical"].unique(), 3))
    common_pressures = sorted(list(p_temp & p_qv & p_u & p_v), reverse=True)

    if len(common_pressures) == 0:
        raise ValueError("No common pressure levels found across T, QV, U, V observations.")

    # one sounding site
    station_lon = float(np.nanmean(t_df["lon_plot"].values))
    station_lat = float(np.nanmean(t_df["latitude"].values))

    obs_panels = {
        "THM": t_df,
        "QVAPOR": q_df,
        "WSPD": wspd_df,
    }

    return common_pressures, station_lon, station_lat, obs_panels

# =============================================================================
# HORIZONTAL PRESSURE-SLICE FIGURES
# =============================================================================
def compute_horizontal_slices(post_file, nature_file, pressure_levels):
    ds_post = xr.open_dataset(post_file)
    ds_nat = xr.open_dataset(nature_file)

    lat2d, lon2d = get_2d_latlon(ds_post)
    x, y, m = setup_basemap_from_file(post_file)

    # p_post = get_pressure_mass(ds_post, TIME_INDEX)
    # p_nat = get_pressure_mass(ds_nat, TIME_INDEX)
    # Use nature pressure for both files because postassim_mean does not contain P/PB
    p_ref = get_pressure_mass(ds_nat, TIME_INDEX)

    results = {}
    for plev in pressure_levels:
        results[plev] = {}
        for var in VARS_COLS:
            f_post = get_mass_field(ds_post, var, TIME_INDEX)
            f_nat = get_mass_field(ds_nat, var, TIME_INDEX)

            # post_2d = interp_3d_to_pressure(f_post, p_post, plev)
            # nat_2d = interp_3d_to_pressure(f_nat, p_nat, plev)

            post_2d = interp_3d_to_pressure(f_post, p_ref, plev)
            nat_2d  = interp_3d_to_pressure(f_nat,  p_ref, plev)
        
            results[plev][var] = post_2d - nat_2d

    ds_post.close()
    ds_nat.close()

    return results, x, y, m, lon2d, lat2d

def plot_horizontal_figure(dom, post_file, nature_file, pressure_levels, obs_panels):
    slices, x, y, m, lon2d, lat2d = compute_horizontal_slices(post_file, nature_file, pressure_levels)

    # shared contour levels per column
    diff_levels = {}
    for var in VARS_COLS:
        arrs = [slices[plev][var] for plev in pressure_levels]
        diff_levels[var] = symmetric_levels(arrs, N_DIFF_LEVELS)

    nrows = len(pressure_levels)
    ncols = len(VARS_COLS)

    fig, axs = plt.subplots(nrows, ncols, figsize=(18.5, 3.0 * nrows), constrained_layout=False)
    plt.subplots_adjust(left=0.04, right=0.93, bottom=0.04, top=0.95, wspace=0.16, hspace=0.16)

    last_cs = {}

    for r, plev in enumerate(pressure_levels):
        for c, var in enumerate(VARS_COLS):
            ax = axs[r, c] if nrows > 1 else axs[c]

            m.drawcoastlines(ax=ax, linewidth=0.45)
            m.drawstates(ax=ax, linewidth=0.25)
            m.drawcountries(ax=ax, linewidth=0.25)

            field = slices[plev][var]
            cs = m.contourf(
                x, y, field,
                levels=diff_levels[var],
                cmap=CMAP_DIFF,
                extend="both",
                ax=ax
            )
            last_cs[var] = cs

            # obs overlay at this pressure level
            obs_df = obs_panels[var]
            obs_sel = obs_df[np.isclose(obs_df["vertical"], plev, atol=1e-3)].copy()

            if not obs_sel.empty:
                sc = m.scatter(
                    obs_sel["lon_plot"].values,
                    obs_sel["latitude"].values,
                    c=obs_sel["observation"].values,
                    cmap=OBS_CMAPS[var],
                    s=OBS_SIZE,
                    edgecolors=OBS_EDGE_COLOR,
                    linewidth=OBS_LINEWIDTH,
                    latlon=True,
                    ax=ax,
                    zorder=30,
                )

                # annotate obs value
                for _, row in obs_sel.iterrows():
                    xt, yt = m(row["lon_plot"], row["latitude"])
                    ax.text(
                        xt, yt,
                        fmt_obs_value(var, row["observation"]),
                        fontsize=OBS_TEXT_FONTSIZE,
                        fontweight="bold",
                        color="k",
                        ha="left",
                        va="bottom",
                        zorder=35,
                        bbox=dict(facecolor="white", edgecolor="none", alpha=0.65, pad=0.4),
                    )

            ax.set_title(f"{var} | {plev/100:.0f} hPa", fontsize=10)

            if c == 0:
                ax.text(
                    0.01, 0.98, f"{plev/100:.0f} hPa",
                    transform=ax.transAxes,
                    ha="left", va="top",
                    fontsize=10, fontweight="bold",
                    bbox=dict(facecolor="white", edgecolor="none", alpha=0.7, pad=0.5),
                )

    # one colorbar per column
    for c, var in enumerate(VARS_COLS):
        cb = fig.colorbar(
            last_cs[var],
            ax=axs[:, c] if nrows > 1 else axs[c],
            location="right",
            pad=0.015,
            fraction=0.025
        )
        cb.set_label(f"{var}: posterior - nature", fontsize=9)
        cb.ax.yaxis.set_major_locator(MaxNLocator(5))
        cb.ax.yaxis.set_major_formatter(ScalarFormatter(useOffset=False))

    fig.suptitle(f"{dom}: Horizontal posterior-minus-nature fields at radiosonde pressure levels", fontsize=13)

    outfig = FIGDIR / f"{dom}_horizontal_posterior_minus_nature_obs_levels_{CYCLE_DIR[36:-27]}_{CYCLE_DIR[49:-19]}_{CYCLE_DIR[-3:-1]}.jpeg"
    plt.savefig(outfig, dpi=300, bbox_inches="tight")
    plt.show()
    print(f"Saved {outfig}")

# =============================================================================
# WEST-EAST PRESSURE CROSS-SECTIONS
# =============================================================================
def compute_west_east_cross_section(post_file, nature_file, station_lon, station_lat):
    ds_post = xr.open_dataset(post_file)
    ds_nat = xr.open_dataset(nature_file)

    lat2d, lon2d = get_2d_latlon(ds_post)
    lon2d_w = wrap_lon(lon2d)

    j0, i0 = nearest_gridpoint(lon2d_w, lat2d, station_lon, station_lat)

    # full-domain west-east line through the sounding row
    lon_line = lon2d_w[j0, :]
    lat_line = lat2d[j0, :]
    distance_km = 111.32 * np.cos(np.deg2rad(station_lat)) * (lon_line - station_lon)

    # pressure levels for the cross-section
    p_levels = np.linspace(100000.0, 5000.0, N_XSEC_PRESSURE_LEVELS)  # Pa, descending

    # p_post = get_pressure_mass(ds_post, TIME_INDEX)[:, j0, :]
    # p_nat  = get_pressure_mass(ds_nat, TIME_INDEX)[:, j0, :]

    # Use nature pressure for both posterior and nature fields
    p_ref = get_pressure_mass(ds_nat, TIME_INDEX)[:, j0, :]

    xsec = {}
    for var in VARS_COLS:
        f_post = get_mass_field(ds_post, var, TIME_INDEX)[:, j0, :]
        f_nat  = get_mass_field(ds_nat, var, TIME_INDEX)[:, j0, :]

        out = np.full((len(p_levels), f_post.shape[1]), np.nan, dtype=float)

        for i in range(f_post.shape[1]):
            # post_prof = interp_column_to_pressure(p_post[:, i], f_post[:, i], p_levels)
            # nat_prof  = interp_column_to_pressure(p_nat[:, i],  f_nat[:, i],  p_levels)
            post_prof = interp_column_to_pressure(p_ref[:, i], f_post[:, i], p_levels)
            nat_prof  = interp_column_to_pressure(p_ref[:, i], f_nat[:, i],  p_levels)

            out[:, i] = post_prof - nat_prof

        xsec[var] = out

    ds_post.close()
    ds_nat.close()

    return distance_km, p_levels, xsec, j0, i0

def plot_cross_section_figure(dom, post_file, nature_file, station_lon, station_lat, obs_panels):
    distance_km, p_levels, xsec, j0, i0 = compute_west_east_cross_section(
        post_file, nature_file, station_lon, station_lat
    )

    p_hpa = p_levels / 100.0

    fig, axs = plt.subplots(3, 1, figsize=(14, 13), constrained_layout=False)
    plt.subplots_adjust(left=0.08, right=0.90, bottom=0.06, top=0.95, hspace=0.22)

    for idx, var in enumerate(VARS_COLS):
        ax = axs[idx]
        levs = symmetric_levels([xsec[var]], N_DIFF_LEVELS)

        cs = ax.contourf(
            distance_km,
            p_hpa,
            xsec[var],
            levels=levs,
            cmap=CMAP_DIFF,
            extend="both"
        )

        ax.axvline(0.0, color="k", linestyle="--", linewidth=1.0)

        # overlay the observed levels at x = 0
        obs_df = obs_panels[var].copy()
        if not obs_df.empty:
            ax.scatter(
                np.zeros(len(obs_df)),
                obs_df["vertical"].values / 100.0,
                c=obs_df["observation"].values,
                cmap=OBS_CMAPS[var],
                s=OBS_SIZE,
                edgecolors=OBS_EDGE_COLOR,
                linewidths=OBS_LINEWIDTH,
                zorder=30,
            )

        ax.set_title(f"{dom}: West–East cross-section of posterior - nature | {var}", fontsize=11)
        ax.set_xlabel("Distance from radiosonde site (km)")
        ax.set_ylabel("Pressure (hPa)")
        ax.invert_yaxis()
        ax.grid(alpha=0.20)

        cb = fig.colorbar(cs, ax=ax, location="right", pad=0.02, fraction=0.035)
        cb.set_label(f"{var}: posterior - nature", fontsize=9)
        cb.ax.yaxis.set_major_locator(MaxNLocator(5))
        cb.ax.yaxis.set_major_formatter(ScalarFormatter(useOffset=False))

    fig.suptitle(f"{dom}: West–East pressure cross-sections through radiosonde site", fontsize=13)
    # OBS_TYPES = ['RADIOSONDE_U_WIND_COMPONENT'] # ['RADIOSONDE_SPECIFIC_HUMIDITY'] #['RADIOSONDE_TEMPERATURE'] #["RADIOSONDE_U_WIND_COMPONENT"]

    outfig = FIGDIR / f"{dom}_west_east_cross_section_posterior_minus_nature_{CYCLE_DIR[36:-27]}_{CYCLE_DIR[49:-19]}_{CYCLE_DIR[-3:-1]}.jpeg"
    plt.savefig(outfig, dpi=300, bbox_inches="tight")
    plt.show()
    print(f"Saved {outfig}")

# =============================================================================
# MAIN
# =============================================================================
def main():
    pressure_levels, station_lon, station_lat, obs_panels = load_obs_profiles(
        OBS_SEQ_FILE,
        OBS_TIME_START,
        OBS_TIME_END,
    )

    print("Common radiosonde pressure levels (Pa):", pressure_levels)
    print(f"Station location: lon={station_lon:.4f}, lat={station_lat:.4f}")

    for dom in DOMS:
        post_file = str(Path(CYCLE_DIR) / f"postassim_mean_{dom}.nc")
        nature_file = NATURE_FILES[dom]

        # Figure 1: horizontal slices
        plot_horizontal_figure(
            dom=dom,
            post_file=post_file,
            nature_file=nature_file,
            pressure_levels=pressure_levels,
            obs_panels=obs_panels,
        )

        # Figure 2: west-east pressure cross section
        plot_cross_section_figure(
            dom=dom,
            post_file=post_file,
            nature_file=nature_file,
            station_lon=station_lon,
            station_lat=station_lat,
            obs_panels=obs_panels,
        )

if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
from pathlib import Path
from matplotlib.ticker import MaxNLocator, ScalarFormatter
from mpl_toolkits.basemap import Basemap

import pydartdiags.obs_sequence.obs_sequence as obsq

# =============================================================================
# USER SETTINGS
# =============================================================================
CYCLE_DIR   = "/gpfs/home/sa24m/stephen_asare/runs/wind_only_18_eakf/output/2015071421/"

NATURE_FILES = {
    "d02": f"/gpfs/research/scratch/sa24m/tqprof/run2/osse_out/ens_wrf/e002/wrfout_d02_2015-07-14_{CYCLE_DIR[-3:-1]}:00:00",
}

DOMS = ["d02"]
VARS_COLS = ["THM", "QVAPOR", "WSPD"]
TIME_INDEX = 0

OBS_SEQ_FILE = CYCLE_DIR + "obs_seq.final"
OBS_TYPES_ALL = [
    "RADIOSONDE_SPECIFIC_HUMIDITY",
    "RADIOSONDE_TEMPERATURE",
    "RADIOSONDE_U_WIND_COMPONENT",
    "RADIOSONDE_V_WIND_COMPONENT",
]

OBS_TIME_START = f"2015-07-14 {int(CYCLE_DIR[-3:-1]) - 1:02d}:30:00"
OBS_TIME_END   = f"2015-07-14 {int(CYCLE_DIR[-3:-1]):02d}:30:00"

# plot style
CMAP_DIFF = "bwr"
N_DIFF_LEVELS = 21
N_XSEC_PRESSURE_LEVELS = 121

# marker style
OBS_SIZE = 90
OBS_EDGE_COLOR = "k"
OBS_LINEWIDTH = 1.4
OBS_TEXT_FONTSIZE = 7

# output
FIGDIR = Path("fig")
FIGDIR.mkdir(exist_ok=True)


# =============================================================================
# BASIC HELPERS
# =============================================================================
def wrap_lon(lon):
    return ((lon + 180.0) % 360.0) - 180.0

def fmt_obs_diff(var, val):
    if np.isnan(val):
        return "NaN"
    if var == "THM":
        return f"{val:+.1f}"
    elif var == "QVAPOR":
        return f"{val:+.4f}"
    elif var == "WSPD":
        return f"{val:+.1f}"
    return f"{val:+.2f}"

def symmetric_levels(arrays, nlevels=21):
    absmax = 0.0
    for arr in arrays:
        local = np.nanmax(np.abs(arr))
        if np.isfinite(local):
            absmax = max(absmax, float(local))
    if absmax == 0.0:
        absmax = 1.0e-12
    return np.linspace(-absmax, absmax, nlevels)

def get_da(ds, name, time_index=0):
    da = ds[name]
    if "Time" in da.dims:
        da = da.isel(Time=time_index)
    return da

def get_2d_latlon(ds):
    lat = get_da(ds, "XLAT", TIME_INDEX).values
    lon = get_da(ds, "XLONG", TIME_INDEX).values
    return lat, lon

def setup_basemap_from_file(nc_path):
    ds = xr.open_dataset(nc_path)
    lat, lon = get_2d_latlon(ds)

    truelat1 = getattr(ds, "TRUELAT1", float(np.nanmean(lat)))
    truelat2 = getattr(ds, "TRUELAT2", float(np.nanmean(lat)))
    stand_lon = getattr(ds, "STAND_LON", float(np.nanmean(lon)))

    m = Basemap(
        projection="lcc",
        resolution="i",
        llcrnrlon=float(lon[0, 0]),
        llcrnrlat=float(lat[0, 0]),
        urcrnrlon=float(lon[-1, -1]),
        urcrnrlat=float(lat[-1, -1]),
        lat_1=truelat1,
        lat_2=truelat2,
        lon_0=stand_lon,
    )

    x, y = m(lon, lat)
    ds.close()
    return x, y, m

def destagger_to_mass(da, var_name):
    if var_name == "U" and "west_east_stag" in da.dims:
        da = 0.5 * (
            da.isel(west_east_stag=slice(0, -1)) +
            da.isel(west_east_stag=slice(1, None))
        ).rename({"west_east_stag": "west_east"})
    elif var_name == "V" and "south_north_stag" in da.dims:
        da = 0.5 * (
            da.isel(south_north_stag=slice(0, -1)) +
            da.isel(south_north_stag=slice(1, None))
        ).rename({"south_north_stag": "south_north"})
    elif var_name == "W" and "bottom_top_stag" in da.dims:
        da = 0.5 * (
            da.isel(bottom_top_stag=slice(0, -1)) +
            da.isel(bottom_top_stag=slice(1, None))
        ).rename({"bottom_top_stag": "bottom_top"})
    return da

def get_mass_field(ds, var, time_index=0):
    if var == "THM":
        if "THM" in ds.variables:
            val = get_da(ds, "THM", time_index).values
        elif "T" in ds.variables:
            val = get_da(ds, "T", time_index).values
        else:
            raise KeyError("THM/T not found in file.")
        
        # FIX: Ensure we are always returning TOTAL Potential Temperature.
        if np.nanmean(val) < 150.0:
            val = val + 300.0
        return val

    elif var == "QVAPOR":
        return get_da(ds, "QVAPOR", time_index).values
    elif var == "U":
        da = get_da(ds, "U", time_index)
        return destagger_to_mass(da, "U").values
    elif var == "V":
        da = get_da(ds, "V", time_index)
        return destagger_to_mass(da, "V").values
    elif var == "WSPD":
        u = get_mass_field(ds, "U", time_index)
        v = get_mass_field(ds, "V", time_index)
        return np.sqrt(u**2 + v**2)
    else:
        raise KeyError(f"Unsupported variable {var}")

def get_pressure_mass(ds, time_index=0):
    p = get_da(ds, "P", time_index).values
    pb = get_da(ds, "PB", time_index).values
    return p + pb

def interp_column_to_pressure(p_col, f_col, p_targets):
    mask = np.isfinite(p_col) & np.isfinite(f_col)
    if mask.sum() < 2:
        return np.full_like(np.atleast_1d(p_targets), np.nan, dtype=float)

    p_use = p_col[mask]
    f_use = f_col[mask]

    order = np.argsort(p_use)
    p_use = p_use[order]
    f_use = f_use[order]

    p_use, unique_idx = np.unique(p_use, return_index=True)
    f_use = f_use[unique_idx]

    p_targets = np.atleast_1d(p_targets).astype(float)
    out = np.interp(p_targets, p_use, f_use, left=np.nan, right=np.nan)

    out[(p_targets < p_use.min()) | (p_targets > p_use.max())] = np.nan
    return out

def interp_3d_to_pressure(field3d, p3d, p_target):
    nz, ny, nx = field3d.shape
    out = np.full((ny, nx), np.nan, dtype=float)

    field_flat = field3d.reshape(nz, -1)
    p_flat = p3d.reshape(nz, -1)

    for idx in range(field_flat.shape[1]):
        out.ravel()[idx] = interp_column_to_pressure(
            p_flat[:, idx],
            field_flat[:, idx],
            np.array([p_target])
        )[0]
    return out

def nearest_gridpoint(lon2d, lat2d, lon0, lat0):
    lon2d_w = wrap_lon(lon2d)
    d2 = (lon2d_w - lon0)**2 + (lat2d - lat0)**2
    j, i = np.unravel_index(np.nanargmin(d2), d2.shape)
    return j, i


# =============================================================================
# OBSERVATION HANDLING
# =============================================================================
def load_obs_profiles(obs_seq_file, t0, t1):
    obs_seq = obsq.ObsSequence(obs_seq_file)
    df = obs_seq.df.copy()

    if df.empty:
        raise ValueError(f"{obs_seq_file} loaded but contains no observations.")

    df["time"] = pd.to_datetime(df["time"])
    t0 = pd.Timestamp(t0)
    t1 = pd.Timestamp(t1)

    df = df[
        (df["time"] >= t0) &
        (df["time"] <= t1) &
        (df["type"].isin(OBS_TYPES_ALL)) &
        (df["posterior_ensemble_mean"] != -888888.0)
    ].copy()

    if df.empty:
        raise ValueError("No valid observations remain after filtering.")

    df["lon_plot"] = wrap_lon(df["longitude"].values)

    q_df = df[df["type"] == "RADIOSONDE_SPECIFIC_HUMIDITY"].copy()
    t_df = df[df["type"] == "RADIOSONDE_TEMPERATURE"].copy()
    u_df = df[df["type"] == "RADIOSONDE_U_WIND_COMPONENT"].copy()
    v_df = df[df["type"] == "RADIOSONDE_V_WIND_COMPONENT"].copy()

    # Note: We only use these dataframes to get lat/lon/pressure.
    # The actual observation values from DART will be ignored,
    # because we will purely sample the gridded files for an apples-to-apples comparison.
    u_work = u_df[["vertical", "longitude", "latitude", "lon_plot", "observation"]].rename(columns={"observation": "u_obs"})
    v_work = v_df[["vertical", "longitude", "latitude", "lon_plot", "observation"]].rename(columns={"observation": "v_obs"})

    u_work["vertical_key"] = u_work["vertical"].round(3)
    v_work["vertical_key"] = v_work["vertical"].round(3)

    wspd_df = pd.merge(
        u_work,
        v_work[["vertical_key", "v_obs"]],
        on="vertical_key",
        how="inner"
    )

    wspd_df["vertical"] = wspd_df["vertical"]
    wspd_df["type"] = "RADIOSONDE_WIND_SPEED"

    p_temp = set(np.round(t_df["vertical"].unique(), 3))
    p_qv   = set(np.round(q_df["vertical"].unique(), 3))
    p_u    = set(np.round(u_df["vertical"].unique(), 3))
    p_v    = set(np.round(v_df["vertical"].unique(), 3))
    common_pressures = sorted(list(p_temp & p_qv & p_u & p_v), reverse=True)

    if 91500.0 not in common_pressures:
        common_pressures.append(91500.0)
        common_pressures = sorted(common_pressures, reverse=True)

    if len(common_pressures) == 0:
        raise ValueError("No common pressure levels found across T, QV, U, V observations.")

    station_lon = float(np.nanmean(t_df["lon_plot"].values))
    station_lat = float(np.nanmean(t_df["latitude"].values))

    obs_panels = {
        "THM": t_df,
        "QVAPOR": q_df,
        "WSPD": wspd_df,
    }

    return common_pressures, station_lon, station_lat, obs_panels

def append_gridded_diff(obs_panels, post_file, nature_file):
    """
    Forces pure grid sampling: Extracts both the posterior and nature run
    data directly from their gridded NetCDF files at the observation's exact 
    location and pressure level, bypassing DART's H(x) completely.
    """
    ds_post = xr.open_dataset(post_file)
    ds_nat = xr.open_dataset(nature_file)
    
    lat2d, lon2d = get_2d_latlon(ds_post)
    # Using the exact same reference pressure as the contour plots
    p_ref = get_pressure_mass(ds_nat, TIME_INDEX)
    
    for var, df in obs_panels.items():
        if df.empty:
            df["post_hx"] = np.nan
            df["nature_hx"] = np.nan
            df["diff_post_nat"] = np.nan
            continue
            
        f_post = get_mass_field(ds_post, var, TIME_INDEX)
        f_nat = get_mass_field(ds_nat, var, TIME_INDEX)
        
        post_hx_vals = []
        nature_hx_vals = []
        
        for _, row in df.iterrows():
            j, i = nearest_gridpoint(lon2d, lat2d, row["longitude"], row["latitude"])
            p_col = p_ref[:, j, i]
            f_p_col = f_post[:, j, i]
            f_n_col = f_nat[:, j, i]
            
            p_val = interp_column_to_pressure(p_col, f_p_col, np.array([row["vertical"]]))[0]
            n_val = interp_column_to_pressure(p_col, f_n_col, np.array([row["vertical"]]))[0]
            
            post_hx_vals.append(p_val)
            nature_hx_vals.append(n_val)
            
        df["post_hx"] = post_hx_vals
        df["nature_hx"] = nature_hx_vals
        
        # Calculate strictly Apples-to-Apples difference directly from the WRF grids
        df["diff_post_nat"] = df["post_hx"] - df["nature_hx"]
        
    ds_post.close()
    ds_nat.close()
    return obs_panels


# =============================================================================
# HORIZONTAL PRESSURE-SLICE FIGURES
# =============================================================================
def compute_horizontal_slices(post_file, nature_file, pressure_levels):
    ds_post = xr.open_dataset(post_file)
    ds_nat = xr.open_dataset(nature_file)

    lat2d, lon2d = get_2d_latlon(ds_post)
    x, y, m = setup_basemap_from_file(post_file)

    p_ref = get_pressure_mass(ds_nat, TIME_INDEX)

    results = {}
    for plev in pressure_levels:
        results[plev] = {}
        for var in VARS_COLS:
            f_post = get_mass_field(ds_post, var, TIME_INDEX)
            f_nat = get_mass_field(ds_nat, var, TIME_INDEX)

            post_2d = interp_3d_to_pressure(f_post, p_ref, plev)
            nat_2d  = interp_3d_to_pressure(f_nat,  p_ref, plev)
        
            results[plev][var] = post_2d - nat_2d

    ds_post.close()
    ds_nat.close()

    return results, x, y, m, lon2d, lat2d

def plot_horizontal_figure(dom, post_file, nature_file, pressure_levels, obs_panels):
    slices, x, y, m, lon2d, lat2d = compute_horizontal_slices(post_file, nature_file, pressure_levels)

    diff_levels = {}
    for var in VARS_COLS:
        arrs = [slices[plev][var] for plev in pressure_levels]
        diff_levels[var] = symmetric_levels(arrs, N_DIFF_LEVELS)

    nrows = len(pressure_levels)
    ncols = len(VARS_COLS)

    fig, axs = plt.subplots(nrows, ncols, figsize=(18.5, 3.0 * nrows), constrained_layout=False)
    plt.subplots_adjust(left=0.04, right=0.93, bottom=0.04, top=0.95, wspace=0.16, hspace=0.16)

    last_cs = {}

    for r, plev in enumerate(pressure_levels):
        for c, var in enumerate(VARS_COLS):
            ax = axs[r, c] if nrows > 1 else axs[c]

            m.drawcoastlines(ax=ax, linewidth=0.45)
            m.drawstates(ax=ax, linewidth=0.25)
            m.drawcountries(ax=ax, linewidth=0.25)

            field = slices[plev][var]
            
            vmin = diff_levels[var][0]
            vmax = diff_levels[var][-1]
            
            cs = m.contourf(
                x, y, field,
                levels=diff_levels[var],
                cmap=CMAP_DIFF,
                extend="both",
                ax=ax
            )
            last_cs[var] = cs

            obs_df = obs_panels[var]
            obs_sel = obs_df[np.isclose(obs_df["vertical"], plev, atol=1e-3)].copy()

            if not obs_sel.empty:
                sc = m.scatter(
                    obs_sel["lon_plot"].values,
                    obs_sel["latitude"].values,
                    c=obs_sel["diff_post_nat"].values,
                    cmap=CMAP_DIFF,
                    vmin=vmin,
                    vmax=vmax,
                    s=OBS_SIZE,
                    edgecolors=OBS_EDGE_COLOR,
                    linewidth=OBS_LINEWIDTH,
                    latlon=True,
                    ax=ax,
                    zorder=30,
                )

                for _, row in obs_sel.iterrows():
                    xt, yt = m(row["lon_plot"], row["latitude"])
                    ax.text(
                        xt, yt,
                        fmt_obs_diff(var, row["diff_post_nat"]),
                        fontsize=OBS_TEXT_FONTSIZE,
                        fontweight="bold",
                        color="k",
                        ha="left",
                        va="bottom",
                        zorder=35,
                        bbox=dict(facecolor="white", edgecolor="none", alpha=0.65, pad=0.4),
                    )

            ax.set_title(f"{var} | {plev/100:.0f} hPa", fontsize=10)

            if c == 0:
                ax.text(
                    0.01, 0.98, f"{plev/100:.0f} hPa",
                    transform=ax.transAxes,
                    ha="left", va="top",
                    fontsize=10, fontweight="bold",
                    bbox=dict(facecolor="white", edgecolor="none", alpha=0.7, pad=0.5),
                )

    for c, var in enumerate(VARS_COLS):
        cb = fig.colorbar(
            last_cs[var],
            ax=axs[:, c] if nrows > 1 else axs[c],
            location="right",
            pad=0.015,
            fraction=0.025
        )
        cb.set_label(f"{var}: Posterior H(x) - Nature Run", fontsize=9)
        cb.ax.yaxis.set_major_locator(MaxNLocator(5))
        cb.ax.yaxis.set_major_formatter(ScalarFormatter(useOffset=False))

    fig.suptitle(f"{dom}: Horizontal Difference Fields (Posterior vs Nature Run, Pure Grid Sampling)", fontsize=13)

    outfig = FIGDIR / f"{dom}_horizontal_posterior_minus_nature_obs_levels_{CYCLE_DIR[36:-19]}_{CYCLE_DIR[-3:-1]}.jpeg"
    plt.savefig(outfig, dpi=300, bbox_inches="tight")
    print(f"Saved {outfig}")

# =============================================================================
# WEST-EAST PRESSURE CROSS-SECTIONS
# =============================================================================
def compute_west_east_cross_section(post_file, nature_file, station_lon, station_lat):
    ds_post = xr.open_dataset(post_file)
    ds_nat = xr.open_dataset(nature_file)

    lat2d, lon2d = get_2d_latlon(ds_post)
    lon2d_w = wrap_lon(lon2d)

    j0, i0 = nearest_gridpoint(lon2d_w, lat2d, station_lon, station_lat)

    lon_line = lon2d_w[j0, :]
    distance_km = 111.32 * np.cos(np.deg2rad(station_lat)) * (lon_line - station_lon)

    p_levels = np.linspace(100000.0, 5000.0, N_XSEC_PRESSURE_LEVELS)

    p_ref = get_pressure_mass(ds_nat, TIME_INDEX)[:, j0, :]

    xsec = {}
    for var in VARS_COLS:
        f_post = get_mass_field(ds_post, var, TIME_INDEX)[:, j0, :]
        f_nat  = get_mass_field(ds_nat, var, TIME_INDEX)[:, j0, :]

        out = np.full((len(p_levels), f_post.shape[1]), np.nan, dtype=float)

        for i in range(f_post.shape[1]):
            post_prof = interp_column_to_pressure(p_ref[:, i], f_post[:, i], p_levels)
            nat_prof  = interp_column_to_pressure(p_ref[:, i], f_nat[:, i],  p_levels)
            out[:, i] = post_prof - nat_prof

        xsec[var] = out

    ds_post.close()
    ds_nat.close()

    return distance_km, p_levels, xsec, j0, i0

def plot_cross_section_figure(dom, post_file, nature_file, station_lon, station_lat, obs_panels):
    distance_km, p_levels, xsec, j0, i0 = compute_west_east_cross_section(
        post_file, nature_file, station_lon, station_lat
    )

    p_hpa = p_levels / 100.0

    fig, axs = plt.subplots(3, 1, figsize=(14, 13), constrained_layout=False)
    plt.subplots_adjust(left=0.08, right=0.90, bottom=0.06, top=0.95, hspace=0.22)

    for idx, var in enumerate(VARS_COLS):
        ax = axs[idx]
        levs = symmetric_levels([xsec[var]], N_DIFF_LEVELS)
        vmin = levs[0]
        vmax = levs[-1]

        cs = ax.contourf(
            distance_km,
            p_hpa,
            xsec[var],
            levels=levs,
            cmap=CMAP_DIFF,
            extend="both"
        )

        ax.axvline(0.0, color="k", linestyle="--", linewidth=1.0)

        obs_df = obs_panels[var].copy()
        if not obs_df.empty:
            ax.scatter(
                np.zeros(len(obs_df)),
                obs_df["vertical"].values / 100.0,
                c=obs_df["diff_post_nat"].values,
                cmap=CMAP_DIFF,
                vmin=vmin,
                vmax=vmax,
                s=OBS_SIZE,
                edgecolors=OBS_EDGE_COLOR,
                linewidths=OBS_LINEWIDTH,
                zorder=30,
            )

        ax.set_title(f"{dom}: West–East cross-section of Posterior - Nature | {var}", fontsize=11)
        ax.set_xlabel("Distance from radiosonde site (km)")
        ax.set_ylabel("Pressure (hPa)")
        ax.invert_yaxis()
        ax.grid(alpha=0.20)

        cb = fig.colorbar(cs, ax=ax, location="right", pad=0.02, fraction=0.035)
        cb.set_label(f"{var}: Posterior - Nature", fontsize=9)
        cb.ax.yaxis.set_major_locator(MaxNLocator(5))
        cb.ax.yaxis.set_major_formatter(ScalarFormatter(useOffset=False))

    fig.suptitle(f"{dom}: West–East pressure cross-sections through radiosonde site (Pure Grid Sampling)", fontsize=13)

    outfig = FIGDIR / f"{dom}_west_east_cross_section_posterior_minus_nature_{CYCLE_DIR[36:-19]}_{CYCLE_DIR[-3:-1]}.jpeg"
    plt.savefig(outfig, dpi=300, bbox_inches="tight")
    print(f"Saved {outfig}")

# =============================================================================
# MAIN
# =============================================================================
def main():
    pressure_levels, station_lon, station_lat, obs_panels = load_obs_profiles(
        OBS_SEQ_FILE,
        OBS_TIME_START,
        OBS_TIME_END,
    )

    print("Target pressure levels (Pa):", pressure_levels)
    print(f"Station location: lon={station_lon:.4f}, lat={station_lat:.4f}")

    for dom in DOMS: 
        post_file = str(Path(CYCLE_DIR) / f"postassim_mean_{dom}.nc")
        nature_file = NATURE_FILES[dom]

        obs_panels = append_gridded_diff(obs_panels, post_file, nature_file)

        plot_horizontal_figure(
            dom=dom,
            post_file=post_file,
            nature_file=nature_file,
            pressure_levels=pressure_levels,
            obs_panels=obs_panels,
        )

        plot_cross_section_figure(
            dom=dom,
            post_file=post_file,
            nature_file=nature_file,
            station_lon=station_lon,
            station_lat=station_lat,
            obs_panels=obs_panels,
        )

if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
from pathlib import Path
from matplotlib.ticker import MaxNLocator, ScalarFormatter
from mpl_toolkits.basemap import Basemap

import pydartdiags.obs_sequence.obs_sequence as obsq

# =============================================================================
# USER SETTINGS
# =============================================================================
CYCLE_DIR   = "/gpfs/home/sa24m/stephen_asare/runs/sph_only_18_eakf/output/2015071421/"

NATURE_FILES = {
    "d02": f"/gpfs/research/scratch/sa24m/tqprof/run2/osse_out/ens_wrf/e002/wrfout_d02_2015-07-14_{CYCLE_DIR[-3:-1]}:00:00",
}

DOMS = ["d02"]
VARS_COLS = ["THM", "QVAPOR", "WSPD"]
TIME_INDEX = 0

OBS_SEQ_FILE = CYCLE_DIR + "obs_seq.final"
OBS_TYPES_ALL = [
    "RADIOSONDE_SPECIFIC_HUMIDITY",
    "RADIOSONDE_TEMPERATURE",
    "RADIOSONDE_U_WIND_COMPONENT",
    "RADIOSONDE_V_WIND_COMPONENT",
]

OBS_TIME_START = f"2015-07-14 {int(CYCLE_DIR[-3:-1]) - 1:02d}:30:00"
OBS_TIME_END   = f"2015-07-14 {int(CYCLE_DIR[-3:-1]):02d}:30:00"

# plot style
CMAP_DIFF = "bwr"
N_DIFF_LEVELS = 21
N_XSEC_PRESSURE_LEVELS = 121

# marker style
OBS_SIZE = 90
OBS_EDGE_COLOR = "k"
OBS_LINEWIDTH = 1.4
OBS_TEXT_FONTSIZE = 7

# output
FIGDIR = Path("figs")
FIGDIR.mkdir(exist_ok=True)


# =============================================================================
# BASIC HELPERS
# =============================================================================
def wrap_lon(lon):
    return ((lon + 180.0) % 360.0) - 180.0

def fmt_obs_diff(var, val):
    if np.isnan(val):
        return "NaN"
    if var == "THM":
        return f"{val:+.1f}"
    elif var == "QVAPOR":
        return f"{val:+.4f}"
    elif var == "WSPD":
        return f"{val:+.1f}"
    return f"{val:+.2f}"

def symmetric_levels(arrays, nlevels=21):
    absmax = 0.0
    for arr in arrays:
        local = np.nanmax(np.abs(arr))
        if np.isfinite(local):
            absmax = max(absmax, float(local))
    if absmax == 0.0:
        absmax = 1.0e-12
    return np.linspace(-absmax, absmax, nlevels)

def get_da(ds, name, time_index=0):
    da = ds[name]
    if "Time" in da.dims:
        da = da.isel(Time=time_index)
    return da

def get_2d_latlon(ds):
    lat = get_da(ds, "XLAT", TIME_INDEX).values
    lon = get_da(ds, "XLONG", TIME_INDEX).values
    return lat, lon

def setup_basemap_from_file(nc_path):
    ds = xr.open_dataset(nc_path)
    lat, lon = get_2d_latlon(ds)

    truelat1 = getattr(ds, "TRUELAT1", float(np.nanmean(lat)))
    truelat2 = getattr(ds, "TRUELAT2", float(np.nanmean(lat)))
    stand_lon = getattr(ds, "STAND_LON", float(np.nanmean(lon)))

    m = Basemap(
        projection="lcc",
        resolution="i",
        llcrnrlon=float(lon[0, 0]),
        llcrnrlat=float(lat[0, 0]),
        urcrnrlon=float(lon[-1, -1]),
        urcrnrlat=float(lat[-1, -1]),
        lat_1=truelat1,
        lat_2=truelat2,
        lon_0=stand_lon,
    )

    x, y = m(lon, lat)
    ds.close()
    return x, y, m

def destagger_to_mass(da, var_name):
    if var_name == "U" and "west_east_stag" in da.dims:
        da = 0.5 * (
            da.isel(west_east_stag=slice(0, -1)) +
            da.isel(west_east_stag=slice(1, None))
        ).rename({"west_east_stag": "west_east"})
    elif var_name == "V" and "south_north_stag" in da.dims:
        da = 0.5 * (
            da.isel(south_north_stag=slice(0, -1)) +
            da.isel(south_north_stag=slice(1, None))
        ).rename({"south_north_stag": "south_north"})
    elif var_name == "W" and "bottom_top_stag" in da.dims:
        da = 0.5 * (
            da.isel(bottom_top_stag=slice(0, -1)) +
            da.isel(bottom_top_stag=slice(1, None))
        ).rename({"bottom_top_stag": "bottom_top"})
    return da

def get_mass_field(ds, var, time_index=0):
    if var == "THM":
        if "THM" in ds.variables:
            val = get_da(ds, "THM", time_index).values
        elif "T" in ds.variables:
            val = get_da(ds, "T", time_index).values
        else:
            raise KeyError("THM/T not found in file.")
        
        # FIX: Ensure we are always returning TOTAL Potential Temperature.
        if np.nanmean(val) < 150.0:
            val = val + 300.0
        return val

    elif var == "QVAPOR":
        return get_da(ds, "QVAPOR", time_index).values
    elif var == "U":
        da = get_da(ds, "U", time_index)
        return destagger_to_mass(da, "U").values
    elif var == "V":
        da = get_da(ds, "V", time_index)
        return destagger_to_mass(da, "V").values
    elif var == "WSPD":
        u = get_mass_field(ds, "U", time_index)
        v = get_mass_field(ds, "V", time_index)
        return np.sqrt(u**2 + v**2)
    else:
        raise KeyError(f"Unsupported variable {var}")

def get_pressure_mass(ds, time_index=0):
    p = get_da(ds, "P", time_index).values
    pb = get_da(ds, "PB", time_index).values
    return p + pb

def interp_column_to_pressure(p_col, f_col, p_targets):
    mask = np.isfinite(p_col) & np.isfinite(f_col)
    if mask.sum() < 2:
        return np.full_like(np.atleast_1d(p_targets), np.nan, dtype=float)

    p_use = p_col[mask]
    f_use = f_col[mask]

    order = np.argsort(p_use)
    p_use = p_use[order]
    f_use = f_use[order]

    p_use, unique_idx = np.unique(p_use, return_index=True)
    f_use = f_use[unique_idx]

    p_targets = np.atleast_1d(p_targets).astype(float)
    out = np.interp(p_targets, p_use, f_use, left=np.nan, right=np.nan)

    out[(p_targets < p_use.min()) | (p_targets > p_use.max())] = np.nan
    return out

def interp_3d_to_pressure(field3d, p3d, p_target):
    nz, ny, nx = field3d.shape
    out = np.full((ny, nx), np.nan, dtype=float)

    field_flat = field3d.reshape(nz, -1)
    p_flat = p3d.reshape(nz, -1)

    for idx in range(field_flat.shape[1]):
        out.ravel()[idx] = interp_column_to_pressure(
            p_flat[:, idx],
            field_flat[:, idx],
            np.array([p_target])
        )[0]
    return out

def nearest_gridpoint(lon2d, lat2d, lon0, lat0):
    lon2d_w = wrap_lon(lon2d)
    lon0_w = wrap_lon(lon0)
    dlon = (lon2d_w - lon0_w + 180.0) % 360.0 - 180.0
    d2 = dlon**2 + (lat2d - lat0)**2
    j, i = np.unravel_index(np.nanargmin(d2), d2.shape)
    return j, i


# =============================================================================
# OBSERVATION HANDLING
# =============================================================================
def load_obs_profiles(obs_seq_file, t0, t1):
    obs_seq = obsq.ObsSequence(obs_seq_file)
    df = obs_seq.df.copy()

    if df.empty:
        raise ValueError(f"{obs_seq_file} loaded but contains no observations.")

    df["time"] = pd.to_datetime(df["time"])
    t0 = pd.Timestamp(t0)
    t1 = pd.Timestamp(t1)

    df = df[
        (df["time"] >= t0) &
        (df["time"] <= t1) &
        (df["type"].isin(OBS_TYPES_ALL)) &
        (df["posterior_ensemble_mean"] != -888888.0)
    ].copy()

    if df.empty:
        raise ValueError("No valid observations remain after filtering.")

    df["lon_plot"] = wrap_lon(df["longitude"].values)

    q_df = df[df["type"] == "RADIOSONDE_SPECIFIC_HUMIDITY"].copy()
    t_df = df[df["type"] == "RADIOSONDE_TEMPERATURE"].copy()
    u_df = df[df["type"] == "RADIOSONDE_U_WIND_COMPONENT"].copy()
    v_df = df[df["type"] == "RADIOSONDE_V_WIND_COMPONENT"].copy()

    # Note: We only use these dataframes to get lat/lon/pressure.
    # The actual observation values from DART will be ignored,
    # because we will purely sample the gridded files for an apples-to-apples comparison.
    u_work = u_df[["vertical", "longitude", "latitude", "lon_plot", "observation"]].rename(columns={"observation": "u_obs"})
    v_work = v_df[["vertical", "longitude", "latitude", "lon_plot", "observation"]].rename(columns={"observation": "v_obs"})

    u_work["vertical_key"] = u_work["vertical"].round(3)
    v_work["vertical_key"] = v_work["vertical"].round(3)

    wspd_df = pd.merge(
        u_work,
        v_work[["vertical_key", "v_obs"]],
        on="vertical_key",
        how="inner"
    )

    wspd_df["vertical"] = wspd_df["vertical"]
    wspd_df["type"] = "RADIOSONDE_WIND_SPEED"

    p_temp = set(np.round(t_df["vertical"].unique(), 3))
    p_qv   = set(np.round(q_df["vertical"].unique(), 3))
    p_u    = set(np.round(u_df["vertical"].unique(), 3))
    p_v    = set(np.round(v_df["vertical"].unique(), 3))
    common_pressures = sorted(list(p_temp & p_qv & p_u & p_v), reverse=True)

    if 91500.0 not in common_pressures:
        common_pressures.append(91500.0)
        common_pressures = sorted(common_pressures, reverse=True)

    if len(common_pressures) == 0:
        raise ValueError("No common pressure levels found across T, QV, U, V observations.")

    station_lon = float(np.nanmean(t_df["lon_plot"].values))
    station_lat = float(np.nanmean(t_df["latitude"].values))

    obs_panels = {
        "THM": t_df,
        "QVAPOR": q_df,
        "WSPD": wspd_df,
    }

    return common_pressures, station_lon, station_lat, obs_panels

def append_gridded_diff(obs_panels, post_file, nature_file):
    """
    Forces pure grid sampling: Extracts both the posterior and nature run
    data directly from their gridded NetCDF files at the observation's exact 
    location and pressure level, bypassing DART's H(x) completely.
    """
    ds_post = xr.open_dataset(post_file)
    ds_nat = xr.open_dataset(nature_file)
    
    lat2d, lon2d = get_2d_latlon(ds_post)
    # Using the exact same reference pressure as the contour plots
    p_ref = get_pressure_mass(ds_nat, TIME_INDEX)
    
    for var, df in obs_panels.items():
        if df.empty:
            df["post_hx"] = np.nan
            df["nature_hx"] = np.nan
            df["diff_post_nat"] = np.nan
            continue
            
        f_post = get_mass_field(ds_post, var, TIME_INDEX)
        f_nat = get_mass_field(ds_nat, var, TIME_INDEX)
        
        post_hx_vals = []
        nature_hx_vals = []
        
        for _, row in df.iterrows():
            j, i = nearest_gridpoint(lon2d, lat2d, row["longitude"], row["latitude"])
            p_col = p_ref[:, j, i]
            f_p_col = f_post[:, j, i]
            f_n_col = f_nat[:, j, i]
            
            p_val = interp_column_to_pressure(p_col, f_p_col, np.array([row["vertical"]]))[0]
            n_val = interp_column_to_pressure(p_col, f_n_col, np.array([row["vertical"]]))[0]
            
            post_hx_vals.append(p_val)
            nature_hx_vals.append(n_val)
            
        df["post_hx"] = post_hx_vals
        df["nature_hx"] = nature_hx_vals
        
        # Calculate strictly Apples-to-Apples difference directly from the WRF grids
        df["diff_post_nat"] = df["post_hx"] - df["nature_hx"]
        
    ds_post.close()
    ds_nat.close()
    return obs_panels


# =============================================================================
# HORIZONTAL PRESSURE-SLICE FIGURES
# =============================================================================
def compute_horizontal_slices(post_file, nature_file, pressure_levels):
    ds_post = xr.open_dataset(post_file)
    ds_nat = xr.open_dataset(nature_file)

    lat2d, lon2d = get_2d_latlon(ds_post)
    x, y, m = setup_basemap_from_file(post_file)

    p_ref = get_pressure_mass(ds_nat, TIME_INDEX)

    results = {}
    for plev in pressure_levels:
        results[plev] = {}
        for var in VARS_COLS:
            f_post = get_mass_field(ds_post, var, TIME_INDEX)
            f_nat = get_mass_field(ds_nat, var, TIME_INDEX)

            post_2d = interp_3d_to_pressure(f_post, p_ref, plev)
            nat_2d  = interp_3d_to_pressure(f_nat,  p_ref, plev)
        
            results[plev][var] = post_2d - nat_2d

    ds_post.close()
    ds_nat.close()

    return results, x, y, m, lon2d, lat2d

def plot_horizontal_figure(dom, post_file, nature_file, pressure_levels, obs_panels):
    slices, x, y, m, lon2d, lat2d = compute_horizontal_slices(post_file, nature_file, pressure_levels)

    diff_levels = {}
    for var in VARS_COLS:
        arrs = [slices[plev][var] for plev in pressure_levels]
        diff_levels[var] = symmetric_levels(arrs, N_DIFF_LEVELS)

    nrows = len(pressure_levels)
    ncols = len(VARS_COLS)

    fig, axs = plt.subplots(nrows, ncols, figsize=(18.5, 3.0 * nrows), constrained_layout=False)
    plt.subplots_adjust(left=0.04, right=0.93, bottom=0.04, top=0.95, wspace=0.16, hspace=0.16)

    last_cs = {}

    for r, plev in enumerate(pressure_levels):
        for c, var in enumerate(VARS_COLS):
            ax = axs[r, c] if nrows > 1 else axs[c]

            m.drawcoastlines(ax=ax, linewidth=0.45)
            m.drawstates(ax=ax, linewidth=0.25)
            m.drawcountries(ax=ax, linewidth=0.25)

            field = slices[plev][var]
            
            vmin = diff_levels[var][0]
            vmax = diff_levels[var][-1]
            
            cs = m.contourf(
                x, y, field,
                levels=diff_levels[var],
                cmap=CMAP_DIFF,
                extend="both",
                ax=ax
            )
            last_cs[var] = cs

            obs_df = obs_panels[var]
            obs_sel = obs_df[np.isclose(obs_df["vertical"], plev, atol=1e-3)].copy()

            if not obs_sel.empty:
                sc = m.scatter(
                    obs_sel["lon_plot"].values,
                    obs_sel["latitude"].values,
                    c=obs_sel["diff_post_nat"].values,
                    cmap=CMAP_DIFF,
                    vmin=vmin,
                    vmax=vmax,
                    s=OBS_SIZE,
                    edgecolors=OBS_EDGE_COLOR,
                    linewidth=OBS_LINEWIDTH,
                    latlon=True,
                    ax=ax,
                    zorder=30,
                )

                for _, row in obs_sel.iterrows():
                    xt, yt = m(row["lon_plot"], row["latitude"])
                    ax.text(
                        xt, yt,
                        fmt_obs_diff(var, row["diff_post_nat"]),
                        fontsize=OBS_TEXT_FONTSIZE,
                        fontweight="bold",
                        color="k",
                        ha="left",
                        va="bottom",
                        zorder=35,
                        bbox=dict(facecolor="white", edgecolor="none", alpha=0.65, pad=0.4),
                    )

            ax.set_title(f"{var} | {plev/100:.0f} hPa", fontsize=10)

            if c == 0:
                ax.text(
                    0.01, 0.98, f"{plev/100:.0f} hPa",
                    transform=ax.transAxes,
                    ha="left", va="top",
                    fontsize=10, fontweight="bold",
                    bbox=dict(facecolor="white", edgecolor="none", alpha=0.7, pad=0.5),
                )

    for c, var in enumerate(VARS_COLS):
        cb = fig.colorbar(
            last_cs[var],
            ax=axs[:, c] if nrows > 1 else axs[c],
            location="right",
            pad=0.015,
            fraction=0.025
        )
        cb.set_label(f"{var}: Posterior H(x) - Nature Run", fontsize=9)
        cb.ax.yaxis.set_major_locator(MaxNLocator(5))
        cb.ax.yaxis.set_major_formatter(ScalarFormatter(useOffset=False))

    fig.suptitle(f"{dom}: Horizontal Difference Fields (Posterior vs Nature Run, Pure Grid Sampling)", fontsize=13)

    outfig = FIGDIR / f"{dom}_horizontal_posterior_minus_nature_obs_levels_{CYCLE_DIR[36:-19]}_{CYCLE_DIR[-3:-1]}.jpeg"
    plt.savefig(outfig, dpi=300, bbox_inches="tight")
    print(f"Saved {outfig}")

# =============================================================================
# WEST-EAST PRESSURE CROSS-SECTIONS
# =============================================================================
def compute_west_east_cross_section(post_file, nature_file, station_lon, station_lat):
    ds_post = xr.open_dataset(post_file)
    ds_nat = xr.open_dataset(nature_file)

    lat2d, lon2d = get_2d_latlon(ds_post)
    lon2d_w = wrap_lon(lon2d)

    j0, i0 = nearest_gridpoint(lon2d_w, lat2d, station_lon, station_lat)

    lon_line = lon2d_w[j0, :]
    distance_km = 111.32 * np.cos(np.deg2rad(station_lat)) * (lon_line - station_lon)

    p_levels = np.linspace(100000.0, 5000.0, N_XSEC_PRESSURE_LEVELS)

    p_ref = get_pressure_mass(ds_nat, TIME_INDEX)[:, j0, :]

    xsec = {}
    for var in VARS_COLS:
        f_post = get_mass_field(ds_post, var, TIME_INDEX)[:, j0, :]
        f_nat  = get_mass_field(ds_nat, var, TIME_INDEX)[:, j0, :]

        out = np.full((len(p_levels), f_post.shape[1]), np.nan, dtype=float)

        for i in range(f_post.shape[1]):
            post_prof = interp_column_to_pressure(p_ref[:, i], f_post[:, i], p_levels)
            nat_prof  = interp_column_to_pressure(p_ref[:, i], f_nat[:, i],  p_levels)
            out[:, i] = post_prof - nat_prof

        xsec[var] = out

    ds_post.close()
    ds_nat.close()

    return distance_km, p_levels, xsec, j0, i0

def plot_cross_section_figure(dom, post_file, nature_file, station_lon, station_lat, obs_panels):
    distance_km, p_levels, xsec, j0, i0 = compute_west_east_cross_section(
        post_file, nature_file, station_lon, station_lat
    )

    p_hpa = p_levels / 100.0

    fig, axs = plt.subplots(3, 1, figsize=(14, 13), constrained_layout=False)
    plt.subplots_adjust(left=0.08, right=0.90, bottom=0.06, top=0.95, hspace=0.22)

    for idx, var in enumerate(VARS_COLS):
        ax = axs[idx]
        levs = symmetric_levels([xsec[var]], N_DIFF_LEVELS)
        vmin = levs[0]
        vmax = levs[-1]

        cs = ax.contourf(
            distance_km,
            p_hpa,
            xsec[var],
            levels=levs,
            cmap=CMAP_DIFF,
            extend="both"
        )

        ax.axvline(0.0, color="k", linestyle="--", linewidth=1.0)

        obs_df = obs_panels[var].copy()
        if not obs_df.empty:
            ax.scatter(
                np.zeros(len(obs_df)),
                obs_df["vertical"].values / 100.0,
                c=obs_df["diff_post_nat"].values,
                cmap=CMAP_DIFF,
                vmin=vmin,
                vmax=vmax,
                s=OBS_SIZE,
                edgecolors=OBS_EDGE_COLOR,
                linewidths=OBS_LINEWIDTH,
                zorder=30,
            )

        ax.set_title(f"{dom}: West–East cross-section of Posterior - Nature | {var}", fontsize=11)
        ax.set_xlabel("Distance from radiosonde site (km)")
        ax.set_ylabel("Pressure (hPa)")
        ax.invert_yaxis()
        ax.grid(alpha=0.20)

        cb = fig.colorbar(cs, ax=ax, location="right", pad=0.02, fraction=0.035)
        cb.set_label(f"{var}: Posterior - Nature", fontsize=9)
        cb.ax.yaxis.set_major_locator(MaxNLocator(5))
        cb.ax.yaxis.set_major_formatter(ScalarFormatter(useOffset=False))

    fig.suptitle(f"{dom}: West–East pressure cross-sections through radiosonde site (Pure Grid Sampling)", fontsize=13)

    outfig = FIGDIR / f"{dom}_west_east_cross_section_posterior_minus_nature_{CYCLE_DIR[36:-19]}_{CYCLE_DIR[-3:-1]}.jpeg"
    plt.savefig(outfig, dpi=300, bbox_inches="tight")
    print(f"Saved {outfig}")

# =============================================================================
# MAIN
# =============================================================================
def main():
    pressure_levels, station_lon, station_lat, obs_panels = load_obs_profiles(
        OBS_SEQ_FILE,
        OBS_TIME_START,
        OBS_TIME_END,
    )

    print("Target pressure levels (Pa):", pressure_levels)
    print(f"Station location: lon={station_lon:.4f}, lat={station_lat:.4f}")

    for dom in DOMS: 
        post_file = str(Path(CYCLE_DIR) / f"postassim_mean_{dom}.nc")
        nature_file = NATURE_FILES[dom]

        obs_panels = append_gridded_diff(obs_panels, post_file, nature_file)

        plot_horizontal_figure(
            dom=dom,
            post_file=post_file,
            nature_file=nature_file,
            pressure_levels=pressure_levels,
            obs_panels=obs_panels,
        )

        plot_cross_section_figure(
            dom=dom,
            post_file=post_file,
            nature_file=nature_file,
            station_lon=station_lon,
            station_lat=station_lat,
            obs_panels=obs_panels,
        )

if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
from pathlib import Path

# Import the pydartdiags module for reading obs_seq.final files
import pydartdiags.obs_sequence.obs_sequence as obsq

# =============================================================================
# USER OPTIONS
# =============================================================================
CYCLE_DIR   = "/gpfs/home/sa24m/stephen_asare/runs/temp_only_18_eakf/output/2015071418/"

TIME_INDEX = 0

# VERTICAL LEVEL SETTINGS
PLOT_LOWEST_LEVEL = True
ZAGL_M = 3000.0

DOMS = ["d01", "d02"]

# Four columns now: wind is separated into U and V
VARS_COLS = ["THM", "QVAPOR", "U", "V"]

NUM_MEMBERS = 30

# Observation file and time window
OBS_SEQ_FILE = CYCLE_DIR + "obs_seq.final"
OBS_TIME_START = f"2015-07-14 {int(CYCLE_DIR[-3:-1]) - 1:02d}:30:00"
OBS_TIME_END   = f"2015-07-14 {int(CYCLE_DIR[-3:-1]):02d}:30:00"

# Use radiosonde location from the obs_seq.final file
# This location is used to sample the ensemble at the nearest WRF grid point.
OBS_TYPES_FOR_LOCATION = [
    "RADIOSONDE_U_WIND_COMPONENT",
    "RADIOSONDE_V_WIND_COMPONENT",
    "RADIOSONDE_SPECIFIC_HUMIDITY",
    "RADIOSONDE_TEMPERATURE",
]

# Optional observation vertical lines.
# For THM, the radiosonde obs is temperature, not THM, so it is not directly plotted.
# For QVAPOR, radiosonde specific humidity is not exactly QVAPOR mixing ratio.
# Therefore, obs lines are only plotted for U and V by default.
PLOT_OBS_LINE = True

OBS_TYPE_BY_VAR = {
    "U": "RADIOSONDE_U_WIND_COMPONENT",
    "V": "RADIOSONDE_V_WIND_COMPONENT",
    "THM": "RADIOSONDE_TEMPERATURE",
    "QVAPOR": "RADIOSONDE_SPECIFIC_HUMIDITY",
}

# Histogram settings
BINS = 14
ALPHA_PRIOR = 0.6
ALPHA_POST  = 0.2

OUTFIG = f"fig/prior_posterior_ensemble_histograms_{CYCLE_DIR[36:-19]}_{CYCLE_DIR[-3:-1]}.jpeg"


# =============================================================================
# Helper functions
# =============================================================================
def _destagger_to_mass(ds, var, time_index=0):
    da = ds[var].isel(Time=time_index) if "Time" in ds[var].dims else ds[var]

    if var == "U" and "west_east_stag" in da.dims:
        return (
            0.5
            * (
                da.isel(west_east_stag=slice(0, -1))
                + da.isel(west_east_stag=slice(1, None))
            )
        ).rename({"west_east_stag": "west_east"})

    if var == "V" and "south_north_stag" in da.dims:
        return (
            0.5
            * (
                da.isel(south_north_stag=slice(0, -1))
                + da.isel(south_north_stag=slice(1, None))
            )
        ).rename({"south_north_stag": "south_north"})

    if var == "W" and "bottom_top_stag" in da.dims:
        return (
            0.5
            * (
                da.isel(bottom_top_stag=slice(0, -1))
                + da.isel(bottom_top_stag=slice(1, None))
            )
        ).rename({"bottom_top_stag": "bottom_top"})

    return da


def _k_at_zagl(ds_ref, zagl_m, time_index=0):
    PH = ds_ref["PH"].isel(Time=time_index).values if "Time" in ds_ref["PH"].dims else ds_ref["PH"].values
    PHB = ds_ref["PHB"].isel(Time=time_index).values if "Time" in ds_ref["PHB"].dims else ds_ref["PHB"].values

    Zstag = (PH + PHB) / 9.81
    Zmass = 0.5 * (Zstag[:-1, :, :] + Zstag[1:, :, :])

    HGT = ds_ref["HGT"].isel(Time=time_index).values if "Time" in ds_ref["HGT"].dims else ds_ref["HGT"].values

    ZAGL = Zmass - HGT[None, :, :]
    return np.argmin(np.abs(ZAGL - zagl_m), axis=0)


def field_at_zagl(ds_field, ds_ref, var, zagl_m, time_index=0, lowest_level=False):
    da = _destagger_to_mass(ds_field, var, time_index=time_index)

    if "bottom_top" not in da.dims:
        return da.values

    v3d = da.values

    if lowest_level:
        return v3d[0, :, :]

    k = _k_at_zagl(ds_ref, zagl_m=zagl_m, time_index=time_index)
    j, i = np.indices(k.shape)
    return v3d[k, j, i]


def get_latlon_2d(ds):
    lat = ds["XLAT"].isel(Time=0).values if "Time" in ds["XLAT"].dims else ds["XLAT"].values
    lon = ds["XLONG"].isel(Time=0).values if "Time" in ds["XLONG"].dims else ds["XLONG"].values
    return lat, lon


def lon_to_minus180_180(lon):
    return ((lon + 180.0) % 360.0) - 180.0


def nearest_ij_to_obs(ds, obs_lat, obs_lon):
    lat2d, lon2d = get_latlon_2d(ds)
    lon2d_plot = lon_to_minus180_180(lon2d)
    obs_lon_plot = lon_to_minus180_180(obs_lon)

    dist2 = (lat2d - obs_lat) ** 2 + (lon2d_plot - obs_lon_plot) ** 2
    j, i = np.unravel_index(np.nanargmin(dist2), dist2.shape)
    return j, i


def read_obs_dataframe():
    obs_seq = obsq.ObsSequence(OBS_SEQ_FILE)
    df = obs_seq.df.copy()

    if df.empty:
        raise RuntimeError(f"{OBS_SEQ_FILE} loaded, but obs dataframe is empty.")

    df["time"] = pd.to_datetime(df["time"])

    t0 = pd.Timestamp(OBS_TIME_START)
    t1 = pd.Timestamp(OBS_TIME_END)

    df = df[
        (df["time"] >= t0)
        & (df["time"] <= t1)
        & (df["type"].isin(OBS_TYPES_FOR_LOCATION))
    ].copy()

    if df.empty:
        raise RuntimeError("No observations found in the requested time/type window.")

    df["lon_plot"] = lon_to_minus180_180(df["longitude"].values)

    return df


def choose_reference_obs_location(df):
    """
    Uses the mean radiosonde location in the selected window.
    If the radiosonde has several vertical levels, this avoids choosing one level only.
    """
    obs_lat = float(df["latitude"].mean())
    obs_lon = float(df["lon_plot"].mean())
    return obs_lat, obs_lon


def obs_value_for_var(df, var):
    """
    Returns the mean observed value for the matching observation type.
    This is mainly appropriate for U and V.
    THM and QVAPOR are not one-to-one with radiosonde temperature/specific humidity.
    """
    obs_type = OBS_TYPE_BY_VAR.get(var, None)
    if obs_type is None:
        return None

    dff = df[df["type"] == obs_type].copy()
    if dff.empty:
        return None

    if "observation" not in dff.columns:
        return None

    return float(dff["observation"].mean())


def member_file(cdir, stage, member, dom):
    """
    stage should be 'preassim' or 'postassim'
    member is 1-based.
    """
    return cdir / f"{stage}_member_{member:04d}_{dom}.nc"


def ensemble_values_at_point(cdir, stage, dom, var, j, i, ref_file):
    vals = []

    with xr.open_dataset(ref_file) as ds_ref:
        for mem in range(1, NUM_MEMBERS + 1):
            f = member_file(cdir, stage, mem, dom)

            if not f.exists():
                raise FileNotFoundError(f"Missing member file: {f}")

            with xr.open_dataset(f) as ds_mem:
                field2d = field_at_zagl(
                    ds_mem,
                    ds_ref,
                    var,
                    ZAGL_M,
                    time_index=TIME_INDEX,
                    lowest_level=PLOT_LOWEST_LEVEL,
                )

                vals.append(float(field2d[j, i]))

    return np.asarray(vals)


# =============================================================================
# Main processing
# =============================================================================
cdir = Path(CYCLE_DIR)

obs_df = read_obs_dataframe()
obs_lat, obs_lon = choose_reference_obs_location(obs_df)

print(f"Using radiosonde/reference location:")
print(f"  latitude  = {obs_lat:.5f}")
print(f"  longitude = {obs_lon:.5f}")
print(f"  obs count = {len(obs_df)}")

DATA = {}
GRIDPOINTS = {}

for dom in DOMS:
    ref_file = cdir / f"postassim_mean_{dom}.nc"

    if not ref_file.exists():
        raise FileNotFoundError(f"Missing reference file: {ref_file}")

    with xr.open_dataset(ref_file) as ds_ref:
        j, i = nearest_ij_to_obs(ds_ref, obs_lat, obs_lon)

    GRIDPOINTS[dom] = (j, i)

    print(f"{dom}: nearest grid point to radiosonde = j={j}, i={i}")

    for var in VARS_COLS:
        prior_vals = ensemble_values_at_point(
            cdir=cdir,
            stage="preassim",
            dom=dom,
            var=var,
            j=j,
            i=i,
            ref_file=ref_file,
        )

        post_vals = ensemble_values_at_point(
            cdir=cdir,
            stage="postassim",
            dom=dom,
            var=var,
            j=j,
            i=i,
            ref_file=ref_file,
        )

        DATA[(dom, var)] = {
            "prior": prior_vals,
            "post": post_vals,
            "prior_mean": float(np.mean(prior_vals)),
            "post_mean": float(np.mean(post_vals)),
            "prior_sd": float(np.std(prior_vals, ddof=1)),
            "post_sd": float(np.std(post_vals, ddof=1)),
            "obs_value": obs_value_for_var(obs_df, var),
        }


# =============================================================================
# Plot: 2 rows x 4 columns
# =============================================================================
fig, axs = plt.subplots(
    len(DOMS),
    len(VARS_COLS),
    figsize=(20, 8.5),
    constrained_layout=False,
)

plt.subplots_adjust(
    left=0.05,
    right=0.98,
    bottom=0.08,
    top=0.90,
    wspace=0.25,
    hspace=0.28,
)

for r, dom in enumerate(DOMS):
    for c, var in enumerate(VARS_COLS):
        ax = axs[r, c]

        prior = DATA[(dom, var)]["prior"]
        post  = DATA[(dom, var)]["post"]

        all_vals = np.concatenate([prior, post])
        xmin = float(np.nanmin(all_vals))
        xmax = float(np.nanmax(all_vals))

        if xmax <= xmin:
            pad = max(1e-12, 1e-6 * (abs(xmin) + 1.0))
            xmin -= pad
            xmax += pad

        bins = np.linspace(xmin, xmax, BINS)

        ax.hist(
            prior,
            bins=bins,
            alpha=ALPHA_PRIOR,
            density=True,
            label="Prior / preassim",
            edgecolor="k",
            linewidth=0.5,
        )

        ax.hist(
            post,
            bins=bins,
            alpha=ALPHA_POST,
            density=True,
            label="Posterior / postassim",
            edgecolor="k",
            linewidth=0.5,
        )

        prior_mean = DATA[(dom, var)]["prior_mean"]
        post_mean  = DATA[(dom, var)]["post_mean"]
        prior_sd   = DATA[(dom, var)]["prior_sd"]
        post_sd    = DATA[(dom, var)]["post_sd"]

        ax.axvline(prior_mean, linestyle="--", linewidth=1.6, label="Prior mean")
        ax.axvline(post_mean, linestyle="-", linewidth=1.6, label="Posterior mean")

        obs_val = DATA[(dom, var)]["obs_value"]

        if PLOT_OBS_LINE and obs_val is not None and var in ["U", "V"]:
            ax.axvline(obs_val, linestyle=":", linewidth=2.0, label="Observation")

        j, i = GRIDPOINTS[dom]

        ax.set_title(f"{dom}: {var} at j={j}, i={i}", fontsize=11)
        ax.set_xlabel(var)
        ax.set_ylabel("Density")

        txt = (
            f"prior μ={prior_mean:.4g}, σ={prior_sd:.4g}\n"
            f"post μ={post_mean:.4g}, σ={post_sd:.4g}"
        )

        ax.text(
            0.03,
            0.97,
            txt,
            transform=ax.transAxes,
            va="top",
            ha="left",
            fontsize=8,
            bbox=dict(facecolor="white", edgecolor="0.7", alpha=0.85),
        )

        if r == 0 and c == 0:
            ax.legend(fontsize=8, loc="best")

if PLOT_LOWEST_LEVEL:
    lvl_str = "lowest model level k=0"
else:
    lvl_str = f"{ZAGL_M:.0f} m AGL"

fig.suptitle(
    f"Prior vs Posterior Ensemble Histograms | cycle={Path(CYCLE_DIR).name} | {lvl_str}",
    fontsize=13,
)

Path("fig").mkdir(exist_ok=True)
plt.savefig(OUTFIG, dpi=300, bbox_inches="tight")
plt.show()

# print(f"Saved figure: {OUTFIG}")
```

```python
#!/usr/bin/env python3
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

import pydartdiags.obs_sequence.obs_sequence as obsq


# =============================================================================
# USER OPTIONS
# =============================================================================
CYCLE_DIR   = "/gpfs/home/sa24m/stephen_asare/runs/temp_only_kde/2015071418/"
OBS_SEQ_FILE = str(Path(CYCLE_DIR) / "obs_seq.final")

NUM_MEMBERS = 30

# Change this for any cycle you want
CYCLE_CENTER = f"2015-07-14 {CYCLE_DIR[-3:-1]}:00:00"
WINDOW_MINUTES = 30

# Assimilated observation types (Used ONLY as fallback if QC is missing)
ASSIM_TYPES = [
    "RADIOSONDE_TEMPERATURE",
]

# Evaluated / not assimilated observation types (Used ONLY as fallback if QC is missing)
EVAL_TYPES = [
    "RADIOSONDE_SPECIFIC_HUMIDITY",
    "RADIOSONDE_U_WIND_COMPONENT",
    "RADIOSONDE_V_WIND_COMPONENT",
]

OBS_TYPES_TO_PLOT = ASSIM_TYPES + EVAL_TYPES

BINS = 20

OUTDIR = Path("fig")
OUTDIR.mkdir(exist_ok=True)

cycle_tag = pd.Timestamp(CYCLE_CENTER).strftime("%Y%m%d%H")
OUTFIG = OUTDIR / f"obs_space_Hx_prior_posterior_{CYCLE_DIR[36:-19]}_{CYCLE_DIR[-3:-1]}.jpeg"

MISSING_VALUE_THRESHOLD = -1.0e5


# =============================================================================
# Helper functions
# =============================================================================
def time_window(center_time, window_minutes):
    center = pd.Timestamp(center_time)
    dt = pd.Timedelta(minutes=window_minutes)
    return center - dt, center + dt


def member_number(col):
    nums = re.findall(r"\d+", col)
    return int(nums[-1]) if nums else 999999


def find_member_columns(df, stage):
    cols = []
    for c in df.columns:
        cl = c.lower()
        if stage in cl and "ensemble_member" in cl:
            cols.append(c)
    return sorted(cols, key=member_number)


def qc_column(df):
    """
    Find the true DART QC column, strictly ignoring the raw 'qc' column.
    """
    for c in df.columns:
        if c.lower() in ["dart_quality_control", "dart_qc"]:
            return c
    return None


def valid_dart_values(vals, missing_threshold=MISSING_VALUE_THRESHOLD):
    vals = np.asarray(vals, dtype=float).ravel()
    good = (
        np.isfinite(vals)
        & (vals > missing_threshold)
    )
    return vals[good]


def read_obs_df():
    obs_seq = obsq.ObsSequence(OBS_SEQ_FILE)
    df = obs_seq.df.copy()

    if df.empty:
        raise RuntimeError(f"{OBS_SEQ_FILE} loaded, but dataframe is empty.")

    if "time" not in df.columns or "type" not in df.columns:
        raise RuntimeError("Missing 'time' or 'type' column in obs_seq dataframe.")

    df["time"] = pd.to_datetime(df["time"])

    t0, t1 = time_window(CYCLE_CENTER, WINDOW_MINUTES)

    df = df[
        (df["time"] >= t0)
        & (df["time"] <= t1)
        & (df["type"].isin(OBS_TYPES_TO_PLOT))
    ].copy()

    if df.empty:
        raise RuntimeError(f"No observations found between {t0} and {t1}.")

    return df


def safe_mean(vals):
    vals = valid_dart_values(vals)
    if len(vals) == 0:
        return np.nan
    return float(np.mean(vals))


def safe_std(vals):
    vals = valid_dart_values(vals)
    if len(vals) < 2:
        return np.nan
    return float(np.std(vals, ddof=1))


def safe_bins(prior_vals, post_vals, obs_vals, nbins):
    arrays = []
    for vals in [prior_vals, post_vals, obs_vals]:
        vals = valid_dart_values(vals)
        if len(vals) > 0:
            arrays.append(vals)

    if not arrays:
        return None

    all_vals = np.concatenate(arrays)
    xmin = float(np.min(all_vals))
    xmax = float(np.max(all_vals))

    if xmax <= xmin:
        pad = max(1e-12, 1e-6 * (abs(xmin) + 1.0))
        xmin -= pad
        xmax += pad

    return np.linspace(xmin, xmax, nbins)


def format_float(x):
    if np.isfinite(x):
        return f"{x:.4g}"
    return "NA"


def count_removed_values(raw_vals, cleaned_vals):
    raw_vals = np.asarray(raw_vals, dtype=float).ravel()
    n_raw_finite = np.sum(np.isfinite(raw_vals))
    n_clean = len(cleaned_vals)
    return int(n_raw_finite - n_clean)


# =============================================================================
# Main
# =============================================================================
df = read_obs_df()

prior_cols = find_member_columns(df, "prior")
post_cols = find_member_columns(df, "posterior")
qc_col = qc_column(df)

if len(prior_cols) != NUM_MEMBERS or len(post_cols) != NUM_MEMBERS:
    print("WARNING:")
    print(f"  Found {len(prior_cols)} prior member columns")
    print(f"  Found {len(post_cols)} posterior member columns")
    print(f"  Expected {NUM_MEMBERS}")

if not prior_cols or not post_cols:
    raise RuntimeError("Missing prior or posterior ensemble member columns.")

# Dynamically set assimilation status for the terminal printout
df["assim_status"] = "Unknown"
if qc_col is not None:
    for obs_type in df["type"].unique():
        type_qcs = df[df["type"] == obs_type][qc_col].dropna().unique()
        
        # If DART explicitly evaluated ANY of these (QC=1), mark the whole type as evaluated
        if 1.0 in type_qcs or 1 in type_qcs:
            df.loc[df["type"] == obs_type, "assim_status"] = "Not assimilated / evaluated"
        elif 0.0 in type_qcs or 0 in type_qcs:
            df.loc[df["type"] == obs_type, "assim_status"] = "Assimilated"
        else:
            df.loc[df["type"] == obs_type, "assim_status"] = "Not assimilated / evaluated"
else:
    df["assim_status"] = np.where(
        df["type"].isin(ASSIM_TYPES),
        "Assimilated",
        "Not assimilated / evaluated"
    )

print("\nObs count by type/status:")
print(df.groupby(["assim_status", "type"]).size())

print("\nUsing:")
print(f"  OBS_SEQ_FILE      = {OBS_SEQ_FILE}")
print(f"  CYCLE_CENTER      = {CYCLE_CENTER}")
print(f"  WINDOW_MINUTES    = ±{WINDOW_MINUTES}")
print(f"  QC column         = {qc_col}")



import math

num_plots = len(OBS_TYPES_TO_PLOT)

# Create a 2-column grid instead of 1 long row
ncols = 2 if num_plots > 1 else 1 
nrows = math.ceil(num_plots / ncols)

# Use constrained_layout=True to automatically fit the suptitle and remove dead space
fig, axs = plt.subplots(
    nrows,
    ncols,
    figsize=(5.5 * ncols, 5.0 * nrows), 
    constrained_layout=True, 
)

# Flatten axes to make them easy to iterate over
if num_plots == 1:
    axs = [axs]
else:
    axs = axs.flatten()

# Hide any extra empty subplots (e.g., if you had 3 obs types in a 2x2 grid)
for i in range(num_plots, len(axs)):
    axs[i].axis("off")


for ax, obs_type in zip(axs, OBS_TYPES_TO_PLOT):
    d = df[df["type"] == obs_type].copy()

    # =========================================================================
    # NEW CODE: Explicitly grab the observation at vertical == 91000.0
    # =========================================================================
    if not d.empty and "vertical" in d.columns:
        # Filter for the exact vertical level
        d = d[d["vertical"] == 85000.0]
        
        # Take just the first match in case of duplicate observations at 91000.0
        if not d.empty:
            d = d.iloc[[0]] 
    # =========================================================================

    if d.empty:
        ax.axis("off")
        ax.set_title(f"{obs_type}\nNo obs at 91000.0")
        continue

    # Dynamically define title status based on the exact same logic
    # if qc_col is not None and qc_col in d.columns:
# ... [rest of your script continues exactly as before] ...
    # # =========================================================================
    # # NEW CODE: Isolate the single observation at the SECOND lowest vertical level
    # # =========================================================================
    # if not d.empty and "vertical" in d.columns:
    #     # 1. Get all unique, non-missing vertical levels and sort them
    #     unique_levels = sorted(d["vertical"].dropna().unique())
        
    #     # 2. Check if we actually have at least 2 different levels
    #     if len(unique_levels) >= 2:
    #         # Index 1 is the second item (the second smallest level)
    #         second_min_vert = unique_levels[1] 
    #         d = d[d["vertical"] == second_min_vert]
    #     elif len(unique_levels) == 1:
    #         # Fallback: If there's only 1 level total, just use that one
    #         d = d[d["vertical"] == unique_levels[0]]
        
    #     # 3. Take just the first match in case of duplicate observations at this level
    #     if not d.empty:
    #         d = d.iloc[[0]] 
            
    #     # Optional: Print out some info so you know exactly which observation was grabbed
    #     print(f"\nPlotted {obs_type} at vertical level: {d['vertical'].values[0]}")
    # =========================================================================

    # if d.empty:
    #     ax.axis("off")
    #     ax.set_title(f"{obs_type}\nNo obs")
    #     continue

    # Dynamically define title status based on the exact same logic
    if qc_col is not None and qc_col in d.columns:
        unique_qcs = d[qc_col].dropna().unique()
        if 1.0 in unique_qcs or 1 in unique_qcs:
            status = "Not assimilated / evaluated"
        elif 0.0 in unique_qcs or 0 in unique_qcs:
            status = "Assimilated"
        else:
            status = "Not assimilated / evaluated"
    else:
        status = "Assimilated" if obs_type in ASSIM_TYPES else "Not assimilated / evaluated"

    raw_prior = d[prior_cols].to_numpy()
    raw_post = d[post_cols].to_numpy()
    raw_obs = d["observation"].to_numpy()

    prior_vals = valid_dart_values(raw_prior)
    post_vals = valid_dart_values(raw_post)
    obs_vals = valid_dart_values(raw_obs)

    if len(prior_vals) == 0 and len(post_vals) == 0 and len(obs_vals) == 0:
        ax.axis("off")
        ax.set_title(f"{status}\n{obs_type}\nNo valid values")
        continue

    bins = safe_bins(prior_vals, post_vals, obs_vals, BINS)

    if bins is None:
        ax.axis("off")
        ax.set_title(f"{status}\n{obs_type}\nNo valid bins")
        continue

    if len(prior_vals) > 0:
        ax.hist(
            prior_vals,
            bins=bins,
            density=True,
            alpha=0.55,
            edgecolor="k",
            linewidth=0.5,
            label="Prior H(x)",
        )

    if len(post_vals) > 0:
        ax.hist(
            post_vals,
            bins=bins,
            density=True,
            alpha=0.35,
            edgecolor="k",
            linewidth=0.5,
            label="Posterior H(x)",
        )

    prior_mean = safe_mean(prior_vals)
    post_mean = safe_mean(post_vals)
    prior_sd = safe_std(prior_vals)
    post_sd = safe_std(post_vals)

    if np.isfinite(prior_mean):
        ax.axvline(
            prior_mean,
            linestyle="--",
            linewidth=1.7,
            label="Prior mean",
        )

    if np.isfinite(post_mean):
        ax.axvline(
            post_mean,
            linestyle="-",
            linewidth=1.7,
            label="Posterior mean",
        )

    if len(obs_vals) > 0:
        obs_mean = safe_mean(obs_vals)
        if np.isfinite(obs_mean):
            ax.axvline(
                obs_mean,
                linestyle=":",
                linewidth=2.2,
                label="Obs mean",
            )

        ymax = ax.get_ylim()[1]
        ax.plot(
            obs_vals,
            np.zeros_like(obs_vals) + 0.02 * ymax,
            "|",
            markersize=12,
            markeredgewidth=1.5,
            label="Obs values",
        )

    title = f"{status}\n{obs_type}"
    ax.set_title(title, fontsize=10)
    ax.set_xlabel("Observation-space value, H(x)")
    ax.set_ylabel("Density")

    txt = (
        f"Nobs = {len(d)}\n"
        f"prior μ = {format_float(prior_mean)}\n"
        f"prior σ = {format_float(prior_sd)}\n"
        f"post μ = {format_float(post_mean)}\n"
        f"post σ = {format_float(post_sd)}"
    )

    ax.text(
        0.03,
        0.97,
        txt,
        transform=ax.transAxes,
        va="top",
        ha="left",
        fontsize=8,
        bbox=dict(facecolor="white", edgecolor="0.7", alpha=0.85),
    )


for ax in axs:
    handles, labels = ax.get_legend_handles_labels()
    if handles:
        ax.legend(fontsize=8, loc="best")
        break


fig.suptitle(
    (
        "Observation-space prior/posterior ensemble distributions of H(x)\n"
        f"cycle center = {CYCLE_CENTER}, window = ±{WINDOW_MINUTES} min"
    ),
    fontsize=14,
)

# plt.savefig(OUTFIG, dpi=300, bbox_inches="tight")
plt.show()

```

```python
#!/usr/bin/env python3
import pandas as pd
from pathlib import Path
import pydartdiags.obs_sequence.obs_sequence as obsq

# =============================================================================
# USER OPTIONS
# Add the main directories you want to scan here.
# The script will look recursively for 'obs_seq.final' files inside these.
# =============================================================================
DIRECTORIES_TO_SCAN = [
    "/gpfs/home/sa24m/stephen_asare/runs/",
]

def find_qc_column(df):
    """Find the true DART QC column, strictly ignoring the raw 'qc' column."""
    for c in df.columns:
        if c.lower() in ["dart_quality_control", "dart_qc"]:
            return c
    return None

def analyze_obs_seq(file_path):
    print(f"\n{'-'*70}")
    print(f"FILE: {file_path}")
    print(f"{'-'*70}")
    
    try:
        obs_seq = obsq.ObsSequence(str(file_path))
        df = obs_seq.df
    except Exception as e:
        print(f"  [!] Failed to load file: {e}")
        return

    if df.empty:
        print("  [!] File loaded, but dataframe is empty.")
        return

    print(f"  Total Observations: {len(df)}")
    
    qc_col = find_qc_column(df)
    if not qc_col:
        print("  [!] Could not find DART QC column. Available columns:")
        print(f"      {df.columns.tolist()}")
        return
        
    print(f"  Using QC Column: '{qc_col}'\n")
    
    # Breakdown by Type and QC Value
    summary = df.groupby(['type', qc_col]).size().reset_index(name='count')
    print("  === QC Breakdown ===")
    print(summary.to_string(index=False))

def main():
    print("Starting DART Data Diagnostic Scan...")
    found_files = 0
    
    for dir_str in DIRECTORIES_TO_SCAN:
        base_path = Path(dir_str)
        if not base_path.exists():
            print(f"\n[!] Directory not found: {base_path}")
            continue
            
        # Search recursively for obs_seq.final files
        print(f"\nScanning: {base_path} ...")
        for obs_file in base_path.rglob("obs_seq.final"):
            found_files += 1
            analyze_obs_seq(obs_file)
            
    if found_files == 0:
        print(f"\nNo 'obs_seq.final' files found. Check your directory paths.")
    else:
        print(f"\n{'-'*70}")
        print(f"Scan complete. Analyzed {found_files} files.")

if __name__ == "__main__":
    main()
```

```python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from pathlib import Path
from matplotlib.lines import Line2D

import pydartdiags.obs_sequence.obs_sequence as obsq


# =============================================================================
# USER SETTINGS
# =============================================================================
OUTPUT_ROOT = Path("/gpfs/home/sa24m/stephen_asare/runs/all_eakf/output/")

CYCLE_DIRS = [
    OUTPUT_ROOT / "2015071412",
    OUTPUT_ROOT / "2015071415",
    OUTPUT_ROOT / "2015071418",
    OUTPUT_ROOT / "2015071421",
]

OBS_SEQ_NAME = "obs_seq.final"

OBS_TYPES_TO_PLOT = [
    "RADIOSONDE_TEMPERATURE",
    "RADIOSONDE_SPECIFIC_HUMIDITY",
    "RADIOSONDE_U_WIND_COMPONENT",
    "RADIOSONDE_V_WIND_COMPONENT",
]

TYPE_LABELS = {
    "RADIOSONDE_TEMPERATURE": "Temperature",
    "RADIOSONDE_SPECIFIC_HUMIDITY": "Specific humidity",
    "RADIOSONDE_U_WIND_COMPONENT": "U wind",
    "RADIOSONDE_V_WIND_COMPONENT": "V wind",
}

MISSING_VALUE_THRESHOLD = -1.0e5

TARGET_LON = None
TARGET_LAT = None
STATION_TOL_DEG = 0.05

PLOT_ALL_STATIONS = False

FIGDIR = Path("fig_obs_seq_update_check")
FIGDIR.mkdir(exist_ok=True)

SAVE_FIG = True
SHOW_FIG = True
DPI = 300


# =============================================================================
# HELPERS
# =============================================================================
def wrap_lon(lon):
    lon = np.asarray(lon, dtype=float)
    return ((lon + 180.0) % 360.0) - 180.0


def cycle_time_from_dir(cycle_dir):
    name = Path(cycle_dir).name
    m = re.search(r"(\d{10})", name)

    if m is None:
        raise ValueError(f"Could not parse YYYYMMDDHH from {cycle_dir}")

    return pd.to_datetime(m.group(1), format="%Y%m%d%H")


def model_cycle_time_bounds(cycle_dirs):
    times = pd.to_datetime([cycle_time_from_dir(cdir) for cdir in cycle_dirs])
    return times.min(), times.max(), times


def clean_array(vals):
    vals = np.asarray(vals, dtype=float).copy()
    vals[~np.isfinite(vals)] = np.nan
    vals[vals <= MISSING_VALUE_THRESHOLD] = np.nan
    return vals


def format_pressure_label(p_pa):
    return f"{p_pa / 100.0:.0f} hPa"


def sanitize_name(s):
    return (
        str(s)
        .replace("/", "_")
        .replace(" ", "_")
        .replace(":", "")
        .replace(",", "")
        .replace(".", "p")
    )


def read_one_obs_seq(cycle_dir):
    cycle_dir = Path(cycle_dir)
    obs_file = cycle_dir / OBS_SEQ_NAME
    cycle_time = cycle_time_from_dir(cycle_dir)

    if not obs_file.exists():
        raise FileNotFoundError(f"Missing obs_seq.final: {obs_file}")

    obs_seq = obsq.ObsSequence(str(obs_file))
    df = obs_seq.df.copy()

    if df.empty:
        raise RuntimeError(f"{obs_file} loaded, but dataframe is empty.")

    required = [
        "type",
        "time",
        "longitude",
        "latitude",
        "vertical",
        "observation",
        "prior_ensemble_mean",
        "posterior_ensemble_mean",
    ]

    for col in required:
        if col not in df.columns:
            raise RuntimeError(f"Required column missing from {obs_file}: {col}")

    df = df[df["type"].isin(OBS_TYPES_TO_PLOT)].copy()

    df["time"] = pd.to_datetime(df["time"])
    df["cycle_time"] = cycle_time
    df["cycle_tag"] = cycle_dir.name
    df["cycle_dir"] = str(cycle_dir)

    df["longitude"] = df["longitude"].astype(float)
    df["latitude"] = df["latitude"].astype(float)
    df["vertical"] = df["vertical"].astype(float)

    df["lon_plot"] = wrap_lon(df["longitude"].values)

    df["obs_clean"] = clean_array(df["observation"].values)
    df["prior_clean"] = clean_array(df["prior_ensemble_mean"].values)
    df["post_clean"] = clean_array(df["posterior_ensemble_mean"].values)

    df["prior_minus_obs"] = df["prior_clean"] - df["obs_clean"]
    df["post_minus_obs"] = df["post_clean"] - df["obs_clean"]
    df["post_minus_prior"] = df["post_clean"] - df["prior_clean"]

    df["abs_prior_err"] = np.abs(df["prior_minus_obs"])
    df["abs_post_err"] = np.abs(df["post_minus_obs"])
    df["abs_update"] = np.abs(df["post_minus_prior"])

    return df


def collect_obs_seq_updates(cycle_dirs):
    frames = []

    for cycle_dir in cycle_dirs:
        print(f"Reading {Path(cycle_dir) / OBS_SEQ_NAME}")
        frames.append(read_one_obs_seq(cycle_dir))

    df = pd.concat(frames, ignore_index=True)

    # Keep only obs whose valid time is within the plotted cycle range.
    # This prevents observations after 21 UTC from appearing.
    tmin, tmax, _ = model_cycle_time_bounds(cycle_dirs)

    before = len(df)
    df = df[(df["time"] >= tmin) & (df["time"] <= tmax)].copy()
    after = len(df)

    print("\nObs rows:")
    print(f"  before time filter = {before}")
    print(f"  after time filter  = {after}")
    print(f"  time range kept    = {tmin} to {tmax}")

    print("\nCounts by cycle and type:")
    print(df.groupby(["cycle_tag", "type"]).size())

    return df


def get_station_list(df):
    if TARGET_LON is not None and TARGET_LAT is not None:
        return [(float(wrap_lon(TARGET_LON)), float(TARGET_LAT))]

    stations = (
        df.assign(
            station_lon=df["lon_plot"].round(4),
            station_lat=df["latitude"].round(4),
        )
        .groupby(["station_lon", "station_lat"])
        .size()
        .reset_index(name="count")
        .sort_values("count", ascending=False)
    )

    print("\nDetected stations:")
    print(stations)

    if PLOT_ALL_STATIONS:
        return list(zip(stations["station_lon"], stations["station_lat"]))

    return [(float(stations.iloc[0]["station_lon"]), float(stations.iloc[0]["station_lat"]))]


def observations_for_station(df, lon0, lat0):
    dlon = df["lon_plot"].values - float(wrap_lon(lon0))
    dlat = df["latitude"].values - float(lat0)
    dist = np.sqrt(dlon**2 + dlat**2)

    return df[dist <= STATION_TOL_DEG].copy()


def summarize_updates(df):
    """
    Print update diagnostics by cycle/type and type/level.
    """
    print("\n" + "=" * 100)
    print("SUMMARY BY CYCLE AND TYPE")
    print("=" * 100)

    summary_cycle_type = (
        df.groupby(["cycle_tag", "type"])
        .agg(
            n=("obs_clean", "count"),
            mean_abs_prior_err=("abs_prior_err", "mean"),
            mean_abs_post_err=("abs_post_err", "mean"),
            mean_abs_update=("abs_update", "mean"),
            max_abs_update=("abs_update", "max"),
        )
        .reset_index()
    )

    summary_cycle_type["improved"] = (
        summary_cycle_type["mean_abs_post_err"]
        < summary_cycle_type["mean_abs_prior_err"]
    )

    print(summary_cycle_type.to_string(index=False))

    print("\n" + "=" * 100)
    print("SUMMARY BY CYCLE, TYPE, AND PRESSURE LEVEL")
    print("=" * 100)

    summary_level = (
        df.groupby(["cycle_tag", "type", "vertical"])
        .agg(
            n=("obs_clean", "count"),
            mean_obs=("obs_clean", "mean"),
            mean_prior=("prior_clean", "mean"),
            mean_post=("post_clean", "mean"),
            mean_abs_prior_err=("abs_prior_err", "mean"),
            mean_abs_post_err=("abs_post_err", "mean"),
            mean_abs_update=("abs_update", "mean"),
            max_abs_update=("abs_update", "max"),
        )
        .reset_index()
    )

    summary_level["level_hpa"] = summary_level["vertical"] / 100.0
    summary_level["improved"] = (
        summary_level["mean_abs_post_err"]
        < summary_level["mean_abs_prior_err"]
    )

    print(
        summary_level[
            [
                "cycle_tag",
                "type",
                "level_hpa",
                "n",
                "mean_obs",
                "mean_prior",
                "mean_post",
                "mean_abs_prior_err",
                "mean_abs_post_err",
                "mean_abs_update",
                "max_abs_update",
                "improved",
            ]
        ].to_string(index=False)
    )

    return summary_cycle_type, summary_level


def aggregate_for_plot(df):
    """
    One value per cycle/type/pressure/station.

    This plots cycle-time summaries of actual DART observation-space prior/posterior.
    """
    group_cols = [
        "cycle_time",
        "cycle_tag",
        "type",
        "vertical",
        "lon_plot",
        "latitude",
    ]

    out = (
        df.groupby(group_cols, as_index=False)
        .agg(
            obs_mean=("obs_clean", "mean"),
            prior_mean=("prior_clean", "mean"),
            post_mean=("post_clean", "mean"),
            mean_abs_prior_err=("abs_prior_err", "mean"),
            mean_abs_post_err=("abs_post_err", "mean"),
            mean_abs_update=("abs_update", "mean"),
            n_obs=("obs_clean", "count"),
        )
    )

    return out


def plot_obs_seq_update_by_level(df_plot, lon0, lat0):
    """
    Grid layout:
      rows    = pressure levels
      columns = observation types

    Plots DART observation-space:
      prior_ensemble_mean
      posterior_ensemble_mean
      observation
    """
    obs_types = [t for t in OBS_TYPES_TO_PLOT if t in df_plot["type"].unique()]
    levels = np.sort(df_plot["vertical"].dropna().unique())[::-1]

    nlev = len(levels)
    nvar = len(obs_types)

    fig_width = max(4.2 * nvar, 12)
    fig_height = max(2.3 * nlev, 8)

    fig, axs = plt.subplots(
        nlev,
        nvar,
        figsize=(fig_width, fig_height),
        sharex=True,
        constrained_layout=False,
    )

    if nlev == 1 and nvar == 1:
        axs = np.array([[axs]])
    elif nlev == 1:
        axs = np.array([axs])
    elif nvar == 1:
        axs = np.array([[ax] for ax in axs])

    plt.subplots_adjust(
        left=0.07,
        right=0.84,
        bottom=0.07,
        top=0.90,
        hspace=0.34,
        wspace=0.25,
    )

    tmin, tmax, tick_times = model_cycle_time_bounds(CYCLE_DIRS)
    xpad = pd.Timedelta(minutes=25)
    xlim_min = tmin - xpad
    xlim_max = tmax + xpad

    for r, level in enumerate(levels):
        level_label = format_pressure_label(level)

        for c, typ in enumerate(obs_types):
            ax = axs[r, c]

            d = df_plot[
                (df_plot["type"] == typ)
                & (np.isclose(df_plot["vertical"], level, atol=1e-6))
            ].copy()

            d = d.sort_values("cycle_time")

            if not d.empty:
                ax.plot(
                    d["cycle_time"],
                    d["prior_mean"],
                    color="black",
                    linestyle="--",
                    linewidth=1.8,
                    label="Prior",
                )

                ax.plot(
                    d["cycle_time"],
                    d["post_mean"],
                    color="black",
                    linestyle="-",
                    linewidth=2.1,
                    label="Posterior",
                )

                ax.plot(
                    d["cycle_time"],
                    d["obs_mean"],
                    color="tab:red",
                    linestyle="none",
                    marker="o",
                    markersize=4.2,
                    markeredgecolor="k",
                    markeredgewidth=0.35,
                    alpha=0.95,
                    label="Observation",
                )

                # Annotate nobs if useful
                for _, row in d.iterrows():
                    ax.text(
                        row["cycle_time"],
                        row["obs_mean"],
                        f"{int(row['n_obs'])}",
                        fontsize=6,
                        ha="left",
                        va="bottom",
                        color="tab:red",
                    )

            if r == 0:
                ax.set_title(TYPE_LABELS.get(typ, typ), fontsize=10, fontweight="bold")

            if c == 0:
                ax.set_ylabel(f"{level_label}\nObs-space value")

            ax.text(
                0.02,
                0.92,
                level_label,
                transform=ax.transAxes,
                ha="left",
                va="top",
                fontsize=8,
                bbox=dict(facecolor="white", edgecolor="0.7", alpha=0.75, pad=1.5),
            )

            ax.grid(alpha=0.25)

            ax.set_xlim(xlim_min, xlim_max)
            ax.set_xticks(tick_times)
            ax.xaxis.set_major_formatter(mdates.DateFormatter("%H"))

    for ax in axs[-1, :]:
        ax.set_xlabel("UTC hour")

    legend_handles = [
        Line2D(
            [0], [0],
            color="black",
            linestyle="--",
            linewidth=1.8,
            label="DART prior ensemble mean",
        ),
        Line2D(
            [0], [0],
            color="black",
            linestyle="-",
            linewidth=2.1,
            label="DART posterior ensemble mean",
        ),
        Line2D(
            [0], [0],
            color="tab:red",
            linestyle="none",
            marker="o",
            markersize=4.2,
            markeredgecolor="k",
            linewidth=0.0,
            label="Observation",
        ),
    ]

    fig.legend(
        handles=legend_handles,
        loc="upper left",
        bbox_to_anchor=(0.855, 0.985),
        fontsize=9,
        frameon=True,
        title="Obs-space quantity",
    )

    fig.suptitle(
        (
            "DART Observation-Space Prior/Posterior Update Check\n"
            f"station lon={lon0:.4f}, lat={lat0:.4f}"
        ),
        fontsize=14,
    )

    outfig = FIGDIR / (
        f"obs_seq_update_check"
        f"_lon_{sanitize_name(f'{lon0:.4f}')}"
        f"_lat_{sanitize_name(f'{lat0:.4f}')}.png"
    )

    # if SAVE_FIG:
    #     plt.savefig(outfig, dpi=DPI, bbox_inches="tight")
    #     print(f"Saved {outfig}")

    if SHOW_FIG:
        plt.show()
    else:
        plt.close(fig)


# =============================================================================
# MAIN
# =============================================================================
def main():
    print("\nCycle directories:")
    for d in CYCLE_DIRS:
        print(f"  {d}")

    df = collect_obs_seq_updates(CYCLE_DIRS)

    stations = get_station_list(df)

    for lon0, lat0 in stations:
        print(f"\nProcessing station lon={lon0:.4f}, lat={lat0:.4f}")

        d_station = observations_for_station(df, lon0, lat0)

        if d_station.empty:
            print("WARNING: no observations at this station.")
            continue

        summarize_updates(d_station)

        d_plot = aggregate_for_plot(d_station)

        print("\nAggregated plotting rows:")
        print(d_plot.groupby(["cycle_tag", "type"]).size())

        plot_obs_seq_update_by_level(d_plot, lon0, lat0)


main()
```

```python
import re
import numpy as np
import pandas as pd
from pathlib import Path

import pydartdiags.obs_sequence.obs_sequence as obsq


# =============================================================================
# USER SETTINGS
# =============================================================================
OUTPUT_ROOT = Path("/gpfs/home/sa24m/scratch/tqprof/run2/osse_out/dart_cycle/output")

CYCLE_DIRS = [
    OUTPUT_ROOT / "2015071412",
    OUTPUT_ROOT / "2015071415",
    OUTPUT_ROOT / "2015071418",
    OUTPUT_ROOT / "2015071421",
]

OBS_SEQ_NAME = "obs_seq.final"

OBS_TYPES = [
    "RADIOSONDE_TEMPERATURE",
    "RADIOSONDE_SPECIFIC_HUMIDITY",
    "RADIOSONDE_U_WIND_COMPONENT",
    "RADIOSONDE_V_WIND_COMPONENT",
]

MISSING_VALUE_THRESHOLD = -1.0e5

# Treat updates smaller than this as effectively zero
UPDATE_TOL = 1.0e-8

# Optional station filter
TARGET_LON = None
TARGET_LAT = None
STATION_TOL_DEG = 0.05


# =============================================================================
# HELPERS
# =============================================================================
def wrap_lon(lon):
    lon = np.asarray(lon, dtype=float)
    return ((lon + 180.0) % 360.0) - 180.0


def cycle_time_from_dir(cycle_dir):
    name = Path(cycle_dir).name
    m = re.search(r"(\d{10})", name)

    if m is None:
        raise ValueError(f"Could not parse YYYYMMDDHH from {cycle_dir}")

    return pd.to_datetime(m.group(1), format="%Y%m%d%H")


def clean_array(vals):
    vals = np.asarray(vals, dtype=float).copy()
    vals[~np.isfinite(vals)] = np.nan
    vals[vals <= MISSING_VALUE_THRESHOLD] = np.nan
    return vals


def read_one_obs_seq(cycle_dir):
    cycle_dir = Path(cycle_dir)
    obs_file = cycle_dir / OBS_SEQ_NAME

    print(f"Reading {obs_file}")

    obs_seq = obsq.ObsSequence(str(obs_file))
    df = obs_seq.df.copy()

    if df.empty:
        raise RuntimeError(f"{obs_file} loaded, but dataframe is empty.")

    required = [
        "type",
        "time",
        "longitude",
        "latitude",
        "vertical",
        "observation",
        "prior_ensemble_mean",
        "posterior_ensemble_mean",
    ]

    for col in required:
        if col not in df.columns:
            raise RuntimeError(f"Missing required column in {obs_file}: {col}")

    df = df[df["type"].isin(OBS_TYPES)].copy()

    df["time"] = pd.to_datetime(df["time"])
    df["cycle_time"] = cycle_time_from_dir(cycle_dir)
    df["cycle_tag"] = cycle_dir.name

    df["longitude"] = df["longitude"].astype(float)
    df["latitude"] = df["latitude"].astype(float)
    df["vertical"] = df["vertical"].astype(float)
    df["lon_plot"] = wrap_lon(df["longitude"].values)

    df["obs_clean"] = clean_array(df["observation"].values)
    df["prior_clean"] = clean_array(df["prior_ensemble_mean"].values)
    df["post_clean"] = clean_array(df["posterior_ensemble_mean"].values)

    df["update"] = df["post_clean"] - df["prior_clean"]
    df["abs_update"] = np.abs(df["update"])

    df["valid_prior_post"] = (
        np.isfinite(df["prior_clean"])
        & np.isfinite(df["post_clean"])
    )

    df["updated"] = (
        df["valid_prior_post"]
        & np.isfinite(df["abs_update"])
        & (df["abs_update"] > UPDATE_TOL)
    )

    return df


def collect_all_obs(cycle_dirs):
    frames = []

    for cycle_dir in cycle_dirs:
        frames.append(read_one_obs_seq(cycle_dir))

    df = pd.concat(frames, ignore_index=True)

    # Restrict to cycle time range only
    cycle_times = pd.to_datetime([cycle_time_from_dir(cdir) for cdir in cycle_dirs])
    tmin = cycle_times.min()
    tmax = cycle_times.max()

    df = df[
        (df["time"] >= tmin)
        & (df["time"] <= tmax)
    ].copy()

    # Optional station filter
    if TARGET_LON is not None and TARGET_LAT is not None:
        lon0 = float(wrap_lon(TARGET_LON))
        lat0 = float(TARGET_LAT)

        dist = np.sqrt(
            (df["lon_plot"].values - lon0) ** 2
            + (df["latitude"].values - lat0) ** 2
        )

        df = df[dist <= STATION_TOL_DEG].copy()

    return df


def pressure_label(pa):
    return f"{pa / 100.0:.0f} hPa"


# =============================================================================
# MAIN VERIFICATION
# =============================================================================
df = collect_all_obs(CYCLE_DIRS)

print("\nRows after filtering:")
print(len(df))

print("\nAvailable pressure levels:")
levels = np.sort(df["vertical"].dropna().unique())[::-1]
print([pressure_label(p) for p in levels])

print("\nCounts by cycle/type/level:")
counts = (
    df.groupby(["cycle_tag", "type", "vertical"])
    .size()
    .reset_index(name="n")
)
counts["level_hpa"] = counts["vertical"] / 100.0
print(counts[["cycle_tag", "type", "level_hpa", "n"]].to_string(index=False))


# =============================================================================
# VERIFY VALID PRIOR/POSTERIOR AND UPDATE BY LEVEL
# =============================================================================
summary = (
    df.groupby(["cycle_tag", "type", "vertical"])
    .agg(
        n_obs=("obs_clean", "count"),
        n_valid_prior_post=("valid_prior_post", "sum"),
        n_updated=("updated", "sum"),
        mean_abs_update=("abs_update", "mean"),
        max_abs_update=("abs_update", "max"),
        mean_prior=("prior_clean", "mean"),
        mean_post=("post_clean", "mean"),
        mean_obs=("obs_clean", "mean"),
    )
    .reset_index()
)

summary["level_hpa"] = summary["vertical"] / 100.0
summary["has_valid_prior_post"] = summary["n_valid_prior_post"] > 0
summary["has_update"] = summary["n_updated"] > 0

print("\n" + "=" * 120)
print("VALID PRIOR/POSTERIOR AND UPDATE SUMMARY")
print("=" * 120)

print(
    summary[
        [
            "cycle_tag",
            "type",
            "level_hpa",
            "n_obs",
            "n_valid_prior_post",
            "n_updated",
            "mean_abs_update",
            "max_abs_update",
            "has_valid_prior_post",
            "has_update",
        ]
    ].to_string(index=False)
)


# =============================================================================
# LOWEST NEAR-SURFACE LEVEL WITH VALID PRIOR/POSTERIOR
# =============================================================================
# Higher pressure = closer to surface.
valid_summary = summary[summary["has_valid_prior_post"]].copy()

lowest_valid = (
    valid_summary
    .sort_values("vertical", ascending=False)
    .groupby(["cycle_tag", "type"], as_index=False)
    .first()
)

print("\n" + "=" * 120)
print("LOWEST NEAR-SURFACE LEVEL WITH VALID PRIOR/POSTERIOR")
print("=" * 120)

if lowest_valid.empty:
    print("No valid prior/posterior values found.")
else:
    lowest_valid["lowest_valid_level"] = lowest_valid["vertical"].apply(pressure_label)

    print(
        lowest_valid[
            [
                "cycle_tag",
                "type",
                "lowest_valid_level",
                "n_valid_prior_post",
                "mean_abs_update",
                "max_abs_update",
            ]
        ].to_string(index=False)
    )


# =============================================================================
# LOWEST NEAR-SURFACE LEVEL WITH ACTUAL NONZERO UPDATE
# =============================================================================
updated_summary = summary[summary["has_update"]].copy()

lowest_updated = (
    updated_summary
    .sort_values("vertical", ascending=False)
    .groupby(["cycle_tag", "type"], as_index=False)
    .first()
)

print("\n" + "=" * 120)
print("LOWEST NEAR-SURFACE LEVEL WITH NONZERO UPDATE")
print("=" * 120)

if lowest_updated.empty:
    print("No nonzero updates found with UPDATE_TOL =", UPDATE_TOL)
else:
    lowest_updated["lowest_updated_level"] = lowest_updated["vertical"].apply(pressure_label)

    print(
        lowest_updated[
            [
                "cycle_tag",
                "type",
                "lowest_updated_level",
                "n_updated",
                "mean_abs_update",
                "max_abs_update",
            ]
        ].to_string(index=False)
    )


# =============================================================================
# OVERALL LOWEST VALID / UPDATED LEVEL ACROSS ALL CYCLES AND TYPES
# =============================================================================
overall = (
    summary.groupby("vertical")
    .agg(
        n_obs=("n_obs", "sum"),
        n_valid_prior_post=("n_valid_prior_post", "sum"),
        n_updated=("n_updated", "sum"),
        mean_abs_update=("mean_abs_update", "mean"),
        max_abs_update=("max_abs_update", "max"),
    )
    .reset_index()
    .sort_values("vertical", ascending=False)
)

overall["level_hpa"] = overall["vertical"] / 100.0
overall["has_valid_prior_post"] = overall["n_valid_prior_post"] > 0
overall["has_update"] = overall["n_updated"] > 0

print("\n" + "=" * 120)
print("OVERALL LEVEL SUMMARY")
print("=" * 120)

print(
    overall[
        [
            "level_hpa",
            "n_obs",
            "n_valid_prior_post",
            "n_updated",
            "mean_abs_update",
            "max_abs_update",
            "has_valid_prior_post",
            "has_update",
        ]
    ].to_string(index=False)
)

valid_levels = overall[overall["has_valid_prior_post"]]["vertical"].to_numpy()
updated_levels = overall[overall["has_update"]]["vertical"].to_numpy()

print("\n" + "=" * 120)
print("FINAL ANSWER")
print("=" * 120)

if len(valid_levels) > 0:
    lowest_valid_overall = np.nanmax(valid_levels)
    print(
        "Lowest near-surface level with valid DART prior/posterior:",
        pressure_label(lowest_valid_overall),
    )
else:
    print("No level has valid DART prior/posterior.")

if len(updated_levels) > 0:
    lowest_updated_overall = np.nanmax(updated_levels)
    print(
        "Lowest near-surface level with nonzero DART update:",
        pressure_label(lowest_updated_overall),
    )
else:
    print("No level has nonzero DART update.")
```

```python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import matplotlib as mpl
from matplotlib.lines import Line2D
from pathlib import Path

import pydartdiags.obs_sequence.obs_sequence as obsq


# =============================================================================
# USER SETTINGS
# =============================================================================
OUTPUT_ROOT = Path("/gpfs/home/sa24m/stephen_asare/runs//output")

CYCLE_DIRS = [
    OUTPUT_ROOT / "2015071412",
    OUTPUT_ROOT / "2015071415",
    OUTPUT_ROOT / "2015071418",
    OUTPUT_ROOT / "2015071421",
]

OBS_SEQ_NAME = "obs_seq.final"

OBS_TYPES_TO_PLOT = [
    "RADIOSONDE_TEMPERATURE",
    "RADIOSONDE_SPECIFIC_HUMIDITY",
    "RADIOSONDE_U_WIND_COMPONENT",
    "RADIOSONDE_V_WIND_COMPONENT",
]

OBS_LABELS = {
    "RADIOSONDE_TEMPERATURE": "Temperature (K)",
    "RADIOSONDE_SPECIFIC_HUMIDITY": "Specific humidity (kg/kg)",
    "RADIOSONDE_U_WIND_COMPONENT": "U wind (m/s)",
    "RADIOSONDE_V_WIND_COMPONENT": "V wind (m/s)",
}

TARGET_LON = None
TARGET_LAT = None
STATION_TOL_DEG = 0.05

MISSING_VALUE_THRESHOLD = -1.0e5
X_PAD_MINUTES = 25

FIGDIR = Path("fig")
FIGDIR.mkdir(exist_ok=True)

SAVE_FIG = True
SHOW_FIG = True
DPI = 300


# =============================================================================
# HELPERS
# =============================================================================
def wrap_lon(lon):
    lon = np.asarray(lon, dtype=float)
    return ((lon + 180.0) % 360.0) - 180.0


def clean_array(vals):
    vals = np.asarray(vals, dtype=float).copy()
    vals[~np.isfinite(vals)] = np.nan
    vals[vals <= MISSING_VALUE_THRESHOLD] = np.nan
    return vals


def cycle_time_from_dir(cycle_dir):
    name = Path(cycle_dir).name
    m = re.search(r"(\d{10})", name)

    if m is None:
        raise ValueError(f"Could not parse cycle time from {cycle_dir}")

    return pd.to_datetime(m.group(1), format="%Y%m%d%H")


def cycle_time_bounds(cycle_dirs):
    times = pd.to_datetime([cycle_time_from_dir(c) for c in cycle_dirs])
    return times.min(), times.max(), times


def format_pressure_label(p_pa):
    return f"{p_pa / 100.0:.0f} hPa"


def sanitize_name(s):
    return (
        str(s)
        .replace("/", "_")
        .replace(" ", "_")
        .replace(":", "")
        .replace(",", "")
        .replace(".", "p")
    )


# =============================================================================
# READ OBS_SEQ.FINAL
# =============================================================================
def read_obs_file(obs_seq_file):
    obs_seq = obsq.ObsSequence(str(obs_seq_file))
    df = obs_seq.df.copy()

    if df.empty:
        return df

    required = [
        "time",
        "type",
        "longitude",
        "latitude",
        "vertical",
        "observation",
        "prior_ensemble_mean",
        "posterior_ensemble_mean",
        "DART_quality_control",
    ]

    for col in required:
        if col not in df.columns:
            raise RuntimeError(f"Missing required column: {col}")

    df["time"] = pd.to_datetime(df["time"])
    df["longitude"] = df["longitude"].astype(float)
    df["latitude"] = df["latitude"].astype(float)
    df["vertical"] = df["vertical"].astype(float)

    df["observation"] = df["observation"].astype(float)
    df["prior_ensemble_mean"] = df["prior_ensemble_mean"].astype(float)
    df["posterior_ensemble_mean"] = df["posterior_ensemble_mean"].astype(float)
    df["DART_quality_control"] = df["DART_quality_control"].astype(float)

    df["lon_plot"] = wrap_lon(df["longitude"].values)

    df["obs_clean"] = clean_array(df["observation"].values)
    df["prior_clean"] = clean_array(df["prior_ensemble_mean"].values)
    df["posterior_clean"] = clean_array(df["posterior_ensemble_mean"].values)
    df["qc_clean"] = clean_array(df["DART_quality_control"].values)

    df = df[df["type"].isin(OBS_TYPES_TO_PLOT)].copy()

    valid = (
        np.isfinite(df["obs_clean"])
        & np.isfinite(df["prior_clean"])
        & np.isfinite(df["posterior_clean"])
        & np.isfinite(df["qc_clean"])
    )

    return df[valid].copy()


def collect_obs_space_values(cycle_dirs):
    frames = []

    for cycle_dir in cycle_dirs:
        obs_file = Path(cycle_dir) / OBS_SEQ_NAME

        if not obs_file.exists():
            print(f"WARNING: missing {obs_file}")
            continue

        df = read_obs_file(obs_file)

        if df.empty:
            continue

        df["cycle_tag"] = Path(cycle_dir).name
        df["cycle_time"] = cycle_time_from_dir(cycle_dir)

        frames.append(df)

    if not frames:
        raise RuntimeError("No valid DART obs-space values found.")

    df = pd.concat(frames, ignore_index=True)

    tmin, tmax, _ = cycle_time_bounds(cycle_dirs)

    df = df[
        (df["time"] >= tmin)
        & (df["time"] <= tmax)
    ].copy()

    group_cols = [
        "type",
        "time",
        "vertical",
        "lon_plot",
        "latitude",
    ]

    df = (
        df
        .groupby(group_cols, as_index=False)
        .agg(
            observation=("obs_clean", "mean"),
            prior=("prior_clean", "mean"),
            posterior=("posterior_clean", "mean"),
            qc=("qc_clean", "first"),
            n_records=("obs_clean", "count"),
        )
    )

    print("\nValid DART obs-space rows:")
    print(df.groupby("type").size())

    print("\nQC values found:")
    print(np.sort(df["qc"].dropna().unique()))

    return df


# =============================================================================
# STATION SELECTION
# =============================================================================
def get_station(df):
    if TARGET_LON is not None and TARGET_LAT is not None:
        return float(wrap_lon(TARGET_LON)), float(TARGET_LAT)

    stations = (
        df.assign(
            station_lon=df["lon_plot"].round(4),
            station_lat=df["latitude"].round(4),
        )
        .groupby(["station_lon", "station_lat"])
        .size()
        .reset_index(name="count")
        .sort_values("count", ascending=False)
    )

    print("\nDetected stations:")
    print(stations)

    return float(stations.iloc[0]["station_lon"]), float(stations.iloc[0]["station_lat"])


def filter_station(df, lon0, lat0):
    dlon = df["lon_plot"].values - float(wrap_lon(lon0))
    dlat = df["latitude"].values - float(lat0)
    dist = np.sqrt(dlon**2 + dlat**2)

    return df[dist <= STATION_TOL_DEG].copy()


# =============================================================================
# QC COLOR SETUP
# =============================================================================
def make_qc_cmap_norm(df):
    qc_vals = np.sort(df["qc"].dropna().unique())

    if len(qc_vals) == 0:
        qc_vals = np.array([0.0])

    qc_vals = qc_vals.astype(int)

    cmap = plt.get_cmap("tab10", len(qc_vals))

    boundaries = np.arange(len(qc_vals) + 1) - 0.5
    norm = mpl.colors.BoundaryNorm(boundaries, cmap.N)

    qc_to_index = {qc: i for i, qc in enumerate(qc_vals)}

    return qc_vals, qc_to_index, cmap, norm


# =============================================================================
# PLOT
# =============================================================================
def plot_obs_space(df_station, lon0, lat0):
    variables = [v for v in OBS_TYPES_TO_PLOT if v in df_station["type"].unique()]
    levels = np.sort(df_station["vertical"].dropna().unique())[::-1]

    nlev = len(levels)
    nvar = len(variables)

    fig_width = max(4.3 * nvar, 12)
    fig_height = max(2.1 * nlev, 8)

    fig, axs = plt.subplots(
        nlev,
        nvar,
        figsize=(fig_width, fig_height),
        sharex=True,
        constrained_layout=False,
    )

    if nlev == 1 and nvar == 1:
        axs = np.array([[axs]])
    elif nlev == 1:
        axs = np.array([axs])
    elif nvar == 1:
        axs = np.array([[ax] for ax in axs])

    plt.subplots_adjust(
        left=0.07,
        right=0.84,
        bottom=0.07,
        top=0.91,
        hspace=0.34,
        wspace=0.30,
    )

    tmin, tmax, tick_times = cycle_time_bounds(CYCLE_DIRS)
    xpad = pd.Timedelta(minutes=X_PAD_MINUTES)

    qc_vals, qc_to_index, qc_cmap, qc_norm = make_qc_cmap_norm(df_station)

    scatter_ref = None

    for r, level in enumerate(levels):
        level_label = format_pressure_label(level)

        for c, var in enumerate(variables):
            ax = axs[r, c]

            d = df_station[
                (df_station["type"] == var)
                & (np.isclose(df_station["vertical"], level, atol=1e-6))
            ].copy()

            d = d.sort_values("time")

            if not d.empty:
                ax.plot(
                    d["time"],
                    d["prior"],
                    color="black",
                    linestyle="--",
                    linewidth=1.7,
                )

                ax.plot(
                    d["time"],
                    d["posterior"],
                    color="black",
                    linestyle="-",
                    linewidth=2.0,
                )

                qc_idx = d["qc"].astype(int).map(qc_to_index).to_numpy()

                scatter_ref = ax.scatter(
                    d["time"],
                    d["observation"],
                    c=qc_idx,
                    cmap=qc_cmap,
                    norm=qc_norm,
                    s=36,
                    edgecolors="k",
                    linewidths=0.35,
                    zorder=20,
                )

            if r == 0:
                ax.set_title(OBS_LABELS.get(var, var), fontsize=10, fontweight="bold")

            if c == 0:
                ax.set_ylabel(f"{level_label}\nValue")

            ax.text(
                0.02,
                0.92,
                level_label,
                transform=ax.transAxes,
                ha="left",
                va="top",
                fontsize=8,
                bbox=dict(facecolor="white", edgecolor="0.7", alpha=0.75, pad=1.5),
            )

            ax.grid(alpha=0.25)

            ax.set_xlim(tmin - xpad, tmax + xpad)
            ax.set_xticks(tick_times)
            ax.xaxis.set_major_formatter(mdates.DateFormatter("%H"))

    for ax in axs[-1, :]:
        ax.set_xlabel("UTC hour")

    # -------------------------------------------------------------------------
    # One shared QC colorbar, placed where the legend used to be
    # -------------------------------------------------------------------------
    cax = fig.add_axes([0.865, 0.77, 0.018, 0.18])

    sm = mpl.cm.ScalarMappable(cmap=qc_cmap, norm=qc_norm)
    sm.set_array([])

    cb = fig.colorbar(sm, cax=cax)
    cb.set_ticks(np.arange(len(qc_vals)))
    cb.set_ticklabels([str(q) for q in qc_vals])
    cb.set_label("DART QC", fontsize=9)
    cb.ax.tick_params(labelsize=8)

    # Line legend below QC colorbar
    legend_handles = [
        Line2D(
            [0], [0],
            color="black",
            linestyle="--",
            linewidth=1.7,
            label="DART prior H(x)",
        ),
        Line2D(
            [0], [0],
            color="black",
            linestyle="-",
            linewidth=2.0,
            label="DART posterior H(x)",
        ),
        Line2D(
            [0], [0],
            color="white",
            marker="o",
            markeredgecolor="k",
            linestyle="none",
            label="Observation",
        ),
    ]

    fig.legend(
        handles=legend_handles,
        loc="upper left",
        bbox_to_anchor=(0.855, 0.73),
        fontsize=8,
        frameon=True,
        title="Line type",
    )

    fig.suptitle(
        (
            "DART Observation-Space Prior/Posterior H(x) with Observations Colored by QC\n"
            f"station lon={lon0:.4f}, lat={lat0:.4f}"
        ),
        fontsize=14,
    )

    outfig = FIGDIR / (
        f"dart_obs_space_QC_{CYCLE_DIRS[0].parts[-3]}.jpeg"
    )

    if SAVE_FIG:
        plt.savefig(outfig, dpi=DPI, bbox_inches="tight")
        print(f"Saved {outfig}")

    if SHOW_FIG:
        plt.show()
    else:
        plt.close(fig)


# =============================================================================
# MAIN
# =============================================================================
df = collect_obs_space_values(CYCLE_DIRS)

lon0, lat0 = get_station(df)

df_station = filter_station(df, lon0, lat0)

print("\nPressure levels used:")
print([format_pressure_label(p) for p in np.sort(df_station["vertical"].unique())[::-1]])

plot_obs_space(df_station, lon0, lat0)
```

```python
#!/usr/bin/env python3
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
from pathlib import Path
from matplotlib.ticker import MaxNLocator, ScalarFormatter
from mpl_toolkits.basemap import Basemap

# Import the pydartdiags module for reading obs_seq.final files
import pydartdiags.obs_sequence.obs_sequence as obsq

# =============================================================================
# USER OPTIONS
# =============================================================================
# CYCLE_DIR   = "/gpfs/home/sa24m/scratch/tqprof/run2/osse_out/dart_cycle/output/2015071412/"
CYCLE_DIR   = "/gpfs/home/sa24m/stephen_asare/runs/finished_all_obs_enkf_inf_0/2015071412/"
TIME_INDEX  = 0

# VERTICAL LEVEL SETTINGS
PLOT_LOWEST_LEVEL = True    # Set to True to plot k=0 (lowest model level)
ZAGL_M            = 3000.0  # Ignored if PLOT_LOWEST_LEVEL is True

DOMS        = ["d01", "d02"]             # row 0 = d01, row 1 = d02
VARS_2COL   = ["THM", "QVAPOR"]          # col 0..1 in each row

INC_SOURCE  = "file"                     # "file" (analysis_increment) or "diff" (post-pre)

# winds (background wind at same level)
PLOT_WIND     = True
WIND_U        = "U"
WIND_V        = "V"
WIND_SKIP     = 7
QUIVER_SCALE  = 350
QUIVER_WIDTH  = 0.002

# plotting style
BG_LEVELS    = 21
INC_LEVELS   = 17
INC_ALPHA    = 0.55
INC_THRESH   = None   # e.g. 1e-4, or None
CMAP_BG      = "Greens"
CMAP_INC     = "bwr"

# PYDARTDIAGS OBS OVERLAY OPTIONS
PLOT_OBS           = True
OBS_SEQ_FILE       = "/gpfs/home/sa24m/scratch/tqprof/run2/osse_out/dart_cycle/output/2015071412/obs_seq.final"
OBS_TIME_START     = "2015-07-14 10:30:00" 
OBS_TIME_END       = "2015-07-14 13:30:00" 
OBS_TYPES          = ["LAND_SFC_TEMPERATURE"]
OBS_VALUE_COL      = "observation"         
OBS_MARKER         = "o"
OBS_SIZE           = 35
OBS_ALPHA          = 0.9
OBS_CMAP           = "bwr"

# =============================================================================
# WRF BASEMAP HELPER
# =============================================================================
def setup_basemap_from_file(nc_path):
    ds = xr.open_dataset(nc_path)
    lat = ds['XLAT'].values[0] if 'Time' in ds['XLAT'].dims else ds['XLAT'].values
    lon = ds['XLONG'].values[0] if 'Time' in ds['XLONG'].dims else ds['XLONG'].values
    
    truelat1 = getattr(ds, 'TRUELAT1', lat.mean())
    truelat2 = getattr(ds, 'TRUELAT2', lat.mean())
    stand_lon = getattr(ds, 'STAND_LON', lon.mean())
    
    m = Basemap(
        projection='lcc', resolution='i',
        llcrnrlon=lon[0, 0], llcrnrlat=lat[0, 0],
        urcrnrlon=lon[-1, -1], urcrnrlat=lat[-1, -1],
        lat_1=truelat1, lat_2=truelat2, lon_0=stand_lon
    )
    x, y = m(lon, lat)
    ds.close()
    return x, y, m

# =============================================================================
# Helpers
# =============================================================================
def safe_levels(vmin, vmax, n, pad_frac=1e-6, pad_abs=1e-12):
    vmin = float(vmin); vmax = float(vmax)
    if not np.isfinite(vmin) or not np.isfinite(vmax):
        return np.linspace(0.0, 1.0, n)
    if vmax <= vmin:
        pad = max(pad_abs, pad_frac * (abs(vmin) + 1.0))
        vmin -= pad
        vmax += pad
    return np.linspace(vmin, vmax, n)

def _destagger_to_mass(ds, var, time_index=0):
    da = ds[var].isel(Time=time_index) if "Time" in ds[var].dims else ds[var]
    if var == "U":
        return 0.5 * (da.isel(west_east_stag=slice(0, -1)) + da.isel(west_east_stag=slice(1, None))).rename({"west_east_stag":"west_east"})
    if var == "V":
        return 0.5 * (da.isel(south_north_stag=slice(0, -1)) + da.isel(south_north_stag=slice(1, None))).rename({"south_north_stag":"south_north"})
    if var == "W":
        return 0.5 * (da.isel(bottom_top_stag=slice(0, -1)) + da.isel(bottom_top_stag=slice(1, None))).rename({"bottom_top_stag":"bottom_top"})
    return da

def _k_at_zagl(ds_ref, zagl_m, time_index=0):
    Zstag = (ds_ref["PH"].isel(Time=time_index).values + ds_ref["PHB"].values) / 9.81
    Zmass = 0.5 * (Zstag[:-1, :, :] + Zstag[1:, :, :])
    HGT   = ds_ref["HGT"].values
    ZAGL  = Zmass - HGT[None, :, :]
    return np.argmin(np.abs(ZAGL - zagl_m), axis=0)

def field_at_zagl(ds_field, ds_ref, var, zagl_m, time_index=0, lowest_level=False):
    # Pass-through for purely 2D surface fields
    if ds_field[var].ndim == 3 and ("bottom_top" not in ds_field[var].dims) and ("bottom_top_stag" not in ds_field[var].dims):
        return ds_field[var].isel(Time=time_index).values

    if "bottom_top" in ds_field[var].dims:
        v3d = ds_field[var].isel(Time=time_index).values
    else:
        da = _destagger_to_mass(ds_field, var, time_index=time_index)
        if "bottom_top" not in da.dims:
            return da.values
        v3d = da.values

    # If requested, just grab k=0 and skip the height interpolation
    if lowest_level:
        return v3d[0, :, :]

    k = _k_at_zagl(ds_ref, zagl_m=zagl_m, time_index=time_index)
    j, i = np.indices(k.shape)
    return v3d[k, j, i]

# =============================================================================
# Load fields for both domains
# =============================================================================
cdir = Path(CYCLE_DIR)
DATA = {}   # (dom,var) -> dict(bg=..., inc=...)
GEO  = {}   # dom -> dict(x,y,m,lon,lat,u,v)

for dom in DOMS:
    post_file = cdir / f"postassim_mean_{dom}.nc"
    pre_file  = cdir / f"preassim_mean_{dom}.nc"
    inc_file  = cdir / f"analysis_increment_{dom}.nc"

    if not post_file.exists(): raise FileNotFoundError(f"Missing: {post_file}")
    if INC_SOURCE == "diff" and (not pre_file.exists()): raise FileNotFoundError(f"Missing: {pre_file}")
    if INC_SOURCE == "file" and (not inc_file.exists()): raise FileNotFoundError(f"Missing: {inc_file}")

    x, y, m = setup_basemap_from_file(str(post_file))

    ds_post = xr.open_dataset(post_file)
    ds_pre  = xr.open_dataset(pre_file) if (INC_SOURCE == "diff") else None
    ds_inc  = xr.open_dataset(inc_file) if (INC_SOURCE == "file") else None

    lon2d = ds_post["XLONG"].values
    lat2d = ds_post["XLAT"].values

    if PLOT_WIND:
        u2d = field_at_zagl(ds_post, ds_post, WIND_U, ZAGL_M, time_index=TIME_INDEX, lowest_level=PLOT_LOWEST_LEVEL)
        v2d = field_at_zagl(ds_post, ds_post, WIND_V, ZAGL_M, time_index=TIME_INDEX, lowest_level=PLOT_LOWEST_LEVEL)
    else:
        u2d = v2d = None

    GEO[dom] = dict(x=x, y=y, m=m, lon=lon2d, lat=lat2d, u=u2d, v=v2d, post_file=str(post_file))

    for var in VARS_2COL:
        bg2d = field_at_zagl(ds_post, ds_post, var, ZAGL_M, time_index=TIME_INDEX, lowest_level=PLOT_LOWEST_LEVEL)

        if INC_SOURCE == "file":
            inc2d = field_at_zagl(ds_inc, ds_post, var, ZAGL_M, time_index=TIME_INDEX, lowest_level=PLOT_LOWEST_LEVEL)
        else:
            pre2d = field_at_zagl(ds_pre, ds_post, var, ZAGL_M, time_index=TIME_INDEX, lowest_level=PLOT_LOWEST_LEVEL)
            inc2d = bg2d - pre2d

        if INC_THRESH is not None:
            inc2d = np.where(np.abs(inc2d) >= INC_THRESH, inc2d, np.nan)

        DATA[(dom, var)] = dict(bg=bg2d, inc=inc2d)

    if ds_inc is not None: ds_inc.close()
    if ds_pre is not None: ds_pre.close()
    ds_post.close()

# =============================================================================
# Shared levels per COLUMN across d01/d02
# =============================================================================
LEVELS = {}
for var in VARS_2COL:
    bmins, bmaxs = [], []
    for dom in DOMS:
        b = DATA[(dom, var)]["bg"]
        bmins.append(np.nanmin(b))
        bmaxs.append(np.nanmax(b))
    levels_bg = safe_levels(np.nanmin(bmins), np.nanmax(bmaxs), BG_LEVELS)

    absmaxs = []
    for dom in DOMS:
        inc = DATA[(dom, var)]["inc"]
        absmaxs.append(np.nanmax(np.abs(inc)))
    inc_absmax = float(np.nanmax(absmaxs))
    if not np.isfinite(inc_absmax) or inc_absmax == 0.0:
        inc_absmax = 1e-12
    levels_inc = np.linspace(-inc_absmax, inc_absmax, INC_LEVELS)

    LEVELS[var] = dict(levels_bg=levels_bg, levels_inc=levels_inc)

# =============================================================================
# Process pydartdiags Observations (Done ONCE before plotting)
# =============================================================================
lon_obs = lat_obs = obs_val = None

if PLOT_OBS:
    try:
        obs_seq = obsq.ObsSequence(OBS_SEQ_FILE)
        df = obs_seq.df.copy()
        
        if df.empty:
            print(f"WARNING: {Path(OBS_SEQ_FILE).name} loaded, but is completely empty.")
        else:
            df["time"] = pd.to_datetime(df["time"])
            
            t0 = pd.Timestamp(OBS_TIME_START)
            t1 = pd.Timestamp(OBS_TIME_END)
            
            df_filtered = df[
                (df["time"] >= t0) &
                (df["time"] <= t1) &
                (df["type"].isin(OBS_TYPES))
            ].copy()
            
            df_filtered["lon_plot"] = ((df_filtered["longitude"] + 180) % 360) - 180
            
            lon_obs = df_filtered["lon_plot"].values
            lat_obs = df_filtered["latitude"].values
            obs_val = df_filtered[OBS_VALUE_COL].values
            
            print(f"SUCCESS: Loaded {len(df_filtered)} observations from {Path(OBS_SEQ_FILE).name} for overlay.")
    except Exception as e:
        print(f"FAILED: Could not process obs_seq file. Error: {e}")

# =============================================================================
# Plot 2 rows x 2 columns + OBS overlay
# =============================================================================
fig, axs = plt.subplots(2, 2, figsize=(13.5, 8.5), constrained_layout=False)
plt.subplots_adjust(left=0.06, right=0.97, bottom=0.06, top=0.92, wspace=0.26, hspace=0.14)

for r, dom in enumerate(DOMS):
    g = GEO[dom]
    m = g["m"]
    x = g["x"]; y = g["y"]

    for c, var in enumerate(VARS_2COL):
        ax = axs[r, c]

        m.drawcoastlines(ax=ax, linewidth=0.5)
        m.drawstates(ax=ax, linewidth=0.3)
        m.drawcountries(ax=ax, linewidth=0.3)

        bg2d  = DATA[(dom, var)]["bg"]
        inc2d = DATA[(dom, var)]["inc"]
        lev_bg  = LEVELS[var]["levels_bg"]

        cs_bg  = m.contourf(x, y, bg2d,  levels=lev_bg,  extend="both", ax=ax, cmap=CMAP_BG)
        
        pos_inc = np.where(inc2d > 0, inc2d, np.nan)
        neg_inc = np.where(inc2d < 0, inc2d, np.nan)
        inc_max = float(np.nanmax(inc2d))
        inc_min = float(np.nanmin(inc2d))

        # Safe contour plotting that prevents the zero-increment crash
        if inc_max > 1e-10:
            pos_levels = np.linspace(0, inc_max, INC_LEVELS)
            m.contour(x, y, pos_inc, levels=pos_levels, ax=ax, cmap="Wistia", alpha=INC_ALPHA)
        
        if inc_min < -1e-10:
            neg_levels = np.linspace(inc_min, 0, INC_LEVELS)
            m.contour(x, y, neg_inc, levels=neg_levels, ax=ax, cmap="cool", alpha=INC_ALPHA)

        if PLOT_WIND:
            m.quiver(
                g["lon"][::WIND_SKIP, ::WIND_SKIP],
                g["lat"][::WIND_SKIP, ::WIND_SKIP],
                g["u"][::WIND_SKIP, ::WIND_SKIP],
                g["v"][::WIND_SKIP, ::WIND_SKIP],
                latlon=True,
                ax=ax,
                scale=QUIVER_SCALE,
                width=QUIVER_WIDTH,
                color="k",
                zorder=10,
            )

        if PLOT_OBS and (lon_obs is not None) and (lon_obs.size > 0):
            m.scatter(
                lon_obs, lat_obs,
                c=obs_val,
                cmap=OBS_CMAP,
                s=OBS_SIZE,
                marker=OBS_MARKER,
                alpha=OBS_ALPHA,
                edgecolor="k",
                linewidth=0.4,
                latlon=True,
                ax=ax,
                zorder=30,
            )

        # Dynamic Title Logic based on PLOT_LOWEST_LEVEL
        if PLOT_LOWEST_LEVEL:
            lvl_str = "Lowest Model Level (k=0)"
        else:
            lvl_str = f"{ZAGL_M:.0f} m AGL"
            
        ax.set_title(f"{dom}: {var} @ {lvl_str}", fontsize=11)

        cb_bg = fig.colorbar(cs_bg, ax=ax, location="right", pad=0.02, fraction=0.046)
        cb_bg.set_label(f"{var}", fontsize=8)
        cb_bg.ax.yaxis.set_major_locator(MaxNLocator(4))
        cb_bg.ax.yaxis.set_major_formatter(ScalarFormatter(useOffset=False))

fig.suptitle(
    f"Impact of {OBS_TYPES[0]} | cycle={Path(CYCLE_DIR).name}",
    fontsize=12
)
plt.show()
```
